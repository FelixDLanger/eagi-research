# EAGI Pipeline — EA Governance Index (Core Integrity, Study 1)

**Purpose.** Construct the disclosure-based EA Governance Index from SEC 10-K filings, add the board TechCom indicator from DEF 14A, merge with the Core Integrity firm-year panel, and estimate H1/H2 with two-way fixed effects and Driscoll–Kraay standard errors, plus Oster bounds and a gated Callaway–Sant'Anna event study.

**Run order.** Upload → Runtime → Run all. Discovery v5: ccaof-tree-first across csv / parquet / feather / pickle / xlsx / dta; header → value-signature → key-token column resolution; split CI file auto-merged; ticker panels rescued via the SEC map. If the panel cannot be resolved, the run continues in *prefetch-only mode* (inventory printed, firm universe extracted, full text layer fetched and banked). Every cell is cascade-proof: an upstream failure makes downstream cells print a skip banner instead of NameError noise, and the **RUN DIAGNOSTIC** cell at the end always executes and names the root failing stage — paste that block together with the traceback it points to.

**Optional second classifier.** To enable the LLM cross-check of TechCom evidence, add a Colab secret named `ANTHROPIC_API_KEY` (left sidebar → key icon → Add new secret → toggle *Notebook access* on). An API key comes from the Claude Developer Platform console and is billed separately from any Claude.ai subscription. Without the key the notebook runs identically using the deterministic rule classifier only; with it, labels are written once to `llm_techcom_labels.csv` and reused verbatim on every later run.

````python
# ============ 1 · SETUP — mount, walk, discover; prefetch-only fallback ============
import sys, subprocess
def pipq(*pkgs):
    """Plain install - the form that worked for eleven versions. Two attempted "improvements"
    each broke it: --no-deps starved linearmodels of formulaic, and a numpy/pandas constraints
    file made the dependency set unresolvable so nothing installed at all. Neither was needed:
    the 'everything unreadable' run was never traced to pip, it was only hypothesised.

    What IS kept, because it has evidence behind it: install only what is missing, check the
    module that is ACTUALLY IMPORTED (linearmodels imports fine while linearmodels.panel does
    not), and SHOW pip's error instead of failing silently three cells later.
    """
    import importlib
    TARGET = {"beautifulsoup4": "bs4", "linearmodels": "linearmodels.panel",
              "pyarrow": "pyarrow", "openpyxl": "openpyxl", "lxml": "lxml",
              "tqdm": "tqdm", "differences": "differences"}
    def _ok(pk):
        try:
            importlib.invalidate_caches(); importlib.import_module(TARGET.get(pk, pk)); return True
        except Exception:
            return False
    need = [pk for pk in pkgs if not _ok(pk)]
    if not need:
        return
    r = subprocess.run([sys.executable, "-m", "pip", "install", "-q", *need],
                       capture_output=True, text=True)
    still = [pk for pk in need if not _ok(pk)]
    if still:
        print(f"  pip could not make {still} importable (exit {r.returncode}).")
        tail = ((r.stderr or "") + (r.stdout or "")).strip().splitlines()[-4:]
        for ln in tail:
            print("    " + ln[:150])
        print("    → Runtime > Restart session, then Run all. A fresh kernel picks up packages")
        print("      installed in this session; the import cache in a live kernel often will not.")

import os, re, json, time, types, warnings, difflib
import numpy as np, pandas as pd
from tqdm.auto import tqdm
warnings.filterwarnings("ignore")
np.random.seed(42)

def G(n):
    return globals().get(n)

# ---------------------------------------------------------------------------
# TWO-WAY FIXED-EFFECTS OLS - implemented directly, no external estimator package.
# linearmodels repeatedly failed to install in this environment, and every patch around the
# installer was a patch around a dependency we do not need. Everything used from PanelOLS is
# elementary: within transformation, clustered sandwich, Driscoll-Kraay. numpy and pandas are
# preinstalled in Colab, so this removes the single most fragile dependency in the pipeline.
class FEResult:
    def __init__(self, names, beta, cov, nobs, r2w, dfree):
        self.params = pd.Series(beta, index=names)
        self.std_errors = pd.Series(np.sqrt(np.maximum(np.diag(cov), 0)), index=names)
        self.bse = self.std_errors
        with np.errstate(divide="ignore", invalid="ignore"):
            self.tstats = self.params / self.std_errors
        self.tvalues = self.tstats
        from math import erf, sqrt as _sq
        self.pvalues = pd.Series(
            [2 * (1 - 0.5 * (1 + erf(abs(t) / _sq(2)))) if t == t else np.nan for t in self.tstats],
            index=names)
        self.nobs = int(nobs); self.rsquared_within = float(r2w); self.df_resid = int(dfree)
        tab = pd.DataFrame({"coef": self.params.round(6), "std_err": self.std_errors.round(6),
                            "t": self.tstats.round(3), "P>|t|": self.pvalues.round(4)})
        self.summary = types.SimpleNamespace(tables=[None, tab.to_string()])

def _demean2(frame, cols, ent, tim, iters=60, tol=1e-11):
    """Alternating projections: on an unbalanced panel a single firm-then-year pass is NOT the
    two-way within transformation, so iterate to convergence."""
    X = frame[cols].astype(float).copy()
    for _ in range(iters):
        prev = X.to_numpy(copy=True)
        X = X - X.groupby(ent, observed=True).transform("mean")
        X = X - X.groupby(tim, observed=True).transform("mean")
        if np.nanmax(np.abs(X.to_numpy() - prev)) < tol:
            break
    return X

def fe_ols(dep, xs, data, se="clustered", bandwidth=None):
    """Two-way FE regression. data must carry a (entity, time) MultiIndex.
    se='clustered' clusters on entity; se='kernel' is Driscoll-Kraay (Bartlett)."""
    d_ = data.dropna(subset=[dep] + list(xs)).copy()
    ent = pd.Series(d_.index.get_level_values(0), index=d_.index)
    tim = pd.Series(d_.index.get_level_values(1), index=d_.index)
    Z = _demean2(d_, [dep] + list(xs), ent, tim)
    y = Z[dep].to_numpy(float); X = Z[list(xs)].to_numpy(float)
    n, k = X.shape
    XtX_inv = np.linalg.pinv(X.T @ X)
    beta = XtX_inv @ X.T @ y
    e = y - X @ beta
    n_ent = int(ent.nunique()); n_tim = int(tim.nunique())
    dfree = max(1, n - k - n_ent - n_tim + 1)
    if se == "kernel":
        u = X * e[:, None]
        h = pd.DataFrame(u, index=tim.to_numpy()).groupby(level=0).sum().to_numpy()  # cross-sec sum
        T = h.shape[0]
        m = bandwidth if bandwidth is not None else int(np.floor(4 * (T / 100.0) ** (2.0 / 9.0)))
        m = max(0, min(m, T - 1))
        S = h.T @ h
        for j in range(1, m + 1):
            w = 1.0 - j / (m + 1.0)
            G_ = h[j:].T @ h[:-j]
            S = S + w * (G_ + G_.T)
        cov = XtX_inv @ S @ XtX_inv
    else:
        meat = np.zeros((k, k))
        codes = ent.to_numpy()
        for g_ in pd.unique(codes):
            msk = codes == g_
            sg = X[msk].T @ e[msk]
            meat += np.outer(sg, sg)
        cov = XtX_inv @ meat @ XtX_inv
        nc = max(1, len(pd.unique(codes)))
        cov *= (nc / max(1, nc - 1)) * ((n - 1) / max(1, dfree))
    tss = float(((y - y.mean()) ** 2).sum())
    r2w = 1 - float(e @ e) / tss if tss > 0 else np.nan
    return FEResult(list(xs), beta, cov, n, r2w, dfree)


from google.colab import drive

def _mount_drive(path="/content/drive", tries=4):
    """Colab's Drive mount fails transiently with 'credential propagation was unsuccessful'.
    That is an ENVIRONMENT fault, not a data fault, and it is usually cleared by a retry.
    Escalate: plain retry, then unmount-and-remount, then a forced remount. Fall through
    immediately if the mount point is already usable, so a rerun costs nothing."""
    if os.path.isdir(os.path.join(path, "MyDrive")):
        print("  Drive already mounted - reusing")
        return True
    last = None
    for a in range(tries):
        try:
            drive.mount(path, force_remount=(a >= 2))
            if os.path.isdir(os.path.join(path, "MyDrive")):
                print(f"  Drive mounted on attempt {a + 1}")
                return True
        except Exception as e:
            last = e
            print(f"  mount attempt {a + 1}/{tries} failed: {type(e).__name__}: {str(e)[:90]}")
            if a == 1:
                try:
                    drive.flush_and_unmount()
                    print("  unmounted stale mount point; retrying")
                except Exception:
                    pass
        time.sleep(3 * (a + 1))
    print("\n  DRIVE MOUNT FAILED after", tries, "attempts.")
    print("  This is a Colab environment fault. Nothing is wrong with the data or this notebook.")
    print("  Causes, in order of frequency:")
    print("    1. transient credential propagation - Runtime > Restart session, then Run all")
    print("    2. third-party cookies blocked in the browser; Drive auth requires them")
    print("    3. an incognito or embedded window, which cannot complete the auth handshake")
    print("    4. an already-authorised session in another tab holding the credential")
    return False

if not _mount_drive():
    raise RuntimeError("Google Drive could not be mounted - see the diagnosis above. "
                       "All source data lives on Drive, so the run cannot proceed. "
                       "This is an environment fault, not a pipeline fault.")

NEED = ["cik", "year", "ci", "roa", "q", "sga", "size", "lev", "rnd", "capex"]
ALT = {
  "cik": ["cik", "cik_code", "cik_int", "sec_cik"],
  "year": ["fyear", "year", "fiscal_year", "fy", "fiscal_yr", "yr", "report_year"],
  "ci": ["ci_score", "ci", "cc_score", "adherence", "adherence_score", "clean_core_score", "core_integrity", "ccaof", "ccaof_score", "ci_composite", "ci_z", "cc_adherence", "adh_score"],
  "roa": ["roa", "return_on_assets", "roa_w", "ni_at", "return_assets"],
  "q": ["tobins_q", "tobinq", "tobin_q", "tobinsq", "tq", "q_w", "q"],
  "sga": ["sga_sales", "sga", "sga_ratio", "sga_intensity", "sga_rev", "sga_at", "sgna", "xsga_sale"],
  "size": ["log_at", "ln_assets", "log_assets", "lnassets", "at_log", "ln_at", "lnta", "log_ta", "size"],
  "lev": ["leverage", "lev", "debt_ratio", "lev_w", "tl_at", "debt_at", "td_at", "leverage_w"],
  "rnd": ["rnd_int", "rd_intensity", "rnd", "xrd_at", "xrd_sale", "rd_sale", "rnd_at", "rd_at", "r_d"],
  "capex": ["capex_int", "capex", "capx_at", "capx_sale", "capex_at_ratio", "capx", "capex_ratio"],
}
TICK_ALIASES = ["ticker", "tic", "symbol"]
EXTS = (".csv", ".parquet", ".pq", ".feather", ".pkl", ".pickle", ".xlsx", ".dta")

def resolve(cols):
    low = {str(c).lower().strip(): c for c in cols}
    m = {}
    for k, cands in ALT.items():
        hit = next((low[c] for c in cands if c in low), None)
        if hit is not None: m[k] = hit
    return m

def _csv_sep(path):
    """SimFin and many EU exports are ';'-delimited — sniff rather than assume."""
    try:
        with open(path, "r", errors="ignore") as f: head = f.readline()
        return ";" if head.count(";") > head.count(",") else ","
    except Exception:
        return ","

READ_ERRORS = []          # (path, error) - a swallowed exception is a lost diagnosis

def read_head(path, n=400, _tries=3):
    """Drive I/O in Colab fails transiently under load, so retry before giving up - and ALWAYS
    record the reason. Returning a bare None turns one systemic error into 'everything unreadable'."""
    ext = os.path.splitext(path)[1].lower()
    last = None
    for _a in range(_tries):
        try:
            return _read_head_once(path, n, ext)
        except Exception as e:
            last = e
            time.sleep(0.4 * (_a + 1))
    READ_ERRORS.append((path, f"{type(last).__name__}: {str(last)[:120]}"))
    return None, None

def _read_head_once(path, n, ext):
    if True:
        if ext == ".csv":
            df = pd.read_csv(path, nrows=n, low_memory=False, sep=_csv_sep(path)); return list(df.columns), df
        if ext in (".parquet", ".pq"):
            import pyarrow.parquet as pq
            pf = pq.ParquetFile(path)
            cols = [c for c in pf.schema_arrow.names]
            try: smp = pf.read_row_group(0).to_pandas().head(n)
            except Exception: smp = pd.read_parquet(path).head(n)
            return cols, smp
        if ext == ".feather":
            df = pd.read_feather(path); return list(df.columns), df.head(n)
        if ext in (".pkl", ".pickle"):
            if os.path.getsize(path) > 300_000_000: return None, None
            df = pd.read_pickle(path)
            if not isinstance(df, pd.DataFrame): return None, None
            return list(df.columns), df.head(n)
        if ext == ".xlsx":
            df = pd.read_excel(path, nrows=n); return list(df.columns), df
        if ext == ".dta":
            it = pd.read_stata(path, chunksize=n); df = next(it); return list(df.columns), df
    return None, None

def read_full(path):
    ext = os.path.splitext(path)[1].lower()
    if ext == ".csv": return pd.read_csv(path, low_memory=False, sep=_csv_sep(path))
    if ext in (".parquet", ".pq"): return pd.read_parquet(path)
    if ext == ".feather": return pd.read_feather(path)
    if ext in (".pkl", ".pickle"): return pd.read_pickle(path)
    if ext == ".xlsx": return pd.read_excel(path)
    if ext == ".dta": return pd.read_stata(path)
    raise ValueError(ext)

def read_cache(path, required=(), label=""):
    """Read a cached CSV only if its SCHEMA still matches what the current code needs.

    A cache written by an earlier version of this notebook can be missing columns that later
    code depends on, which surfaces as an opaque KeyError several cells downstream. Validating
    the schema on read turns that into an automatic, silent recompute - the same self-healing
    contract the rest of the pipeline follows."""
    if not os.path.exists(path):
        return None
    try:
        d_ = pd.read_csv(path)
    except Exception as e:
        print(f"  cache unreadable ({label or os.path.basename(path)}): {type(e).__name__} - recomputing")
        return None
    miss = [c for c in required if c not in d_.columns]
    if miss:
        print(f"  cached {label or os.path.basename(path)} predates current code (missing {miss}) - recomputing")
        return None
    return d_

def read_col(path, col):
    ext = os.path.splitext(path)[1].lower()
    try:
        if ext == ".csv": return pd.read_csv(path, usecols=[col], low_memory=False, sep=_csv_sep(path))[col]
        if ext in (".parquet", ".pq"): return pd.read_parquet(path, columns=[col])[col]
        return read_full(path)[col]
    except Exception:
        try: return read_full(path)[col]
        except Exception: return None

print("Scanning Drive … [discovery v11 · csv/parquet/feather/pkl/xlsx/dta · ccaof-first · content inference · CIK registry validation · derivation layer · prefetch fallback · cascade-proof]")
data_files, cache_dirs, edgar_dirs = [], [], []
for root, dirs, files in os.walk(DRIVE_ROOT):
    dirs[:] = [dd for dd in dirs if not dd.startswith(".")]
    base = os.path.basename(root).lower()
    if base == "eagi_cache": cache_dirs.append(root)
    if PROJECT_HINT in root.lower() and base == "edgar": edgar_dirs.append((root, len(files)))
    if any(x in root.lower() for x in ("eagi_out", "eagi_cache")):
        continue                                    # our own artefacts: never sources, and scanning
    for f in files:                                 # them multiplies Drive I/O for nothing
        if f.lower().endswith(EXTS): data_files.append(os.path.join(root, f))
in_project = [p for p in data_files if PROJECT_HINT in p.lower()]
print(f"  {len(data_files)} data files · {len(in_project)} in '{PROJECT_HINT}' tree · {len(cache_dirs)} eagi_cache dir(s)")
for ed, nf in edgar_dirs:
    print(f"  existing project edgar folder: {ed} ({nf} files) — not auto-wired; share a sample filename to reuse it")

# ---- SEC contact identity: Colab secrets → Drive cache → prompt ----
try:
    from google.colab import userdata
    for _k in ("SEC_EMAIL", "SEC_CONTACT_EMAIL"):
        if not USER_AGENT:
            try:
                _v = userdata.get(_k)
                if _v and "@" in _v: USER_AGENT = _v.strip(); print(f"  SEC identity from Colab secret {_k}")
            except Exception: pass
except Exception: pass
ua_seed = next((os.path.join(cdir, "user_agent.txt") for cdir in cache_dirs if os.path.exists(os.path.join(cdir, "user_agent.txt"))), os.path.join(DRIVE_ROOT, ".eagi_ua.txt"))
if not USER_AGENT and os.path.exists(ua_seed): USER_AGENT = open(ua_seed).read().strip()
while not USER_AGENT or "@" not in USER_AGENT:
    USER_AGENT = input("SEC requires a contact identity. Enter 'Name email@domain': ").strip()
if " " not in USER_AGENT:
    USER_AGENT = "Research Contact " + USER_AGENT   # SEC prefers 'Name email'
open(os.path.join(DRIVE_ROOT, ".eagi_ua.txt"), "w").write(USER_AGENT)

def tick2cik():
    p = os.path.join(DRIVE_ROOT, ".eagi_company_tickers.json")
    if not os.path.exists(p):
        import requests
        r = requests.get("https://www.sec.gov/files/company_tickers.json", headers={"User-Agent": USER_AGENT}, timeout=30)
        r.raise_for_status(); open(p, "w").write(r.text)
    j = json.load(open(p))
    return {v["ticker"].upper(): int(v["cik_str"]) for v in j.values()}

def cik_share(vals):
    """Fraction of candidate identifiers that are real SEC CIKs (registry match)."""
    try:
        u = set(tick2cik().values())
        v = pd.Series(list(vals)).dropna().astype(int)
        return float(v.isin(list(u)).mean()) if len(v) else 0.0
    except Exception:
        return None

def name_score(p):
    n = os.path.basename(p).lower()
    s = 0
    if "panel" in n: s += 3
    if any(k in n for k in ("ccaof", "core_integrity", "coreintegrity", "adher", "clean")): s += 2
    if re.search(r"\bci\b|ci_|_ci", n): s += 1
    if "eagi" in n: s -= 3
    return s

def profile(sample):
    info = {}
    for c in sample.columns:
        try:
            s = sample[c].dropna()
            if len(s) < 20: continue
            if pd.api.types.is_bool_dtype(s): continue
            if pd.api.types.is_numeric_dtype(s):
                s = pd.to_numeric(s, errors="coerce").dropna()
                if len(s) < 20: continue
                info[c] = dict(kind="num", med=float(s.median()), p05=float(s.quantile(.05)), p95=float(s.quantile(.95)),
                               sd=float(s.std() or 0), nun=int(s.nunique()), ints=bool((s % 1 == 0).mean() > .95))
            else:
                u = s.astype(str).str.strip()
                info[c] = dict(kind="obj", nun=int(u.nunique()), tick=bool(u.str.fullmatch(r"[A-Za-z][A-Za-z0-9\.\-]{0,5}").mean() > .8))
        except Exception:
            continue
    return info

def content_guess(info, missing):
    prop, cands, used = {}, {k: [] for k in missing}, set()
    def fits(k, st):
        if st["kind"] != "num": return False
        m, lo, hi = st["med"], st["p05"], st["p95"]
        return {
          "year":  st["ints"] and 1990 <= m <= 2030 and 3 <= st["nun"] <= 45,
          "cik":   st["ints"] and lo >= 1000 and hi <= 2_000_000 and st["nun"] > 50,
          "size":  (not st["ints"]) and 3 <= m <= 15 and 0.4 <= st["sd"] <= 4,
          "lev":   -0.02 <= lo and hi <= 2.5 and 0.02 <= m <= 0.9,
          "roa":   -0.6 <= lo and hi <= 0.6 and -0.15 <= m <= 0.25,
          "q":     lo >= -0.01 and 0.4 <= m <= 6 and hi <= 30,
          "sga":   lo >= -0.01 and hi <= 3 and 0.03 <= m <= 0.8,
          "rnd":   lo >= -0.01 and hi <= 0.8 and m <= 0.2,
          "capex": lo >= -0.01 and hi <= 0.8 and m <= 0.25,
          "ci":    (abs(m) < 0.6 and 0.4 <= st["sd"] <= 2.5) or (lo >= -0.01 and hi <= 1.05) or (lo >= -0.5 and 50 <= hi <= 100.5),
        }.get(k, False)
    KEYTOK = {"cik": {"cik"}, "year": {"year", "fy", "fyear", "fiscal"}, "ci": {"ci", "adher", "ccaof", "clean", "core", "score", "adh"},
              "roa": {"roa", "ni", "margin", "return"}, "q": {"q", "tobin"}, "sga": {"sga", "sgna", "sganda"},
              "size": {"ln", "log", "size", "lnta"}, "lev": {"lev", "leverage", "debt", "liab"},
              "rnd": {"rnd", "rd", "xrd", "research"}, "capex": {"capex", "capx"}}
    def match_name(k, col):
        cl_ = str(col).lower()
        toks = set(re.split(r"[^a-z0-9]+", cl_))
        kt = KEYTOK[k]
        return bool(toks & kt) or any(t in cl_ for t in kt if len(t) >= 3)
    order = [k for k in ["year", "cik", "size", "q", "lev", "sga", "roa", "capex", "rnd", "ci"] if k in missing]
    for k in order:
        cl = [c for c, st in info.items() if c not in used and fits(k, st)]
        cands[k] = cl
        pick = None
        strong = [c for c in cl if match_name(k, c)]
        pool = strong if strong else cl
        if pool:
            sims = [(max((difflib.SequenceMatcher(None, str(c).lower(), a).ratio() for a in ALT[k]), default=0), c) for c in pool]
            sims.sort(reverse=True)
            if len(strong) == 1: pick = strong[0]
            elif strong: pick = sims[0][1]
            elif len(cl) == 1 or sims[0][0] >= 0.72: pick = sims[0][1]
        if pick: prop[k] = pick; used.add(pick)
    tick_cols = [c for c, st in info.items() if st["kind"] == "obj" and st.get("tick") and st["nun"] > 30 and c not in used]
    if "cik" in missing and "cik" not in prop and tick_cols: prop["_ticker"] = tick_cols[0]
    return prop, cands

# ---- Pass 1: project tree (header + content). Pass 2: global (header-only). ----
PANEL_PATH, MAP, TICKER_COL, best = None, {}, None, []
pool1 = sorted(in_project, key=name_score, reverse=True)
pool2 = sorted([p for p in data_files if p not in in_project], key=name_score, reverse=True)[:150]
for phase, pool in ((1, pool1), (2, pool2)):
    for p in pool:
        cols, smp = read_head(p)
        if cols is None: continue
        low = {str(c).lower(): c for c in cols}
        hdr_tick = next((low[a] for a in TICK_ALIASES if a in low), None)
        m = resolve(cols)
        missing = [k for k in NEED if k not in m]
        cg = {}
        if phase == 1 and missing and smp is not None:
            cg, _c = content_guess(profile(smp), missing)
            for k, c in cg.items():
                if k != "_ticker": m[k] = c
            missing = [k for k in NEED if k not in m]
            if missing == ["cik"] and "_ticker" in cg: TICKER_COL = cg["_ticker"]; missing = []
        best.append((len(NEED) - len(missing), p, missing, m, cg.get("_ticker") or hdr_tick))
        if missing and all(k in ("capex",) for k in missing):
            print(f"  accepting {os.path.basename(p)} - only optional field(s) {missing} absent")
            missing = []
        if not missing:
            PANEL_PATH, MAP = p, m
            break
    if PANEL_PATH: break

# ---- CI-split rescue (content-based, overlap-validated) ----
CI_PATH, CI_MAP, CI_TICK = (CI_FILE_OVERRIDE or None), {}, None
def find_ci_file(exclude, anchor_ciks=None):
    """Any project file with year + (cik|ticker) + a CI-shaped column; validated by firm overlap."""
    for p2 in pool1:
        if p2 in exclude: continue
        c2, s2 = read_head(p2)
        if c2 is None: continue
        m2 = resolve(c2)
        low2 = {str(c).lower(): c for c in c2}
        t2 = next((low2[a] for a in TICK_ALIASES if a in low2), None)
        if s2 is not None and ("ci" not in m2 or "year" not in m2):
            gg, _cc = content_guess(profile(s2), [k for k in ("ci", "year", "cik") if k not in m2])
            for k in ("ci", "year", "cik"):
                if k in gg: m2[k] = gg[k]
            t2 = t2 or gg.get("_ticker")
        if not ("ci" in m2 and "year" in m2 and ("cik" in m2 or t2)):
            continue
        if anchor_ciks:
            try:
                if "cik" in m2:
                    ids = pd.to_numeric(read_col(p2, m2["cik"]), errors="coerce").dropna().astype(int)
                else:
                    ids = read_col(p2, t2).astype(str).str.upper().map(tick2cik()).dropna().astype(int)
                if ids.isin(list(anchor_ciks)).mean() < 0.5:
                    continue
            except Exception:
                continue
        return p2, m2, t2
    return None, {}, None

if PANEL_PATH is None:
    best.sort(key=lambda t: t[0], reverse=True)
    for score, p, missing, m, tk in best[:12]:
        if set(missing) <= {"ci"} or (set(missing) <= {"ci", "cik"} and tk):
            anchor = None
            try:
                if "cik" in m:
                    anchor = set(pd.to_numeric(read_col(p, m["cik"]), errors="coerce").dropna().astype(int).unique())
            except Exception:
                anchor = None
            p2, m2, t2 = find_ci_file({p}, anchor)
            if p2:
                PANEL_PATH, MAP, TICKER_COL = p, m, tk
                CI_PATH, CI_MAP, CI_TICK = p2, m2, t2
                print(f"CI-split rescue: fundamentals {p} + CI {p2}")
                break

# ---- CIK sanity: reject identifier columns that are not SEC CIKs ----
CIK_SHARE = None
if PANEL_PATH is not None and "cik" in MAP:
    _v = read_col(PANEL_PATH, MAP["cik"])
    if _v is not None:
        CIK_SHARE = cik_share(pd.to_numeric(_v, errors="coerce").dropna().astype(int).unique())
        if CIK_SHARE is not None:
            print(f"CIK column '{MAP['cik']}' → SEC registry match {CIK_SHARE:.0%}")
            if CIK_SHARE < 0.5:
                cols_p, _sx = read_head(PANEL_PATH, 5)
                lowp = {str(c).lower(): c for c in (cols_p or [])}
                tcol = next((lowp[a] for a in TICK_ALIASES if a in lowp), None)
                if tcol:
                    print(f"→ rejected as CIK; using ticker column '{tcol}' via the SEC map instead")
                    MAP.pop("cik", None); TICKER_COL = tcol
                else:
                    print("⚠ identifier is likely NOT a CIK and no ticker column exists — expect mass skips; paste the RUN DIAGNOSTIC block")

# ============ DERIVATION LAYER ============
# The CCAOF panel stores CONSTITUENTS, not canonical fields: Core Integrity lives in the five
# CIAOF KPI columns, and outcomes/controls are raw line items. Derive rather than hunt.
KPI_NAMES = ["standardization_kpi", "data_harmonization_kpi", "api_integration_kpi",
             "modular_extensibility_kpi", "ops_automation_kpi"]
OUTCOMES  = ["roa", "q", "sga"]             # dependent variables: required per-regression, NOT panel-wide
NEED_OPT  = ["capex"]                       # nice-to-have control: absence must not block the run
NEED_CORE = [k for k in NEED if k not in NEED_OPT + OUTCOMES]   # cik, year, ci, size, lev, rnd
FIELD_PATS = {
  "assets":    [r"^assets$", r"^assets[_ ]?sf$", r"total[_ ]?assets", r"^at$"],
  "revenue":   [r"^revenue$", r"^revenue[_ ]?sf$", r"^sales$", r"total[_ ]?revenue", r"^revt$"],
  "netinc":    [r"^net[_ ]?income$", r"^net[_ ]?income[_ ]?sf$", r"net[_ ]?income", r"^ni$"],
  "liab":      [r"total[_ ]?liabilities$", r"^liabilities$", r"^liab", r"^lt$", r"total[_ ]?liab"],
  "sga_exp":   [r"sga[_ ]?expense", r"^sga$", r"^sga[_ ]?sf$", r"selling.*general"],
  "rd_exp":    [r"rd[_ ]?expense", r"^rd[_ ]?sf$", r"research.*development", r"^xrd$"],
  "capex_exp": [r"^capex", r"capital[_ ]?expenditure", r"^capx$", r"purchase.*fixed[_ ]?assets"],
  "mktcap":    [r"^mktcap", r"market[_ ]?cap", r"market[_ ]?value.*equity"],
  "tobin":     [r"tobin"],
  "ci_col":    [r"^ccai[_ ]?equal$", r"^ccai$", r"^ci[_ ]?score$", r"^core[_ ]?integrity$"],
  "ci_alt":    [r"^ccai[_ ]?pca$", r"^ccai[_ ]?textrich$"],
  "roa_col":   [r"^roa$", r"return[_ ]?on[_ ]?assets"],
  "lev_col":   [r"^leverage$", r"debt[_ ]?ratio", r"^lev$"],
  "sga_col":   [r"^sga[_ ]?intensity", r"^sga[_ ]?ratio"],
  "rnd_col":   [r"^rd[_ ]?intensity", r"^rnd[_ ]?int", r"^rd[_ ]?sale"],
  "size_col":  [r"^log[_ ]?assets$", r"^ln[_ ]?assets$", r"log[_ ]?at$"],
  "capsoft":   [r"^capsoft[_ ]?intensity", r"capitali[sz]ed[_ ]?software"],
  "cik_col":   [r"^cik$", r"^cik[_ ]?code$", r"^sec[_ ]?cik$"],
  "tick_col":  [r"^ticker$", r"^tic$", r"^symbol$"],
}

AUX_PATS = {   # adjacent in-panel signals: convergent-validity checks against EAGI
  "aux_gov_it":        [r"^gov[_ ]?it[_ ]?signal", r"gov.*it.*signal"],
  "aux_it_intensity":  [r"^it[_ ]?intensity$"],
  "aux_ccai_textrich": [r"^ccai[_ ]?textrich$"],
  "aux_n_tokens":      [r"^n[_ ]?tokens$"],
  "aux_gh_api":        [r"^gh[_ ]?api[_ ]?ratio$"],
  "aux_gh_modular":    [r"^gh[_ ]?modular[_ ]?ratio$"],
  "aux_gh_ops":        [r"^gh[_ ]?ops[_ ]?ratio$"],
}

def _pick(cols, pats, prefer_sf=True):
    """First pattern match; among matches prefer the SimFin (_sf) family so ratio
    numerators and denominators come from the same reporting source."""
    low = [(str(c).lower(), c) for c in cols]
    hits = []
    for pat in pats:
        for lc, orig in low:
            if re.search(pat, lc) and orig not in hits: hits.append(orig)
    if not hits: return None
    if prefer_sf:
        sf = [h for h in hits if str(h).lower().endswith("_sf")]
        if sf: return sf[0]
    return hits[0]

def _num(fr, c):
    return pd.to_numeric(fr[c], errors="coerce")

def cronbach(X):
    X = X.dropna()
    k = X.shape[1]
    if k < 2 or len(X) < 10: return None
    tot = X.sum(axis=1).var(ddof=1)
    if not tot: return None
    return float(k / (k - 1) * (1 - X.var(ddof=1).sum() / tot))

def derive_fields(fr):
    """Return ({canonical: Series}, {canonical: provenance string})."""
    got, how = {}, {}
    C = list(fr.columns)
    f = {k: _pick(C, p) for k, p in FIELD_PATS.items()}
    # --- Core Integrity: prefer the PUBLISHED composite over any recomputation ---
    kpi = [c for c in C if str(c).lower() in KPI_NAMES] or [c for c in C if str(c).lower().endswith("_kpi")]
    alpha = None
    if len(kpi) >= 2:
        alpha = cronbach(pd.concat([_num(fr, c) for c in kpi], axis=1))
    if f["ci_col"]:
        got["ci"] = _num(fr, f["ci_col"])
        how["ci"] = f"published composite '{f['ci_col']}'" + (f" · KPI α={alpha:.3f}" if alpha is not None else "")
    elif len(kpi) >= 3:
        X = pd.concat([_num(fr, c) for c in kpi], axis=1); X.columns = [str(c) for c in kpi]
        got["ci"] = ((X - X.mean()) / X.std().replace(0, 1)).mean(axis=1)
        how["ci"] = f"z-mean of {len(kpi)} CIAOF KPIs" + (f" · α={alpha:.3f}" if alpha is not None else "")
    if f["ci_alt"]:
        got["ci_alt"] = _num(fr, f["ci_alt"]); how["ci_alt"] = f"robustness composite '{f['ci_alt']}'"
    P = lambda pats: _pick(C, pats, prefer_sf=False)
    c = dict(
        assets_sf=P([r"^assets[_ ]?sf$"]), assets=P([r"^assets$"]),
        rev_sf=P([r"^revenue[_ ]?sf$"]),   rev=P([r"^revenue$"]),
        ni_sf=P([r"^net[_ ]?income[_ ]?sf$"]), ni=P([r"^net[_ ]?income$"]),
        liab_sf=P([r"^liab[_ ]?sf$"]), liab=P([r"total[_ ]?liabilities$", r"^liabilities$"]),
        sga_int=P([r"^sga[_ ]?intensity$"]), sga_sf=P([r"^sga[_ ]?sf$"]), sga_exp=P([r"^sga[_ ]?expense$"]),
        rd_int=P([r"^rd[_ ]?intensity[_ ]?sf$", r"^rd[_ ]?intensity$"]), rd_sf=P([r"^rd[_ ]?sf$"]), rd_exp=P([r"^rd[_ ]?expense$"]),
        mkt_sf=P([r"^mktcap[_ ]?sf$"]), tobin=P([r"tobin"]),
        nm_sf=P([r"^net[_ ]?margin[_ ]?sf$"]), at_sf=P([r"^asset[_ ]?turnover[_ ]?sf$"]),
        logat=P([r"^log[_ ]?assets$", r"^ln[_ ]?assets$"]),
    )
    def cc(k):
        return _num(fr, c[k]) if c.get(k) else None
    def rat(n_, d_, absolute=False):
        if n_ is None or d_ is None: return None
        return (n_.abs() if absolute else n_) / d_.replace(0, np.nan)
    def coal(name, *cands):
        """Fill gaps across ordered fallbacks, recording HOW MANY rows each source contributed
        so measurement heterogeneity is auditable rather than hidden."""
        ser, used = None, []
        for lbl, s_ in cands:
            if s_ is None: continue
            if ser is None:
                ser = s_.copy(); used.append((lbl, int(ser.notna().sum())))
            else:
                before = int(ser.notna().sum())
                ser = ser.fillna(s_)
                used.append((lbl, int(ser.notna().sum()) - before))
        if ser is not None:
            got[name] = ser
            live = [(l, k) for l, k in used if k > 0]
            how[name] = " + ".join(f"{l} [{k} rows]" for l, k in live) + (" (coalesced)" if len(live) > 1 else "")
        return ser

    A = cc("assets_sf") if c.get("assets_sf") else cc("assets")
    A2 = cc("assets") if c.get("assets") else None
    R = cc("rev_sf") if c.get("rev_sf") else cc("rev")
    R2 = cc("rev") if c.get("rev") else None

    coal("size", (f"direct '{c['logat']}'", cc("logat")),
                 (f"log({c.get('assets_sf') or c.get('assets')})", np.log(A.clip(lower=1)) if A is not None else None),
                 (f"log({c.get('assets')})", np.log(A2.clip(lower=1)) if A2 is not None else None))
    coal("roa", (f"direct '{f['roa_col']}'", cc("roa_col") if f["roa_col"] else None),
                (f"{c['ni_sf']}/{c.get('assets_sf')}", rat(cc("ni_sf"), A)),
                (f"{c['nm_sf']}×{c['at_sf']}", (cc("nm_sf") * cc("at_sf")) if (c.get("nm_sf") and c.get("at_sf")) else None),
                (f"{c['ni']}/{c.get('assets')}", rat(cc("ni"), A2)))
    coal("lev", (f"direct '{f['lev_col']}'", cc("lev_col") if f["lev_col"] else None),
                (f"{c['liab_sf']}/{c.get('assets_sf')}", rat(cc("liab_sf"), A)),
                (f"{c['liab']}/{c.get('assets')}", rat(cc("liab"), A2)))
    coal("sga", (f"direct '{c['sga_int']}'", cc("sga_int")),
                (f"|{c['sga_sf']}|/{c.get('rev_sf')}", rat(cc("sga_sf"), R, absolute=True)),
                (f"|{c['sga_exp']}|/{c.get('rev')}", rat(cc("sga_exp"), R2, absolute=True)))
    coal("rnd", (f"direct '{c['rd_int']}'", cc("rd_int")),
                (f"|{c['rd_sf']}|/{c.get('rev_sf')}", rat(cc("rd_sf"), R, absolute=True)),
                (f"|{c['rd_exp']}|/{c.get('rev')}", rat(cc("rd_exp"), R2, absolute=True)))
    coal("q", (f"direct '{c['tobin']}'", cc("tobin")),
              (f"({c['mkt_sf']}+{c['liab_sf']})/{c.get('assets_sf')}",
               rat(cc("mkt_sf") + (cc("liab_sf") if c.get("liab_sf") else 0), A) if c.get("mkt_sf") else None))
    # R&D: absence of an R&D line is a reported zero, not missing data (standard convention)
    if "rnd" in got:
        _m = got["rnd"].isna()
        got["rnd_missing"] = _m.astype(float)
        got["rnd"] = got["rnd"].fillna(0.0)
        how["rnd"] += f" · missing→0 for {int(_m.sum())} rows + rnd_missing dummy"
        how["rnd_missing"] = "indicator: firm-year reports no R&D line"
    if f["capsoft"]:
        got["capsoft"] = _num(fr, f["capsoft"]); how["capsoft"] = f"capital-intensity proxy '{f['capsoft']}' (cash capex unavailable)"
    if f["capex_exp"] and A is not None:
        got["capex"] = _num(fr, f["capex_exp"]).abs() / A.replace(0, np.nan); how["capex"] = f"|{f['capex_exp']}|/{c.get('assets_sf')}"
    for _kc in [c for c in C if str(c).lower().endswith("_kpi")]:
        got["kpi_" + str(_kc)] = _num(fr, _kc)
        how["kpi_" + str(_kc)] = f"CIAOF dimension '{_kc}' (carried for dimension-level tests)"
    _ind = _pick(C, [r"^industry$", r"^sector$", r"^sic", r"^gics"], prefer_sf=False)
    if _ind is not None:
        got["industry_raw"] = fr[_ind].astype(str); how["industry_raw"] = f"industry label '{_ind}'"
    for _ak, _ap in AUX_PATS.items():
        _c = _pick(C, _ap, prefer_sf=False)
        if _c is not None:
            got[_ak] = _num(fr, _c); how[_ak] = f"validation signal '{_c}'"
    return got, how

def simfin_id_to_cik():
    """Bridge SimFinId → Ticker → CIK using the SimFin cache (semicolon-delimited)."""
    for p in pool1:
        if not p.lower().endswith(".csv") or _is_ours(p): continue
        try:
            fr = pd.read_csv(p, sep=_csv_sep(p), nrows=300000, low_memory=False)
        except Exception:
            continue
        low = {str(c).lower().replace(" ", ""): c for c in fr.columns}
        if "simfinid" in low and "ticker" in low:
            t2c = tick2cik()
            mm = fr[[low["simfinid"], low["ticker"]]].drop_duplicates()
            mm["_cik"] = mm[low["ticker"]].astype(str).str.upper().map(t2c)
            mm = mm.dropna(subset=["_cik"])
            print(f"  SimFin bridge from {os.path.basename(p)}: {len(mm)} ids → CIK")
            return {str(k): int(v) for k, v in zip(mm[low["simfinid"]], mm["_cik"])}
    return {}

CACHE_ARTIFACTS = ("term_counts.csv", "techcom_raw.csv", "panel_with_eagi.csv",
                   "eagi_main_results.csv", "eagi_loto.csv", "company_tickers.json")

def _is_ours(p):
    """Never treat our own caches/outputs as source data."""
    pl = p.lower()
    return ("eagi_cache" in pl) or ("eagi_out" in pl) or (os.path.basename(pl) in CACHE_ARTIFACTS)

def find_cik_series(fr):
    """Locate CIKs by CONTENT (SEC registry match), never by column name."""
    try:
        tkmap = tick2cik(); reg = set(tkmap.values())
    except Exception as e:
        print("  registry unavailable:", type(e).__name__); return None, None
    bestc, bests = None, 0.0
    for c in fr.columns:
        try:
            v = pd.to_numeric(fr[c], errors="coerce").dropna()
            if len(v) < 10: continue
            if (v % 1 == 0).mean() < 0.99: continue
            if v.min() < 1 or v.max() > 2_000_000: continue
            sh = float(v.astype(int).isin(reg).mean())
            if sh > bests: bestc, bests = c, sh
        except Exception:
            continue
    if bestc is not None and bests >= 0.9:
        print(f"  CIK by content → column '{bestc}' · registry match {bests:.0%}")
        return pd.to_numeric(fr[bestc], errors="coerce"), f"content:{bestc}"
    for c in fr.columns:
        try:
            u = fr[c].astype(str).str.upper().str.strip()
            sh = float(u.isin(tkmap.keys()).mean())
            if sh >= 0.8:
                print(f"  CIK via ticker column '{c}' · {sh:.0%} of rows map")
                return u.map(tkmap), f"ticker:{c}"
        except Exception:
            continue
    if bestc is not None:
        print(f"  best numeric identifier '{bestc}' only {bests:.0%} registry match — rejected")
    return None, None

def assemble_panel():
    """Merge project tables on (firm key, year), derive canonical fields, attach real CIKs."""
    frames = []
    for p in pool1:
        if _is_ours(p): continue
        cols, _s = read_head(p, 5)
        if cols is None: continue
        low = {str(c).lower(): c for c in cols}
        kid = next((low[k] for k in ("firm_id", "cik", "ticker", "tic", "symbol") if k in low), None)
        yid = next((low[k] for k in ("year", "fyear", "fiscal_year", "fy", "fiscal year") if k in low), None)
        if not (kid and yid): continue
        try:
            fr = read_full(p)
        except Exception as e:
            print("  skip", os.path.basename(p), type(e).__name__); continue
        if len(fr) > 400_000:
            print("  skip (too large)", os.path.basename(p)); continue
        fr = fr.copy()
        fr["_fid"] = fr[kid].astype(str)
        fr["_yr"] = pd.to_numeric(fr[yid], errors="coerce")
        fr = fr.dropna(subset=["_yr"])
        if not len(fr): continue
        fr["_yr"] = fr["_yr"].astype(int)
        fr = fr.drop_duplicates(subset=["_fid", "_yr"])
        frames.append((p, fr))
    if not frames:
        print("  no table carries a firm key + year — derivation impossible"); return None
    frames.sort(key=lambda t: -t[1].shape[1])
    base_path, merged = frames[0][0], frames[0][1].copy()
    print(f"  base: {os.path.basename(base_path)} — {merged.shape[0]} rows × {merged.shape[1]} cols")
    print(f"  base columns: {[c for c in merged.columns if not str(c).startswith('_')]}")
    keys = set(merged["_fid"])
    for p, fr in frames[1:]:
        ov = float(fr["_fid"].isin(keys).mean())
        if ov < 0.2:
            print(f"    – {os.path.basename(p)}: key overlap {ov:.0%} → skipped"); continue
        newc = [c for c in fr.columns if c not in merged.columns]
        if not newc: continue
        try:
            merged = merged.merge(fr[["_fid", "_yr"] + newc], on=["_fid", "_yr"], how="left")
            print(f"    + {os.path.basename(p)} (overlap {ov:.0%}, +{len(newc)} cols)")
        except Exception:
            continue
    got, how = derive_fields(merged)
    globals()["DERIV_HOW"] = how
    out = pd.DataFrame({"year": merged["_yr"].values})
    ciks, prov = find_cik_series(merged)
    if ciks is None:
        bridge = simfin_id_to_cik()
        if bridge:
            ciks = merged["_fid"].astype(str).map(bridge)
            print(f"  SimFinId bridge → CIK for {ciks.notna().mean():.0%} of rows")
            if ciks.notna().mean() < 0.5: ciks = None
    if ciks is None:
        print("  ✗ no usable CIK — cannot align the panel to the filing text layer")
        print("   full merged column list:", list(merged.columns))
        return None
    out["cik"] = ciks.values
    for k, v in got.items(): out[k] = v.values if hasattr(v, "values") else v
    print("  derived fields:")
    for k in NEED + ["capsoft", "ci_alt"] + [a for a in AUX_PATS if a in out.columns]:
        print(f"    {k:6s} ← " + (how.get(k, "(direct)") if k in out.columns else "MISSING"))
    out = out.dropna(subset=["cik", "year"])
    out["cik"] = out["cik"].astype(int); out["year"] = out["year"].astype(int)
    _dup = int(out.duplicated(subset=["cik", "year"]).sum())
    if _dup:
        print(f"  ⚠ {_dup} duplicate (cik,year) rows — collapsing by mean "
              f"(distinct firm_ids sharing one CIK, e.g. dual share classes)")
        _ind_keep = (out.groupby(["cik", "year"], as_index=False)["industry_raw"].first()
                     if "industry_raw" in out.columns else None)
        out = out.groupby(["cik", "year"], as_index=False).mean(numeric_only=True)
        if _ind_keep is not None:
            out = out.merge(_ind_keep, on=["cik", "year"], how="left")
    miss = [k for k in NEED_CORE if k not in out.columns]
    for k in NEED_OPT:
        if k not in out.columns: print(f"  note: optional control '{k}' unavailable — dropped from the control set")
    if miss:
        print(f"  ✗ still missing {miss} — full merged column list:")
        print("   ", list(merged.columns))
        return None
    print(f"  field coverage before complete-case filter ({len(out)} rows):")
    for _k in NEED_CORE + OUTCOMES + [x for x in ("capex", "capsoft", "rnd_missing") if x in out.columns]:
        print(f"    {_k:8s} {int(out[_k].notna().sum()):5d}  ({out[_k].notna().mean():.0%})")
    _full = out.dropna(subset=NEED_CORE)
    for _k in NEED_CORE:
        _alt = out.dropna(subset=[c for c in NEED_CORE if c != _k])
        if len(_alt) > 1.2 * max(1, len(_full)):
            print(f"    ⚠ excluding '{_k}' would raise the panel {len(_full)} → {len(_alt)} rows")
    if "industry_raw" in out.columns: out["industry_raw"] = out["industry_raw"].fillna("unknown")
    out = out.dropna(subset=NEED_CORE)
    _oc = [o for o in OUTCOMES if o in out.columns]
    if _oc:
        out = out[out[_oc].notna().any(axis=1)]
        print("  outcome availability inside the panel: " + " · ".join(f"{o}={int(out[o].notna().sum())}" for o in _oc))
    print(f"  ✓ DERIVED PANEL: {len(out)} firm-years · {out.cik.nunique()} firms · {out.year.min()}–{out.year.max()}")
    return out

DERIVED_PANEL, DERIVE_LOG = None, ""
if PANEL_PATH is None:
    import io, contextlib
    print("\nDirect resolution failed → DERIVING canonical panel from project tables …")
    _buf = io.StringIO()
    try:
        with contextlib.redirect_stdout(_buf):
            DERIVED_PANEL = assemble_panel()
    except Exception as e:
        _buf.write(f"  derivation error: {type(e).__name__}: {e}\n")
    DERIVE_LOG = _buf.getvalue()
    print(DERIVE_LOG)

# ---- Prefetch-only fallback ----
PREFETCH_ONLY, FIRM_CIKS = False, None
if PANEL_PATH is None and DERIVED_PANEL is None:
    print("\n──────── PROJECT DATA INVENTORY (full column lists) ────────")
    for p in pool1[:90]:
        cols, _ = read_head(p, n=5)
        rel = p.replace(DRIVE_ROOT + "/", "")
        print(f"· {rel}\n    cols[{0 if cols is None else len(cols)}]: {list(cols) if cols else 'UNREADABLE'}")
    if READ_ERRORS:
        from collections import Counter
        kinds = Counter(e.split(":")[0] for _, e in READ_ERRORS)
        print(f"\n  ⚠ {len(READ_ERRORS)} files could not be read. Error types: {dict(kinds)}")
        for _pp, _ee in READ_ERRORS[:5]:
            print(f"      {os.path.basename(_pp)}: {_ee}")
        print("  If EVERY file failed, this is an ENVIRONMENT fault, not a data fault. The usual")
        print("  cause is pip upgrading numpy/pandas inside a live kernel: Runtime → Restart")
        print("  session, then Run all. If it recurs after a restart, the error types above")
        print("  identify the real cause - do not assume pip.")
    best.sort(key=lambda t: t[0], reverse=True)
    if best:
        s0 = best[0]
        print(f"\nBest candidate: {s0[1]} — resolved {s0[0]}/{len(NEED)}, missing {s0[2]}")
    for score, p, missing, m, tk in best:
        ciks = None
        if "cik" in m:
            v = read_col(p, m["cik"])
            if v is not None: ciks = pd.to_numeric(v, errors="coerce").dropna().astype(int)
        if ciks is None and tk:
            v = read_col(p, tk)
            if v is not None:
                try:
                    t2c = tick2cik()
                    ciks = v.astype(str).str.upper().map(t2c).dropna().astype(int)
                except Exception as e:
                    print("ticker→CIK unavailable:", type(e).__name__)
        if ciks is not None:
            _sh = cik_share(ciks.unique())
            if _sh is not None:
                if _sh < 0.5:
                    print(f"  {p}: identifier rejected as CIK ({_sh:.0%} SEC-registry match)")
                    ciks = None
                elif _sh < 0.9:
                    u = set(tick2cik().values()); n0 = ciks.nunique()
                    ciks = ciks[ciks.isin(list(u))]
                    print(f"  {p}: filtered to SEC-valid CIKs {ciks.nunique()}/{n0} (registry match {_sh:.0%})")
                CIK_SHARE = _sh
        if ciks is not None and ciks.nunique() >= 20:
            FIRM_CIKS = sorted(ciks.unique().tolist())
            print(f"\nPLAN B: firm universe from {p} → {len(FIRM_CIKS)} CIKs. Text-layer prefetch will run and cache to Drive.")
            PREFETCH_ONLY = True
            DATA_DIR = os.path.dirname(p)
            break
    if FIRM_CIKS is None:
        raise RuntimeError("Panel unresolved AND no CIK/ticker column found anywhere — paste the inventory above; mapping is one COLMAP line.")
elif DERIVED_PANEL is not None:
    FIRM_CIKS = sorted(DERIVED_PANEL.cik.unique().tolist())
    CIK_SHARE = cik_share(FIRM_CIKS)
    DATA_DIR = os.path.join(DRIVE_ROOT, PROJECT_HINT + "-research", "data", "processed")
    if not os.path.isdir(DATA_DIR):
        DATA_DIR = os.path.dirname(pool1[0]) if pool1 else DRIVE_ROOT
else:
    DATA_DIR = os.path.dirname(PANEL_PATH)

CACHE_DIR = max(cache_dirs, key=lambda dd: len(os.listdir(dd)), default=os.path.join(DATA_DIR, "eagi_cache"))
OUT_DIR = os.path.join(DATA_DIR, "eagi_out")
os.makedirs(CACHE_DIR, exist_ok=True); os.makedirs(OUT_DIR, exist_ok=True)
open(os.path.join(CACHE_DIR, "user_agent.txt"), "w").write(USER_AGENT)

if PREFETCH_ONLY:
    print(f"\nMODE → PREFETCH-ONLY · caches → {CACHE_DIR}")
elif DERIVED_PANEL is not None:
    print(f"\nMODE → DERIVED PANEL · {len(DERIVED_PANEL)} firm-years · {len(FIRM_CIKS)} firms · registry match {CIK_SHARE:.0%}")
    print(f"CACHE  → {CACHE_DIR}  ({len(os.listdir(CACHE_DIR))} files reused)")
else:
    print(f"\nPANEL  → {PANEL_PATH}")
    print(f"MAP    → {MAP}" + (f"  · ticker column: {TICKER_COL}" if TICKER_COL else ""))
    if CI_PATH: print(f"CI     → {CI_PATH}  map {CI_MAP}" + (f" · ticker {CI_TICK}" if CI_TICK else ""))
    print(f"CACHE  → {CACHE_DIR}  ({len(os.listdir(CACHE_DIR))} files reused)")
````

````python
# ============ 2 · LOAD PANEL (skipped in prefetch-only mode) ============
def load_panel():
    df = read_full(PANEL_PATH)
    if COLMAP: df = df.rename(columns={v: k for k, v in COLMAP.items()})
    df = df.rename(columns={v: k for k, v in MAP.items()})
    if "cik" not in df.columns and TICKER_COL:
        t2c = tick2cik()
        df["cik"] = df[TICKER_COL].astype(str).str.upper().map(t2c)
        n0 = len(df); df = df.dropna(subset=["cik"])
        print(f"ticker→CIK: {len(df)}/{n0} rows mapped")
    if CI_PATH:
        ci = read_full(CI_PATH).rename(columns={v: k for k, v in CI_MAP.items()})
        if "cik" not in ci.columns and CI_TICK:
            t2c = tick2cik(); ci["cik"] = ci[CI_TICK].astype(str).str.upper().map(t2c); ci = ci.dropna(subset=["cik"])
        ci["cik"] = pd.to_numeric(ci["cik"], errors="coerce"); ci["year"] = pd.to_numeric(ci["year"], errors="coerce")
        ci = ci.dropna(subset=["cik", "year"]); ci["cik"] = ci["cik"].astype(int); ci["year"] = ci["year"].astype(int)
        df = df.merge(ci[["cik", "year", "ci"]], on=["cik", "year"], how="inner")
        print(f"CI merged from separate file → {len(df)} firm-years")
    missing = [c for c in NEED if c not in df.columns]
    assert not missing, f"Panel missing {missing} after mapping — set COLMAP."
    df["cik"] = pd.to_numeric(df["cik"], errors="coerce"); df["year"] = pd.to_numeric(df["year"], errors="coerce")
    df = df.dropna(subset=["cik", "year"])
    df["cik"] = df["cik"].astype(int); df["year"] = df["year"].astype(int)
    out = df[(df.year >= YEAR_MIN) & (df.year <= YEAR_MAX)].copy()
    assert out.cik.nunique() >= 30 and out.year.nunique() >= 3, (
        f"Resolved table looks wrong: {out.cik.nunique()} firms × {out.year.nunique()} years from {PANEL_PATH} — "
        "likely mis-detected; set COLMAP / CI_FILE_OVERRIDE and paste the RUN DIAGNOSTIC block.")
    return out

if G("PREFETCH_ONLY") is None:
    panel = None
    print("UPSTREAM INCOMPLETE — cell 1 did not finish. See the RUN DIAGNOSTIC cell at the end.")
elif G("DERIVED_PANEL") is not None:
    panel = DERIVED_PANEL[(DERIVED_PANEL.year >= YEAR_MIN) & (DERIVED_PANEL.year <= YEAR_MAX)].copy()
    FIRM_CIKS = sorted(panel.cik.unique().tolist())
    print(panel.shape, "firm-years ·", panel.cik.nunique(), "firms  [derived]")
elif PREFETCH_ONLY:
    panel = None
    print(f"PREFETCH-ONLY — {len(FIRM_CIKS)} firms; estimation cells will self-skip until the panel is mapped.")
else:
    panel = load_panel()
    FIRM_CIKS = sorted(panel.cik.unique().tolist())
    _sh = cik_share(FIRM_CIKS)
    if _sh is not None:
        globals()["CIK_SHARE"] = _sh
        if _sh < 0.9:
            _u = set(tick2cik().values()); _n0 = len(FIRM_CIKS)
            FIRM_CIKS = [c for c in FIRM_CIKS if c in _u]
            print(f"CIK sanity: kept {len(FIRM_CIKS)}/{_n0} SEC-valid CIKs (registry match {_sh:.0%})")
        else:
            print(f"CIK sanity: registry match {_sh:.0%}")
    print(panel.shape, "firm-years ·", panel.cik.nunique(), "firms")
````

````python
# ============ 3 · EDGAR FETCHERS (cached, rate-limited, lazy headers) ============
import requests
from bs4 import BeautifulSoup

def _hd():
    return {"User-Agent": G("USER_AGENT") or "", "Accept-Encoding": "gzip, deflate"}

def _get(url, binary=False):
    time.sleep(SLEEP)
    r = requests.get(url, headers=_hd(), timeout=30)
    r.raise_for_status()
    return r.content if binary else r.text

def submissions(cik):
    p = f"{CACHE_DIR}/sub_{cik}.json"
    if os.path.exists(p):
        return json.load(open(p))
    j = json.loads(_get(f"https://data.sec.gov/submissions/CIK{int(cik):010d}.json"))
    json.dump(j, open(p, "w"))
    return j

def filings_of(cik, form):
    j = submissions(cik); rec = j["filings"]["recent"]
    for form_i, acc, doc, fdate, rdate in zip(rec["form"], rec["accessionNumber"], rec["primaryDocument"], rec["filingDate"], rec.get("reportDate", rec["filingDate"])):
        if form_i != form:
            continue
        ry = (rdate or fdate)[:4]
        yield int(ry), acc.replace("-", ""), doc

def doc_text(cik, acc, doc):
    p = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
    if os.path.exists(p):
        return open(p, encoding="utf-8", errors="ignore").read()
    url = f"https://www.sec.gov/Archives/edgar/data/{int(cik)}/{acc}/{doc}"
    html = _get(url)
    text = BeautifulSoup(html, "lxml").get_text(" ").lower()
    text = re.sub(r"\s+", " ", text)
    open(p, "w", encoding="utf-8").write(text)
    return text

print("fetchers ready — raw filings cached under", G("CACHE_DIR"))
````

````python
# ============ 4 · EAGI CONSTRUCTION (firm-level fault isolation) ============
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    counts = None; Z = None; df = None
    print("UPSTREAM INCOMPLETE — EAGI construction skipped. See the RUN DIAGNOSTIC cell.")
else:
    def count_terms(text, terms):
        words = max(1, len(text.split()))
        out = {}
        for t in terms:
            patt = r"\b" + re.escape(t).replace(r"\-", "[- ]") + r"\b"
            out[t] = len(re.findall(patt, text))
        out["_words"] = words
        return out

    tc_path = f"{CACHE_DIR}/term_counts.csv"
    counts = None
    if os.path.exists(tc_path):
        prev = pd.read_csv(tc_path)
        if set(LEXICON + PLACEBO + ["_words", "cik", "year"]).issubset(prev.columns):
            counts = prev; print("Reusing cached term_counts.csv —", len(counts), "firm-years")
        else:
            print("Cached term_counts.csv lacks current terms → recounting from cached filings")
    if counts is None:
        recs, firm_fail = [], 0
        for cik in tqdm(FIRM_CIKS, desc="10-K"):
            try:
                fl = list(filings_of(cik, "10-K"))
            except Exception as e:
                firm_fail += 1; print("skip firm", cik, type(e).__name__, getattr(getattr(e, 'response', None), 'status_code', '') or ''); continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX):
                    continue
                try:
                    c = count_terms(doc_text(cik, acc, doc), LEXICON + PLACEBO)
                    c.update({"cik": cik, "year": fy}); recs.append(c)
                except Exception as e:
                    print("skip doc", cik, fy, type(e).__name__)
        if firm_fail: print(f"firms skipped entirely: {firm_fail}")
        counts = pd.DataFrame(recs).groupby(["cik", "year"], as_index=False).first()
        counts.to_csv(tc_path, index=False)
        cov = counts.cik.nunique() / max(1, len(FIRM_CIKS))
        print(f"coverage: {counts.cik.nunique()}/{len(FIRM_CIKS)} firms with in-window 10-Ks ({cov:.0%})")
        if cov < 0.5:
            print("⚠ LOW COVERAGE — identifier may not be CIK, or window mismatch. Paste the RUN DIAGNOSTIC block.")

    Z = counts[["cik", "year"]].copy()
    for t in LEXICON:
        rate = np.log1p(counts[t] / counts["_words"] * 10000)
        Z[t] = (rate - rate.mean()) / (rate.std() or 1)
    Z["eagi"] = Z[LEXICON].mean(axis=1)

    # ---- construct diagnostics: does the lexicon measure ONE thing? ----
    _ZT = Z[LEXICON]
    _k = _ZT.shape[1]
    _tot = _ZT.sum(axis=1).var(ddof=1)
    LEX_ALPHA = float(_k / (_k - 1) * (1 - _ZT.var(ddof=1).sum() / _tot)) if _tot else float("nan")
    _C = _ZT.corr().to_numpy()
    _iu = np.triu_indices(_k, 1)
    LEX_MEAN_R = float(np.nanmean(_C[_iu]))
    try:
        _ev = np.linalg.eigvalsh(np.nan_to_num(_C, nan=0.0))
        LEX_PC1 = float(_ev[-1] / _ev.sum())
    except Exception:
        LEX_PC1 = float("nan")
    _prev = (counts[LEXICON] > 0).mean().sort_values()
    LEX_DEAD = [t for t, v in _prev.items() if v < 0.05]
    globals().update(LEX_ALPHA=LEX_ALPHA, LEX_MEAN_R=LEX_MEAN_R, LEX_PC1=LEX_PC1,
                     LEX_DEAD=LEX_DEAD, LEX_PREV=_prev)
    print(f"\nLexicon construct diagnostics — α={LEX_ALPHA:.3f} · mean inter-item r={LEX_MEAN_R:.3f} · PC1 share={LEX_PC1:.1%}")
    print(f"  sd(EAGI)={Z['eagi'].std():.3f}  (independence benchmark 1/sqrt(k)={1/np.sqrt(_k):.3f}; one-factor → ~1.0)")
    print(f"  term prevalence — lowest: {dict(_prev.head(4).round(3))}")
    print(f"                    highest: {dict(_prev.tail(4).round(3))}")
    if LEX_DEAD:
        print(f"  ⚠ {len(LEX_DEAD)} term(s) appear in <5% of filings and contribute noise: {LEX_DEAD}")
    if LEX_MEAN_R < 0.15:
        print("  ⚠ terms are close to uncorrelated → the composite averages idiosyncratic variation")
        print("    rather than a latent EA-governance factor. Treat downstream nulls as MEASUREMENT results.")
    pr = sum(np.log1p(counts[t] / counts["_words"] * 10000) for t in PLACEBO) / len(PLACEBO)
    Z["placebo"] = (pr - pr.mean()) / (pr.std() or 1)

    if G("panel") is None:
        df = None
        print(f"Text layer banked: {len(counts)} firm-year 10-K term counts → {tc_path} · merge deferred until panel is mapped.")
    else:
        # ---- fiscal-year label reconciliation (deterministic, no network) ----
        # SEC reportDate gives the PERIOD-END date. Compustat/SimFin label fiscal years ending
        # Jan–May to the PRIOR calendar year. Both candidate labels are computed from cached
        # submissions, and the one matching this firm's panel years is chosen per firm.
        def filing_months():
            mm = {}
            for fn in os.listdir(CACHE_DIR):
                if not (fn.startswith("sub_") and fn.endswith(".json")): continue
                try:
                    j = json.load(open(os.path.join(CACHE_DIR, fn))); ck = int(fn[4:-5])
                except Exception:
                    continue
                rec = j.get("filings", {}).get("recent", {})
                for form, fd, rd in zip(rec.get("form", []), rec.get("filingDate", []),
                                        rec.get("reportDate", rec.get("filingDate", []))):
                    if form != "10-K": continue
                    ds = rd or fd
                    if not ds or len(ds) < 7: continue
                    mm[(ck, int(ds[:4]))] = int(ds[5:7])
            return mm

        MON = filing_months()
        Za = Z[["cik", "year", "eagi", "placebo"]].copy()
        Za["pe_month"] = [MON.get((int(c), int(y)), 12) for c, y in zip(Za.cik, Za.year)]
        Za["fy_cal"] = Za["year"]
        Za["fy_comp"] = np.where(Za["pe_month"] >= 6, Za["year"], Za["year"] - 1)
        _pk = set(zip(panel.cik.astype(int), panel.year.astype(int)))
        parts, n_shift, n_firms = [], 0, 0
        for ck, grp in Za.groupby("cik"):
            n_firms += 1
            m_cal = sum((int(ck), int(y)) in _pk for y in grp.fy_cal)
            m_cmp = sum((int(ck), int(y)) in _pk for y in grp.fy_comp)
            use = "fy_comp" if m_cmp > m_cal else "fy_cal"
            if use == "fy_comp": n_shift += 1
            g2 = grp.copy(); g2["year"] = g2[use].astype(int)
            parts.append(g2)
        Za = pd.concat(parts, ignore_index=True).drop_duplicates(subset=["cik", "year"], keep="first")
        FY_SHIFT_SHARE = n_shift / max(1, n_firms)
        globals()["FY_ALIGN_NOTE"] = (f"fiscal-year labels reconciled from SEC period-end dates: "
                                      f"{n_shift}/{n_firms} firms ({FY_SHIFT_SHARE:.0%}) required the "
                                      f"Jan–May prior-year convention")
        print("  " + FY_ALIGN_NOTE)

        df_naive = panel.merge(Z[["cik", "year", "eagi"]], on=["cik", "year"], how="inner")
        df = panel.merge(Za[["cik", "year", "eagi", "placebo"]], on=["cik", "year"], how="inner")
        cov = len(df) / max(1, len(panel))
        print(df.shape, "merged ·  EAGI μ={:.2f} σ={:.2f}".format(df.eagi.mean(), df.eagi.std()))
        print(f"EAGI merge coverage: {len(df)}/{len(panel)} panel firm-years ({cov:.0%})"
              f"  ·  unaligned merge would have been {len(df_naive)} rows"
              f" (+{len(df) - len(df_naive)} recovered by reconciliation)")
        if cov < 0.5:
            print("  ⚠ LOW MERGE COVERAGE — estimation sample will be thin.")
````

````python
# ============ 4b · INSTRUMENT VARIANTS & RELIABILITY ============
# The v7.9 diagnostics said the equal-weighted, whole-document, 20-term index has no common
# factor. This cell builds and scores alternatives from the SAME cached filings — no network.
INSTR_TABLE = None
if G("counts") is None or G("CACHE_DIR") is None:
    print("SKIPPED — no term counts available.")
else:
    RF = re.compile(r"item\s*1a[\.\s:\-]{0,3}risk\s*factors")
    MD = re.compile(r"item\s*7[\.\s:\-]{0,3}management")
    END7 = re.compile(r"item\s*8[\.\s:\-]{0,3}(financial|consolidated)")
    END1A = re.compile(r"item\s*(1b|2)[\.\s:\-]")

    def narrative(txt):
        """Risk Factors + MD&A only; falls back to full text when headings are unparseable."""
        out = []
        ms = list(RF.finditer(txt))
        if ms:
            st = ms[-1].end(); en = END1A.search(txt, st)
            out.append(txt[st: en.start() if en else min(len(txt), st + 400000)])
        ms = list(MD.finditer(txt))
        if ms:
            st = ms[-1].end(); en = END7.search(txt, st)
            out.append(txt[st: en.start() if en else min(len(txt), st + 400000)])
        joined = " ".join(out)
        return (joined, True) if len(joined.split()) > 500 else (txt, False)

    sec_path = f"{CACHE_DIR}/term_counts_sections.csv"
    if os.path.exists(sec_path):
        sec = pd.read_csv(sec_path); print(f"Reusing section-scoped counts — {len(sec)} firm-years")
    else:
        recs, parsed = [], 0
        for cik in tqdm(FIRM_CIKS, desc="section scoping"):
            try: fl = list(filings_of(cik, "10-K"))
            except Exception: continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX): continue
                fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
                if not os.path.exists(fp): continue
                txt = open(fp, encoding="utf-8", errors="ignore").read()
                nar, ok = narrative(txt); parsed += int(ok)
                c = count_terms(nar, LEXICON + PLACEBO)
                c.update({"cik": cik, "year": fy, "parsed": int(ok)}); recs.append(c)
        sec = pd.DataFrame(recs).groupby(["cik", "year"], as_index=False).first()
        sec.to_csv(sec_path, index=False)
        print(f"Section scoping: {len(sec)} firm-years · headings parsed for {parsed} ({parsed/max(1,len(sec)):.0%})")

    def build_index(cnt, terms, mode="rate"):
        if mode == "binary":
            X = (cnt[terms] > 0).astype(float)
        else:
            X = np.log1p(cnt[terms].div(cnt["_words"], axis=0) * 10000)
        Xz = (X - X.mean()) / X.std().replace(0, 1)
        return Xz

    def score(Xz, label):
        k = Xz.shape[1]
        tot = Xz.sum(axis=1).var(ddof=1)
        alpha = float(k / (k - 1) * (1 - Xz.var(ddof=1).sum() / tot)) if tot else np.nan
        Cm = Xz.corr().to_numpy(); iu = np.triu_indices(k, 1)
        mr = float(np.nanmean(Cm[iu]))
        h1 = Xz.iloc[:, 0::2].mean(axis=1); h2 = Xz.iloc[:, 1::2].mean(axis=1)
        sh = float(h1.corr(h2)); sb = 2 * sh / (1 + sh) if sh == sh and sh > -1 else np.nan
        idx = Xz.mean(axis=1)
        return dict(variant=label, k=k, alpha=round(alpha, 3), mean_r=round(mr, 3),
                    split_half_r=round(sh, 3), spearman_brown=round(sb, 3), sd=round(float(idx.std()), 3)), idx, h1, h2

    prev_full = (counts[LEXICON] > 0).mean()
    live = [t for t in LEXICON if prev_full.get(t, 0) >= 0.05]
    rows, built = [], {}
    for cnt, tag in [(counts, "full-doc"), (sec, "section")]:
        for terms, tt in [(LEXICON, "all20"), (live, f"pruned{len(live)}")]:
            for mode in ("rate", "binary"):
                if len(terms) < 4: continue
                Xz = build_index(cnt, terms, mode)
                r, idx, h1, h2 = score(Xz, f"{tag}·{tt}·{mode}")
                rows.append(r)
                built[r["variant"]] = (cnt[["cik", "year"]].assign(v=idx.values, hA=h1.values, hB=h2.values))
    INSTR_TABLE = pd.DataFrame(rows).sort_values("spearman_brown", ascending=False)
    print("\nInstrument variants (higher split-half / Spearman–Brown = a real common factor):")
    print(INSTR_TABLE.to_string(index=False))
    best = INSTR_TABLE.iloc[0]["variant"]
    print(f"\nBest variant: {best}  (Spearman–Brown {INSTR_TABLE.iloc[0]['spearman_brown']})")
    if INSTR_TABLE.spearman_brown.max() < 0.5:
        print("⚠ NO variant reaches SB≥0.5 → the lexicon does not capture a single latent construct")
        print("  under ANY of these constructions. That is the measurement finding.")
    globals()["BEST_VARIANT"] = best
    if G("df") is not None:
        bv = built[best].rename(columns={"v": "eagi_best", "hA": "eagi_hA", "hB": "eagi_hB"})
        globals()["df"] = df.merge(bv, on=["cik", "year"], how="left")
        print(f"merged eagi_best/hA/hB into df ({df['eagi_best'].notna().sum()} rows)")
    try:
        INSTR_TABLE.to_csv(f"{OUT_DIR}/eagi_instrument_variants.csv", index=False)
    except Exception: pass

    # ============ FIRM-LEVEL AGGREGATION — the CIAOF analogue for measurement ============
    # CIAOF: the variation existed but the ESTIMATOR discarded it (within vs between).
    # EAGI:  the variation exists but the MEASUREMENT discards it. A term at 4% per filing
    # appears in ~25% of firms observed 7 years. Aggregating to the firm raises prevalence AND
    # matches the theory (EA governance is a firm type, not a firm-year event).
    FIRM_INSTR = None
    for cnt, tag in [(counts, "full-doc"), (sec, "section")]:
        try:
            pres = (cnt[LEXICON] > 0).astype(float)
            fp = pres.groupby(cnt["cik"]).max()            # "firm EVER uses this vocabulary"
            fr_ = pres.groupby(cnt["cik"]).mean()           # share of the firm's filings using it
            for lab, M in [(f"{tag}·firm-EVER", fp), (f"{tag}·firm-SHARE", fr_)]:
                live_t = [t for t in LEXICON if M[t].std() > 0]
                if len(live_t) < 4: continue
                Mz = (M[live_t] - M[live_t].mean()) / M[live_t].std().replace(0, 1)
                k = Mz.shape[1]; tot = Mz.sum(axis=1).var(ddof=1)
                al = float(k / (k - 1) * (1 - Mz.var(ddof=1).sum() / tot)) if tot else np.nan
                Cm = Mz.corr().to_numpy(); iu = np.triu_indices(k, 1)
                mr = float(np.nanmean(Cm[iu]))
                h1 = Mz.iloc[:, 0::2].mean(axis=1); h2 = Mz.iloc[:, 1::2].mean(axis=1)
                sh = float(h1.corr(h2)); sb = 2 * sh / (1 + sh) if sh == sh and sh > -1 else np.nan
                idx_f = Mz.mean(axis=1)
                row = dict(variant=lab, k=k, alpha=round(al, 3), mean_r=round(mr, 3),
                           split_half_r=round(sh, 3), spearman_brown=round(sb, 3),
                           sd=round(float(idx_f.std()), 3), level="firm")
                FIRM_INSTR = pd.concat([FIRM_INSTR, pd.DataFrame([row])], ignore_index=True) \
                             if FIRM_INSTR is not None else pd.DataFrame([row])
                globals()[f"EAGI_FIRM_{lab.replace('·','_').replace('-','_')}"] = idx_f
        except Exception as e:
            print("firm-level aggregation skipped for", tag, type(e).__name__)
    if FIRM_INSTR is not None:
        print("\nFIRM-LEVEL instrument variants (prevalence rises with aggregation):")
        print(FIRM_INSTR.to_string(index=False))
        INSTR_TABLE = pd.concat([INSTR_TABLE.assign(level="firm-year"), FIRM_INSTR], ignore_index=True)
        globals()["INSTR_TABLE"] = INSTR_TABLE
        best_f = FIRM_INSTR.sort_values("spearman_brown", ascending=False).iloc[0]
        print(f"  best firm-level build: {best_f['variant']} · SB={best_f['spearman_brown']}")
        if float(best_f["spearman_brown"]) >= 0.5:
            print("  ⇒ AGGREGATION RESCUES THE INSTRUMENT: report the firm-level index as primary.")
        else:
            print("  ⇒ aggregation does NOT rescue it → the null is a property of the CORPUS,")
            print("    not of the unit of analysis. This is the sentence that survives review.")
        # firm-level index onto the panel for the cross-sectional tests
        try:
            bi = globals().get("EAGI_FIRM_" + str(best_f["variant"]).replace("·", "_").replace("-", "_"))
            if bi is not None and G("df") is not None:
                globals()["df"] = df.merge(bi.rename("eagi_firm").reset_index().rename(columns={"index": "cik"}),
                                           on="cik", how="left")
                print(f"  merged firm-level index → df ({df['eagi_firm'].notna().sum()} rows)")
        except Exception as e:
            print("  firm-level merge skipped:", type(e).__name__)

    # ---- term prevalence table: which vocabulary is usable at all? ----
    prev_tbl = (pd.DataFrame({"term": LEXICON,
                              "share_filings": [(counts[t] > 0).mean() for t in LEXICON],
                              "share_firms_ever": [float((counts[[t]].assign(c=counts["cik"])
                                                          .groupby("c")[t].max() > 0).mean()) for t in LEXICON]})
                 .sort_values("share_firms_ever", ascending=False))
    globals()["PREV_TBL"] = prev_tbl
    print("\nTerm prevalence (filing level vs firm-ever level):")
    print(prev_tbl.round(3).to_string(index=False))
    try: prev_tbl.to_csv(f"{OUT_DIR}/lexicon_prevalence.csv", index=False)
    except Exception: pass

    # ---- is an alternative corpus available? (earnings transcripts are unscripted) ----
    tdirs = []
    for root, dirs, files in os.walk(os.path.dirname(os.path.dirname(CACHE_DIR))):
        if os.path.basename(root).lower() in ("transcripts", "transcript"):
            tdirs.append((root, len(files)))
    if tdirs:
        print("\nALTERNATIVE CORPUS DETECTED (10-K language is legally drafted; call Q&A is not):")
        for t, n_ in tdirs: print(f"  {t} — {n_} files")
        print("  Earnings-call Q&A is the strongest remaining corpus for this construct.")
````

````python
# ============ 4d · LEXICON CEILING — a deterministic bound, run BEFORE any LLM ============
# The same extreme-coding logic used for TechCom, applied to vocabulary coverage.
#
# ANALYTIC FACT: any multi-word term containing a root can only appear in a SUBSET of the
# documents containing that root. "architecture review board", "target architecture" and
# "architecture governance" are all subsets of documents containing "architect*".
#   LOWER BOUND = the current 20-term lexicon (what we observe)
#   UPPER BOUND = documents containing ANY domain root (the ceiling for ANY keyword method)
#
# If the ceiling is LOW, "a better keyword list exists" is excluded BY CONSTRUCTION, and no
# semantic reader can help either: an architecture-governance passage cannot be written without
# any domain root word. Deterministic, no API, no model drift — strictly better than LLM scoring
# for this specific question.
CEIL_NOTE, CEIL_TBL = [], None
ROOTS = ["architect", "modern", "legacy", "integrat", "standardi", "consolidat", "governan",
         "platform", "interoperab", "middleware", "microservic", "monolith", "refactor",
         "decommission", "rationali", "\\bapi\\b", "\\berp\\b", "migrat", "technical debt",
         "system landscape", "application portfolio", "data model"]
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    print("SKIPPED — no cached text layer.")
else:
    ceil_path = f"{CACHE_DIR}/root_counts.csv"
    rc = read_cache(ceil_path, required=["cik", "year", "_words"], label="root_counts.csv")
    if rc is not None:
        print(f"Reusing root counts — {len(rc)} firm-years")
    else:
        recs = []
        for cik in tqdm(FIRM_CIKS, desc="root ceiling"):
            try: fl = list(filings_of(cik, "10-K"))
            except Exception: continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX): continue
                fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
                if not os.path.exists(fp): continue
                t = open(fp, encoding="utf-8", errors="ignore").read()
                r = {"cik": cik, "year": fy, "_words": max(1, len(t.split()))}
                for rt in ROOTS:
                    pat = rt if rt.startswith("\\b") or " " in rt else r"\b" + rt + r"\w*"
                    r["rt_" + re.sub(r"\W", "", rt)] = len(re.findall(pat, t))
                recs.append(r)
        rc = pd.DataFrame(recs).groupby(["cik", "year"], as_index=False).first() if recs else pd.DataFrame()
        if len(rc): rc.to_csv(ceil_path, index=False)
    if len(rc):
        rtc = [c for c in rc.columns if c.startswith("rt_")]
        prev = (rc[rtc] > 0).mean().sort_values(ascending=False)
        any_root = float((rc[rtc] > 0).any(axis=1).mean())
        two_root = float(((rc[rtc] > 0).sum(axis=1) >= 2).mean())
        lex_any = float((counts[LEXICON] > 0).any(axis=1).mean()) if G("counts") is not None else float("nan")
        CEIL_TBL = prev.rename("share_of_filings").reset_index().rename(columns={"index": "root"})
        CEIL_NOTE.append("LEXICON CEILING (deterministic bound on any keyword method):")
        CEIL_NOTE.append(f"  LOWER bound — current 20-term lexicon, any term present: {lex_any:.1%} of filings")
        CEIL_NOTE.append(f"  UPPER bound — ANY domain root present:                   {any_root:.1%} of filings")
        CEIL_NOTE.append(f"  at least TWO distinct roots present:                     {two_root:.1%} of filings")
        CEIL_NOTE.append("  top roots by prevalence:")
        for _, r_ in CEIL_TBL.head(8).iterrows():
            CEIL_NOTE.append(f"    {r_['root'].replace('rt_',''):22s} {r_['share_of_filings']:6.1%}")
        # does a ROOT-based index hold together where the phrase index did not?
        Xr = np.log1p(rc[rtc].div(rc["_words"], axis=0) * 10000)
        Xz = (Xr - Xr.mean()) / Xr.std().replace(0, 1)
        k = Xz.shape[1]; tot = Xz.sum(axis=1).var(ddof=1)
        al = float(k / (k - 1) * (1 - Xz.var(ddof=1).sum() / tot)) if tot else float("nan")
        h1 = Xz.iloc[:, 0::2].mean(axis=1); h2 = Xz.iloc[:, 1::2].mean(axis=1)
        sh = float(h1.corr(h2)); sb = 2 * sh / (1 + sh) if sh == sh and sh > -1 else float("nan")
        CEIL_NOTE.append(f"  root-based index reliability: α={al:.3f} · split-half={sh:.3f} · Spearman-Brown={sb:.3f}")
        # VERDICT
        if any_root < 0.25:
            CEIL_NOTE.append("  ⇒ CEILING IS LOW. No keyword list can materially exceed this, because every")
            CEIL_NOTE.append("    phrase is a subset of its root. 'A better lexicon exists' is EXCLUDED BY")
            CEIL_NOTE.append("    CONSTRUCTION — and a semantic reader cannot help either: the passage could")
            CEIL_NOTE.append("    not be written without a domain root word. SKIP LLM SCORING; the construct")
            CEIL_NOTE.append("    is genuinely absent from 10-K narrative. This is the finding.")
        elif any_root >= 0.25 and lex_any < 0.5 * any_root:
            CEIL_NOTE.append("  ⇒ CEILING IS HIGH and our phrasing captured less than half of it. The LEXICON")
            CEIL_NOTE.append("    was too specific, not the corpus too thin. Rebuild the index on roots and")
            CEIL_NOTE.append("    re-run reliability BEFORE considering any semantic method.")
            if sb == sb and sb >= 0.5:
                CEIL_NOTE.append("    NOTE: the root index already reaches usable reliability — rebuild on it.")
        else:
            CEIL_NOTE.append("  ⇒ ceiling is moderate and largely captured by the current lexicon; keyword")
            CEIL_NOTE.append("    refinement has little headroom left.")
        CEIL_NOTE.append("  Reproducibility: fully deterministic. No API key, no model version, no drift.")
        globals()["CEIL_TBL"] = CEIL_TBL
        try: CEIL_TBL.to_csv(f"{OUT_DIR}/lexicon_ceiling.csv", index=False)
        except Exception: pass
    globals()["CEIL_NOTE"] = CEIL_NOTE
    print("\n".join(CEIL_NOTE))
````

````python
# ============ 4e · PROXIMITY LEXICON — the middle layer the ceiling result demands ============
# Cell 4d overturned the earlier conclusion: the ceiling is 100%, not thin. The construct IS in the
# corpus; the 20 exact phrases were too rigid. But the raw roots are TOO LOOSE - "consolidat" also
# matches consolidated financial statements, "governan" matches corporate governance, "integrat"
# matches integrated operations. A root index at alpha 0.58 may simply be measuring document length.
#
# The middle layer: a HEAD term (architecture / system / application / platform / data / process)
# must occur within a WINDOW of a GOVERNANCE term (governance / standard / consolidat / rationalis /
# roadmap / review board / modernis / retire / decommission). Specific enough to exclude accounting
# boilerplate, general enough to survive paraphrase. Deterministic; no API.
PROX_NOTE, PROX_TBL = [], None
HEADS = ["architect\\w*", "system\\w*", "application\\w*", "platform\\w*", "\\bdata\\b",
         "process\\w*", "interface\\w*", "integration"]
GOVS  = ["governance", "standardi\\w*", "consolidat\\w*", "rationali\\w*", "roadmap",
         "review board", "moderni\\w*", "retire\\w*", "decommission\\w*", "harmoni\\w*"]
WINDOW = 60      # characters between the two terms
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    print("SKIPPED - no cached text layer.")
else:
    prox_path = f"{CACHE_DIR}/proximity_counts.csv"
    pc = read_cache(prox_path, required=["cik", "year", "_words"], label="proximity_counts.csv")
    if pc is not None:
        print(f"Reusing proximity counts - {len(pc)} firm-years")
    else:
        pats = {}
        for h in HEADS:
            for gv in GOVS:
                key = f"{re.sub(chr(92)+'W','',h)[:10]}__{re.sub(chr(92)+'W','',gv)[:10]}"
                pats[key] = re.compile(h + r".{0," + str(WINDOW) + r"}?" + gv + r"|" +
                                       gv + r".{0," + str(WINDOW) + r"}?" + h)
        recs = []
        for cik in tqdm(FIRM_CIKS, desc="proximity lexicon"):
            try: fl = list(filings_of(cik, "10-K"))
            except Exception: continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX): continue
                fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
                if not os.path.exists(fp): continue
                t = open(fp, encoding="utf-8", errors="ignore").read()
                r = {"cik": cik, "year": fy, "_words": max(1, len(t.split()))}
                for k_, pt in pats.items(): r["px_" + k_] = len(pt.findall(t))
                recs.append(r)
        pc = pd.DataFrame(recs).groupby(["cik", "year"], as_index=False).first() if recs else pd.DataFrame()
        if len(pc): pc.to_csv(prox_path, index=False)
    if len(pc):
        cols = [c for c in pc.columns if c.startswith("px_")]
        prev = (pc[cols] > 0).mean()
        keep = [c for c in cols if 0.05 <= prev[c] <= 0.95]     # drop dead AND near-universal terms
        PROX_NOTE.append(f"PROXIMITY LEXICON: {len(cols)} head x governance pairs, "
                         f"{len(keep)} retained (prevalence 5-95%)")
        if len(keep) >= 6:
            X = np.log1p(pc[keep].div(pc["_words"], axis=0) * 10000)
            Xz = (X - X.mean()) / X.std().replace(0, 1)
            k = Xz.shape[1]; tot = Xz.sum(axis=1).var(ddof=1)
            al = float(k / (k - 1) * (1 - Xz.var(ddof=1).sum() / tot)) if tot else np.nan
            Cm = Xz.corr().to_numpy(); iu = np.triu_indices(k, 1); mr = float(np.nanmean(Cm[iu]))
            h1 = Xz.iloc[:, 0::2].mean(axis=1); h2 = Xz.iloc[:, 1::2].mean(axis=1)
            sh = float(h1.corr(h2)); sb = 2 * sh / (1 + sh) if sh == sh and sh > -1 else np.nan
            idx = Xz.mean(axis=1)
            PROX_NOTE.append(f"  reliability: alpha={al:.3f} - mean r={mr:.3f} - split-half={sh:.3f} "
                             f"- Spearman-Brown={sb:.3f} - sd={idx.std():.3f}")
            PROX_NOTE.append(f"  mean prevalence of retained pairs: {prev[keep].mean():.1%}")
            out = pc[["cik", "year"]].copy(); out["eagi_prox"] = idx.values
            if G("df") is not None:
                globals()["df"] = df.merge(out, on=["cik", "year"], how="left")
                # THE decisive check: is this measuring content, or document length?
                sub = df.dropna(subset=["eagi_prox"])
                for other, lab in [("aux_n_tokens", "filing length"), ("size", "firm size"),
                                   ("placebo", "placebo lexicon"), ("eagi", "old 20-term index")]:
                    if other in sub.columns and sub[other].notna().any():
                        PROX_NOTE.append(f"    corr(prox index, {lab}) = {sub['eagi_prox'].corr(sub[other]):+.3f}")
                _len = sub["aux_n_tokens"].corr(sub["eagi_prox"]) if "aux_n_tokens" in sub.columns else np.nan
                if sb == sb and sb >= 0.5 and (_len != _len or abs(_len) < 0.4):
                    PROX_NOTE.append("  => USABLE INSTRUMENT: reliable AND not a length proxy.")
                    PROX_NOTE.append("     Re-estimate H1/H2/H3 on eagi_prox before writing Paper 2.")
                elif _len == _len and abs(_len) >= 0.4:
                    PROX_NOTE.append("  => REJECT: correlates with filing length; it is measuring verbosity.")
                else:
                    PROX_NOTE.append("  => reliability still below 0.5; the middle layer does not rescue it either.")
            PROX_TBL = out
            try: out.to_csv(f"{OUT_DIR}/eagi_proximity_index.csv", index=False)
            except Exception: pass
        else:
            PROX_NOTE.append("  too few pairs survive the prevalence filter to build an index.")
    globals()["PROX_NOTE"] = PROX_NOTE
    print("\n".join(PROX_NOTE))
````

````python
# ============ 4c · MEASUREMENT OF LAST RESORT — semantic scoring + alternative corpus ============
# Keyword counting cannot see PARAPHRASE. A firm may describe architecture governance at length
# without using any of the twenty terms. Until that is tested, two explanations remain
# observationally equivalent:
#     (a) the construct is ABSENT from the corpus        (a finding)
#     (b) our keywords MISSED it                          (a method failure)
# Semantic scoring separates them. Labels are cached by document hash and ship as data, so the
# result is reproducible whether or not the API is reachable.
SEM_NOTE, SEM_TBL = [], None
# CONTINGENT ON CELL 4d. If the deterministic ceiling is low, a better keyword list is excluded
# by construction and a semantic reader cannot help — so this cell reports that and stops.
_ceil = G("CEIL_NOTE") or []
_low = any("CEILING IS LOW" in x for x in _ceil)
if _low:
    SEM_NOTE.append("SEMANTIC SCORING NOT REQUIRED: cell 4d found the keyword ceiling to be low, so")
    SEM_NOTE.append("  'a better lexicon exists' is already excluded deterministically. Running an LLM")
    SEM_NOTE.append("  here would add a non-reproducible dependency without changing the conclusion.")
SEM_N = 180                 # stratified subsample: enough for reliability, cheap enough to rerun
if G("counts") is None or G("CACHE_DIR") is None:
    print("SKIPPED — text layer unavailable.")
else:
    import hashlib
    # ---------- A · alternative corpus: is the vocabulary simply absent from FILINGS? ----------
    tdir = None
    for root, dirs, files in os.walk(os.path.dirname(os.path.dirname(CACHE_DIR))):
        if os.path.basename(root).lower() in ("transcripts", "transcript") and files:
            tdir = root; break
    if tdir:
        tf = [f for f in os.listdir(tdir) if f.lower().endswith((".txt", ".json", ".csv"))][:400]
        hits, tot = 0, 0
        for f in tf:
            try:
                t = open(os.path.join(tdir, f), encoding="utf-8", errors="ignore").read().lower()
            except Exception:
                continue
            tot += 1
            if any(re.search(r"\b" + re.escape(x).replace(r"\-", "[- ]") + r"\b", t) for x in LEXICON):
                hits += 1
        if tot:
            SEM_NOTE.append(f"ALTERNATIVE CORPUS ({tot} transcripts sampled from {tdir}):")
            SEM_NOTE.append(f"  documents containing ANY lexicon term: {hits}/{tot} = {hits/tot:.1%}")
            SEM_NOTE.append(f"  same statistic in 10-K filings: "
                            f"{float((counts[LEXICON] > 0).any(axis=1).mean()):.1%}")
            SEM_NOTE.append("  materially higher in transcripts ⇒ the construct exists but 10-K prose")
            SEM_NOTE.append("  suppresses it (legal drafting). Similar ⇒ genuinely absent from disclosure.")
    else:
        SEM_NOTE.append("ALTERNATIVE CORPUS: no transcripts directory found — not testable here.")

    # ---------- B · LLM semantic scoring with test-retest reliability ----------
    key = None
    if G("USE_LLM_CODING") and not _low:
        try:
            from google.colab import userdata
            key = userdata.get("ANTHROPIC_API_KEY")
        except Exception:
            key = None
    sem_path = f"{CACHE_DIR}/semantic_eagi_scores.csv"
    cache = pd.read_csv(sem_path) if os.path.exists(sem_path) else pd.DataFrame(columns=["key", "pass", "score", "evidence"])
    base = counts[["cik", "year"]].copy()
    if G("Z") is not None and "eagi" in Z.columns:
        base = base.merge(Z[["cik", "year", "eagi"]], on=["cik", "year"], how="left")
        base["strat"] = pd.qcut(base["eagi"].rank(method="first"), 3, labels=["low", "mid", "high"]).astype(str)
    else:
        base["strat"] = "all"
    parts = [g_.sample(min(SEM_N // 3, len(g_)), random_state=G("SEED") or 42)
             for _, g_ in base.groupby("strat", observed=True)]
    samp = pd.concat(parts, ignore_index=True) if parts else base.head(0)
    RUBRIC = ("You are scoring a US 10-K excerpt for evidence of ENTERPRISE ARCHITECTURE GOVERNANCE: "
              "deliberate management of the application landscape — architecture standards or review "
              "bodies, application rationalisation or consolidation, deliberate management of technical "
              "debt, integration or API standards, master data governance, or a modernisation roadmap. "
              "Judge SUBSTANCE, not vocabulary: paraphrase counts, boilerplate does not.\n"
              "Reply with exactly: SCORE=<0-10>|EVIDENCE=<=15 words, or NONE>\n"
              "0 = no evidence at all. 10 = detailed, specific, managed programme.\n\nEXCERPT:\n")
    if key and len(samp):
        import requests as _rq
        rows = []
        for _, r in tqdm(list(samp.iterrows()), desc="semantic scoring"):
            fp = None
            try:
                fl = [x for x in filings_of(int(r.cik), "10-K") if x[0] == int(r.year)]
                if fl: fp = f"{CACHE_DIR}/txt_{int(r.cik)}_{fl[0][1]}.txt"
            except Exception:
                pass
            if not fp or not os.path.exists(fp): continue
            txt = open(fp, encoding="utf-8", errors="ignore").read()
            exc = txt[:6000] + " ... " + txt[len(txt) // 2: len(txt) // 2 + 6000]
            kk = hashlib.md5(exc.encode("utf-8")).hexdigest()
            for pss in (1, 2):                     # two passes → test-retest reliability
                if ((cache["key"] == kk) & (cache["pass"] == pss)).any(): continue
                try:
                    rp = _rq.post("https://api.anthropic.com/v1/messages", timeout=60,
                                  headers={"x-api-key": key, "anthropic-version": "2023-06-01",
                                           "content-type": "application/json"},
                                  json={"model": LLM_MODEL, "max_tokens": 60, "temperature": LLM_TEMPERATURE,
                                        "messages": [{"role": "user", "content": RUBRIC + exc}]})
                    w = rp.json().get("content", [{}])[0].get("text", "")
                    sc = float(re.search(r"SCORE=(\d+(?:\.\d+)?)", w).group(1))
                    ev = (re.search(r"EVIDENCE=(.*)", w).group(1).strip()[:120]
                          if re.search(r"EVIDENCE=(.*)", w) else "")
                except Exception:
                    sc, ev = np.nan, "ERROR"
                rows.append({"key": kk, "pass": pss, "score": sc, "evidence": ev,
                             "cik": int(r.cik), "year": int(r.year)})
        if rows:
            cache = pd.concat([cache, pd.DataFrame(rows)], ignore_index=True).drop_duplicates(["key", "pass"])
            cache.to_csv(sem_path, index=False)
    if len(cache) and "score" in cache.columns and cache["score"].notna().any():
        w = cache.pivot_table(index="key", columns="pass", values="score", aggfunc="first")
        if w.shape[1] >= 2:
            tr = float(w[1].corr(w[2]))
            SEM_NOTE.append(f"SEMANTIC SCORING (n={len(w)} filings, two independent passes):")
            SEM_NOTE.append(f"  test-retest reliability r={tr:.3f} "
                            + ("(usable)" if tr >= 0.7 else "(UNSTABLE — the scorer disagrees with itself)"))
            m = w.mean(axis=1)
            SEM_NOTE.append(f"  score distribution: mean={m.mean():.2f} sd={m.std():.2f} · "
                            f"share scoring 0-2 = {(m <= 2).mean():.0%} · share ≥6 = {(m >= 6).mean():.0%}")
            if "cik" in cache.columns:
                j = cache.dropna(subset=["score"]).groupby(["cik", "year"], as_index=False)["score"].mean()
                if G("Z") is not None:
                    j = j.merge(Z[["cik", "year", "eagi"]], on=["cik", "year"], how="inner")
                    if len(j) > 30:
                        SEM_NOTE.append(f"  corr(semantic score, keyword EAGI) = {j['score'].corr(j['eagi']):+.3f}")
                        SEM_NOTE.append("    high ⇒ the keyword index was measuring the right thing badly;")
                        SEM_NOTE.append("    near zero ⇒ keywords and substance are unrelated in this corpus.")
                globals()["SEM_SCORES"] = j
                try: j.to_csv(f"{OUT_DIR}/semantic_eagi_scores_firmyear.csv", index=False)
                except Exception: pass
            if (m <= 2).mean() > 0.7:
                SEM_NOTE.append("  ⇒ VERDICT: even a semantic reader finds little or nothing to score.")
                SEM_NOTE.append("    'Keywords missed it' is now EXCLUDED. The construct is genuinely absent")
                SEM_NOTE.append("    from 10-K disclosure — this is the finding, not a method limitation.")
            elif tr >= 0.7 and (m >= 6).mean() > 0.15:
                SEM_NOTE.append("  ⇒ VERDICT: substance IS present and reliably scoreable. The keyword index")
                SEM_NOTE.append("    was the failure, not the corpus. Rebuild EAGI on semantic scores.")
    elif not key:
        SEM_NOTE.append("SEMANTIC SCORING: requires ANTHROPIC_API_KEY. Until it runs, 'absent from the")
        SEM_NOTE.append("  corpus' and 'missed by our keywords' remain observationally equivalent —")
        SEM_NOTE.append("  and that is the one open question a referee will press on Paper 2.")
    globals()["SEM_NOTE"] = SEM_NOTE
    print("\n".join(SEM_NOTE))
````

````python
# ============ 5 · TECHCOM FROM DEF 14A (firm-level fault isolation) ============
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    tc = None
    print("UPSTREAM INCOMPLETE — TechCom skipped. See the RUN DIAGNOSTIC cell.")
else:
    TC_PATT = re.compile(r"(technology|digital|innovation)\s+committee(\s+of\s+the\s+board)?")
    tcm_path = f"{CACHE_DIR}/techcom_raw.csv"
    tc = read_cache(tcm_path, required=["cik", "year", "techcom"], label="techcom_raw.csv")
    if tc is not None:
        print("Reusing cached techcom_raw.csv —", len(tc), "firm-years")
    else:
        recs = []
        for cik in tqdm(FIRM_CIKS, desc="DEF 14A"):
            try:
                fl = list(filings_of(cik, "DEF 14A"))
            except Exception as e:
                print("skip firm", cik, type(e).__name__); continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX):
                    continue
                try:
                    hit = int(bool(TC_PATT.search(doc_text(cik, acc, doc))))
                    recs.append({"cik": cik, "year": fy, "techcom": hit})
                except Exception as e:
                    print("skip doc", cik, fy, type(e).__name__)
        tc = pd.DataFrame(recs, columns=["cik", "year", "techcom"]).groupby(["cik", "year"], as_index=False).max() if recs else pd.DataFrame(columns=["cik", "year", "techcom"])
        tc.to_csv(tcm_path, index=False)
        print("NOTE: manually verify every positive before inference (impression-management risk).")

    if G("df") is None:
        print(f"TechCom layer banked: {len(tc)} firm-years → {tcm_path}")
        print("\nPREFETCH COMPLETE. Map the panel, re-run — full analysis will finish from cache in minutes.")
    else:
        df = df.merge(tc, on=["cik", "year"], how="left")
        df["techcom"] = df["techcom"].fillna(0).astype(int)
        print("TechCom share = {:.1%}".format(df.techcom.mean()))
````

````python
# ============ 5b · CONVERGENT VALIDITY — EAGI vs adjacent in-panel signals ============
VALIDITY_TBL = None
if G("df") is None:
    print("SKIPPED — no merged panel.")
else:
    aux = [c for c in df.columns if str(c).startswith("aux_")]
    if not aux:
        print("No adjacent signals present in the panel — validity check not applicable.")
    else:
        rows = []
        for a in aux:
            sub = df[["cik", "eagi", a]].dropna()
            if len(sub) < 30: continue
            if sub[a].std(ddof=0) < 1e-12 or sub["eagi"].std(ddof=0) < 1e-12:
                print(f"  {a.replace('aux_','')}: zero variance in the overlapping sample — skipped"); continue
            w = sub.copy()
            for c in ("eagi", a):
                w[c] = w[c] - w.groupby("cik")[c].transform("mean")   # within-firm: what FE actually uses
            rows.append(dict(signal=a.replace("aux_", ""), n=len(sub),
                             pearson=round(float(sub["eagi"].corr(sub[a])), 3),
                             spearman=round(float(sub["eagi"].corr(sub[a], method="spearman")), 3),
                             within_firm=round(float(w["eagi"].corr(w[a])), 3)))
        if rows:
            VALIDITY_TBL = pd.DataFrame(rows).sort_values("pearson", key=abs, ascending=False)
            print("Convergent validity — EAGI vs adjacent panel signals:")
            print(VALIDITY_TBL.to_string(index=False))
            print("\nRead: 'within_firm' is the correlation the FE estimator actually exploits.")
            print("Sizeable |r| with gov_it → convergent validity for EAGI; near-zero → the two indices")
            print("are measuring different constructs, which must be stated rather than assumed away.")
        else:
            print("Adjacent signals present but too sparse to correlate (n<30).")
````

````python
# ============ 5e · MANDATORY-DISCLOSURE INSTRUMENTS ============
# Voluntary narrative failed (0.7% prevalence). Mandatory disclosure is a different object:
# firms must file it, auditors attest parts of it, and it exists for every firm every year.
#
#   (1) ITGC MATERIAL WEAKNESS — SOX 404 / Item 9A. AUDITOR-ATTESTED. Access management, change
#       control and segregation of duties in ERP systems. Not rhetoric: a failure someone signed.
#       This INVERTS the measurement problem — governance FAILURE is far more observable than
#       governance presence, because failure is what regulation compels you to report.
#   (2) ITEM 106 CYBER GOVERNANCE — Reg S-K, FY2023 onward. Mandatory: which board committee
#       oversees, how often, and management's expertise. Required of every registrant.
#
# All extraction is deterministic and runs from cached filings. No API, no model drift.
MAND_NOTE, MAND_TBL = [], None
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    print("SKIPPED — no cached text layer.")
else:
    IT_CUES = ["information technology", "it general control", "itgc", "user access", "access control",
               "change management", "segregation of duties", "logical access", "system access",
               "program change", "erp", "information system"]
    MW = re.compile(r"material weakness")
    # negation and affirmation cues, scored on the immediate context of each match
    NEG_PAT = re.compile(r"(no|not|never|without|nor)\s+(?:\w+\s+){0,4}$|did not identify\s*(?:\w+\s+){0,3}$|"
                         r"were not aware of\s*(?:\w+\s+){0,3}$|absence of\s*(?:\w+\s+){0,2}$|"
                         r"free (?:from|of)\s*(?:\w+\s+){0,3}$")
    NEG_RIGHT = re.compile(r"^(?:es)?\s*(?:were|was|has|have)\s+not\s+(?:been\s+)?identified|"
                           r"^(?:es)?\s+(?:did not|do not|does not)")
    AFF_PAT = re.compile(r"(identified|disclosed|reported|concluded that (?:a|the)|remediat\w+|"
                         r"following|previously reported|existence of|resulted in)\s*(?:\w+\s+){0,4}$")
    AFF_RIGHT = re.compile(r"^(?:es)?\s+(?:in|relating to|related to|with respect to|described|"
                           r"associated with|that|existed|remains|was identified|were identified)\b")
    NOTEFF = re.compile(r"internal control over financial reporting (?:was|were) not effective")
    I106 = re.compile(r"item\s*1c[\.\s:\-]|cybersecurity risk management, strategy")
    BOARD_OV = re.compile(r"(audit|technology|risk|cybersecurity)\s+committee")
    MGMT_EXP = re.compile(r"(chief information security officer|ciso|years of experience|prior experience)")
    mand_path = f"{CACHE_DIR}/mandatory_signals.csv"
    md = read_cache(mand_path, required=["cik", "year", "mw_it", "mw_it_aff", "mw_it_incl",
                                         "icfr_not_effective", "item106_present"],
                    label="mandatory_signals.csv")
    if md is not None:
        print(f"Reusing mandatory signals — {len(md)} firm-years")
    else:
        recs = []
        for cik in tqdm(FIRM_CIKS, desc="mandatory disclosure"):
            try: fl = list(filings_of(cik, "10-K"))
            except Exception: continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX): continue
                fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
                if not os.path.exists(fp): continue
                t = open(fp, encoding="utf-8", errors="ignore").read()
                # NEGATION HANDLING. "material weakness" occurs in ~99% of filings, almost always
                # in a negative attestation ("no material weakness was identified"). Classifying on
                # proximity alone would count those as positives. Each match is therefore scored
                # against negation and affirmation cues in its immediate left context, and the
                # residue is bounded both ways rather than adjudicated by hand.
                mw_any, mw_it, mw_it_aff, mw_it_amb = 0, 0, 0, 0
                for m in MW.finditer(t):
                    mw_any = 1
                    w = t[max(0, m.start() - 600): m.end() + 600]
                    if not any(c in w for c in IT_CUES):
                        continue
                    mw_it = 1                                    # proximity only (legacy definition)
                    left = t[max(0, m.start() - 220): m.start()]
                    right = t[m.end(): m.end() + 220]
                    neg = bool(NEG_PAT.search(left)) or bool(NEG_RIGHT.search(right))
                    aff = bool(AFF_PAT.search(left)) or bool(AFF_RIGHT.search(right))
                    # PRECEDENCE: a negation cue suppresses the match outright, even when an
                    # affirmation cue co-occurs. "no material weakness ... in our IT systems" carries
                    # both, and the negation is decisive. Neither counter is incremented, so the
                    # match is excluded rather than pushed into the ambiguous residue.
                    if aff and not neg: mw_it_aff = 1
                    elif not aff and not neg: mw_it_amb = 1
                rec = {"cik": cik, "year": fy,
                       "mw_any": mw_any, "mw_it": mw_it,
                       "mw_it_aff": mw_it_aff, "mw_it_amb": mw_it_amb,
                       "mw_it_incl": int(bool(mw_it_aff or mw_it_amb)),
                       "icfr_not_effective": int(bool(NOTEFF.search(t))),
                       "item106_present": int(bool(I106.search(t)))}
                if rec["item106_present"]:
                    j = I106.search(t).start()
                    seg = t[j: j + 20000]
                    rec["cyber_board_committee"] = int(bool(BOARD_OV.search(seg)))
                    rec["cyber_mgmt_expertise"] = int(bool(MGMT_EXP.search(seg)))
                else:
                    rec["cyber_board_committee"] = 0; rec["cyber_mgmt_expertise"] = 0
                recs.append(rec)
        md = pd.DataFrame(recs).groupby(["cik", "year"], as_index=False).max() if recs else pd.DataFrame()
        if len(md): md.to_csv(mand_path, index=False)
    if len(md):
        MAND_NOTE.append("MANDATORY-DISCLOSURE PREVALENCE (contrast with 0.7% for the voluntary lexicon):")
        for c in ["mw_any", "mw_it", "mw_it_aff", "mw_it_incl", "icfr_not_effective", "item106_present",
                  "cyber_board_committee", "cyber_mgmt_expertise"]:
            if c in md.columns:
                MAND_NOTE.append(f"    {c:24s} {md[c].mean():6.1%} of firm-years · "
                                 f"{md.groupby('cik')[c].max().mean():6.1%} of firms ever")
        post = md[md.year >= 2023]
        if len(post):
            MAND_NOTE.append(f"  Item 106 (mandatory FY2023+): present in {post['item106_present'].mean():.1%} "
                             f"of post-2023 filings — a required disclosure, unlike the lexicon.")
        if G("df") is not None:
            keep = [c for c in md.columns if c not in ("cik", "year")]
            globals()["df"] = df.merge(md, on=["cik", "year"], how="left")
            for c in keep: df[c] = df[c].fillna(0)
            # governance FAILURE as the instrument: does an audited ITGC weakness predict outcomes?
            try:
                import statsmodels.api as sm
                W = df.sort_values(["cik", "year"]).copy()
                lgm = W[["cik", "year", "mw_it", "mw_any"]].copy(); lgm["year"] += 1
                W = W.merge(lgm.rename(columns={"mw_it": "mw_it_L", "mw_any": "mw_any_L"}),
                            on=["cik", "year"], how="left")
                ctl = [c for c in ["size", "lev", "rnd"] if c in W.columns]
                # EXTREME-CODING BOUNDS on the ambiguous residue: lower bound counts only
                # affirmed matches, upper bound counts affirmed plus ambiguous. If both bounds agree,
                # the residue cannot change the conclusion and no hand adjudication is required.
                lgm2 = W[["cik", "year", "mw_it_aff", "mw_it_incl"]].copy(); lgm2["year"] += 1
                W = W.merge(lgm2.rename(columns={"mw_it_aff": "mw_aff_L", "mw_it_incl": "mw_incl_L"}),
                            on=["cik", "year"], how="left")
                MAND_NOTE.append("  NEGATION-AWARE ITGC (bounds on the ambiguous residue):")
                for xv in ["mw_it_L", "mw_aff_L", "mw_incl_L", "mw_any_L"]:
                    S = W.dropna(subset=[xv, "roa"] + ctl)
                    if len(S) < 80 or S[xv].std() == 0: continue
                    X = S[[xv] + ctl].astype(float)
                    X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y",
                                                     drop_first=True, dtype=float)], axis=1)
                    if "industry_raw" in S.columns and S["industry_raw"].nunique() > 1:
                        X = pd.concat([X, pd.get_dummies(S["industry_raw"].astype(str), prefix="i",
                                                         drop_first=True, dtype=float)], axis=1)
                    r_ = sm.OLS(S["roa"].values, sm.add_constant(X)).fit(
                        cov_type="cluster", cov_kwds={"groups": S["cik"].values})
                    _lbl = {"mw_it_L": "proximity only (legacy)", "mw_aff_L": "LOWER bound: affirmed only",
                            "mw_incl_L": "UPPER bound: affirmed + ambiguous",
                            "mw_any_L": "unrestricted (boilerplate control)"}.get(xv, xv)
                    MAND_NOTE.append(f"    {_lbl:36s} β={r_.params[xv]:+.4f} p={r_.pvalues[xv]:.3f} "
                                     f"n={len(S)} · treated={S[xv].mean():.1%}")
                try:
                    _lo = S2 if False else None
                    _b1 = [x for x in MAND_NOTE if "LOWER bound" in x]
                    _b2 = [x for x in MAND_NOTE if "UPPER bound" in x]
                    if _b1 and _b2:
                        _p1 = float(re.search(r"p=([\d.]+)", _b1[-1]).group(1))
                        _p2 = float(re.search(r"p=([\d.]+)", _b2[-1]).group(1))
                        _s1 = float(re.search(r"β=([-+][\d.]+)", _b1[-1]).group(1))
                        _s2 = float(re.search(r"β=([-+][\d.]+)", _b2[-1]).group(1))
                        if (_p1 < .05) == (_p2 < .05) and (_s1 > 0) == (_s2 > 0):
                            MAND_NOTE.append("    ⇒ BOUNDS AGREE: the ambiguous residue cannot change the")
                            MAND_NOTE.append("      conclusion. No hand adjudication is required.")
                        else:
                            MAND_NOTE.append("    ⇒ BOUNDS DISAGREE: the ambiguous residue is decisive.")
                            MAND_NOTE.append("      Code techcom-style gold standard for these matches before publishing.")
                except Exception:
                    pass
                # convergent check against an independently-worded mandatory signal
                if "icfr_not_effective" in df.columns and df["mw_it_aff"].std() > 0:
                    MAND_NOTE.append(f"    convergent: corr(affirmed ITGC, ICFR-not-effective) = "
                                     f"{df['mw_it_aff'].corr(df['icfr_not_effective']):+.3f} "
                                     "(independent wording, same underlying failure)")
                MAND_NOTE.append("  NEGATIVE β on an ITGC weakness is the predicted sign: audited control")
                MAND_NOTE.append("  failure in the system estate is governance failure made observable.")
                if "eagi_best" in df.columns:
                    cc = df.dropna(subset=["eagi_best", "mw_it"])
                    if len(cc) > 50:
                        MAND_NOTE.append(f"  corr(text EAGI, ITGC weakness) = {cc['eagi_best'].corr(cc['mw_it']):+.3f}"
                                         " — near zero would mean the two measure different things entirely.")
            except Exception as e:
                MAND_NOTE.append("outcome test unavailable: " + type(e).__name__)
        # ---------- (3) 8-K ITEM CODES — zero network: the submissions JSON already lists them ----------
        # Item 1.05 = material cybersecurity incident (mandatory FY2024+, tiny base rate).
        # Item 4.02 = non-reliance on previously issued financials, i.e. a RESTATEMENT. Available
        # across the whole window, higher base rate, and a long-established measure of control
        # quality strongly linked to IT general-control failure — the same construct through a
        # different mandatory channel, and dated, so it supports event-time analysis.
        try:
            # SELF-HEALING: if a cached submissions file predates the 'items' field, refresh it
            # automatically. One cheap metadata call per affected firm — never a manual step.
            stale = []
            for fn in sorted(os.listdir(CACHE_DIR)):
                if not (fn.startswith("sub_") and fn.endswith(".json")): continue
                try:
                    jj = json.load(open(os.path.join(CACHE_DIR, fn)))
                    if not jj.get("filings", {}).get("recent", {}).get("items"):
                        stale.append(int(fn[4:-5]))
                except Exception:
                    stale.append(int(fn[4:-5]))
            if stale:
                print(f"  refreshing {len(stale)} cached submissions missing the 'items' field …")
                for ck in tqdm(stale, desc="refresh submissions"):
                    try:
                        os.remove(os.path.join(CACHE_DIR, f"sub_{ck}.json")); submissions(ck)
                    except Exception:
                        pass
            ev = []
            for fn in sorted(os.listdir(CACHE_DIR)):
                if not (fn.startswith("sub_") and fn.endswith(".json")): continue
                try:
                    j = json.load(open(os.path.join(CACHE_DIR, fn))); ck = int(fn[4:-5])
                except Exception:
                    continue
                rec = j.get("filings", {}).get("recent", {})
                forms = rec.get("form", []); dates = rec.get("filingDate", []); items = rec.get("items", [])
                if not items: continue                      # older cache format: field absent
                for f_, d_, it_ in zip(forms, dates, items):
                    if f_ != "8-K" or not d_: continue
                    yy = int(d_[:4])
                    if not (YEAR_MIN <= yy <= YEAR_MAX): continue
                    codes = [x.strip() for x in str(it_).split(",") if x.strip()]
                    ev.append({"cik": ck, "year": yy,
                               "ev_cyber_105": int(any(c.startswith("1.05") for c in codes)),
                               "ev_restate_402": int(any(c.startswith("4.02") for c in codes)),
                               "ev_impair_206": int(any(c.startswith("2.06") for c in codes)),
                               "ev_8k_count": 1})
            if ev:
                E = pd.DataFrame(ev).groupby(["cik", "year"], as_index=False).agg(
                    ev_cyber_105=("ev_cyber_105", "max"), ev_restate_402=("ev_restate_402", "max"),
                    ev_impair_206=("ev_impair_206", "max"), ev_8k_count=("ev_8k_count", "sum"))
                MAND_NOTE.append("8-K ITEM CODES (parsed from cached submissions — no network):")
                MAND_NOTE.append(f"    Item 1.05 cyber incident   {E['ev_cyber_105'].sum():4d} firm-years · "
                                 f"{E.groupby('cik')['ev_cyber_105'].max().sum():3d} firms "
                                 "(mandatory FY2024+ only — descriptive, not a test)")
                MAND_NOTE.append(f"    Item 4.02 RESTATEMENT      {E['ev_restate_402'].sum():4d} firm-years · "
                                 f"{E.groupby('cik')['ev_restate_402'].max().sum():3d} firms "
                                 "(full window — the usable one)")
                MAND_NOTE.append(f"    Item 2.06 impairment       {E['ev_impair_206'].sum():4d} firm-years")
                MAND_NOTE.append(f"    median 8-K filings/firm-year: {E['ev_8k_count'].median():.0f}")
                if G("df") is not None:
                    globals()["df"] = df.merge(E, on=["cik", "year"], how="left")
                    for c in ["ev_cyber_105", "ev_restate_402", "ev_impair_206", "ev_8k_count"]:
                        df[c] = df[c].fillna(0)
                    try:
                        import statsmodels.api as sm
                        W2 = df.sort_values(["cik", "year"]).copy()
                        l4 = W2[["cik", "year", "ev_restate_402"]].copy(); l4["year"] += 1
                        W2 = W2.merge(l4.rename(columns={"ev_restate_402": "restate_L"}),
                                      on=["cik", "year"], how="left")
                        ctl2 = [c for c in ["size", "lev", "rnd"] if c in W2.columns]
                        S2 = W2.dropna(subset=["restate_L", "roa"] + ctl2)
                        if len(S2) > 80 and S2["restate_L"].std() > 0:
                            X2 = S2[["restate_L"] + ctl2].astype(float)
                            X2 = pd.concat([X2, pd.get_dummies(S2["year"].astype(str), prefix="y",
                                                               drop_first=True, dtype=float)], axis=1)
                            if "industry_raw" in S2.columns and S2["industry_raw"].nunique() > 1:
                                X2 = pd.concat([X2, pd.get_dummies(S2["industry_raw"].astype(str), prefix="i",
                                                                   drop_first=True, dtype=float)], axis=1)
                            rr = sm.OLS(S2["roa"].values, sm.add_constant(X2)).fit(
                                cov_type="cluster", cov_kwds={"groups": S2["cik"].values})
                            _tr = float(S2["restate_L"].mean())
                            MAND_NOTE.append(f"    restatement(t-1) → ROA: β={rr.params['restate_L']:+.4f} "
                                             f"p={rr.pvalues['restate_L']:.3f} n={len(S2)} · "
                                             f"treated={_tr:.1%}")
                            if _tr < 0.01:
                                MAND_NOTE.append("      ⚠ SUPPRESSED: treated share under 1% — this estimate")
                                MAND_NOTE.append("        rests on a handful of firm-years. Do not report it, and")
                                MAND_NOTE.append("        do not report any correlation computed against it.")
                        if ("mw_it" in df.columns and df["ev_restate_402"].std() > 0
                                and df["ev_restate_402"].mean() >= 0.01):
                            MAND_NOTE.append(f"    corr(ITGC weakness, restatement) = "
                                             f"{df['mw_it'].corr(df['ev_restate_402']):+.3f} — two mandatory")
                            MAND_NOTE.append("      channels on the same construct; convergence supports both.")
                    except Exception as e2:
                        MAND_NOTE.append("    restatement test unavailable: " + type(e2).__name__)
            else:
                MAND_NOTE.append("8-K ITEM CODES: no in-window 8-K filings carried item codes after refresh.")
        except Exception as e:
            MAND_NOTE.append("8-K item parsing unavailable: " + type(e).__name__)

        MAND_NOTE.append("  POWER CAVEAT: ITGC weaknesses are rare in large-cap panels. A low treated share")
        MAND_NOTE.append("  means limited power — report the share alongside any null so the reader can judge.")
        MAND_TBL = md
        try: md.to_csv(f"{OUT_DIR}/mandatory_signals.csv", index=False)
        except Exception: pass
    globals()["MAND_NOTE"] = MAND_NOTE
    print("\n".join(MAND_NOTE))
````

````python
# ============ 5f · FILING-METADATA INSTRUMENTS — free, mandatory, high base rate ============
# Everything here is already inside the cached submissions JSON. No new documents, no parsing of
# prose, no discretion. The strongest item is FILING LAG: days from period end to filing.
# Closing the books fast is downstream of system quality — it is continuous, universal, and
# recorded for every firm every year, which is exactly what the text index was not.
META_NOTE, META_TBL = [], None
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    print("SKIPPED — no cached submissions.")
else:
    rows, prof = [], []
    for ck in FIRM_CIKS:
        fp = f"{CACHE_DIR}/sub_{ck}.json"
        if not os.path.exists(fp): continue
        try:
            j = json.load(open(fp))
        except Exception:
            continue
        prof.append({"cik": ck,
                     "sic": str(j.get("sic") or ""), "sic_desc": str(j.get("sicDescription") or ""),
                     "fye": str(j.get("fiscalYearEnd") or ""),
                     "filer_category": str(j.get("category") or ""),
                     "state_inc": str(j.get("stateOfIncorporation") or ""),
                     "n_former_names": len(j.get("formerNames") or [])})
        rec = j.get("filings", {}).get("recent", {})
        forms = rec.get("form", []); fdt = rec.get("filingDate", []); rdt = rec.get("reportDate", fdt)
        for f_, d_, r_ in zip(forms, fdt, rdt):
            if not d_: continue
            yy = int(d_[:4])
            if not (YEAR_MIN - 1 <= yy <= YEAR_MAX): continue
            base = {"cik": ck, "form": f_, "fdate": d_, "rdate": r_}
            rows.append(base)
    if rows:
        R = pd.DataFrame(rows)
        R["fdate"] = pd.to_datetime(R["fdate"], errors="coerce")
        R["rdate"] = pd.to_datetime(R["rdate"], errors="coerce")
        ann = R[R["form"] == "10-K"].dropna(subset=["fdate", "rdate"]).copy()
        ann["year"] = ann["rdate"].dt.year
        ann["filing_lag"] = (ann["fdate"] - ann["rdate"]).dt.days
        ann = ann[(ann["filing_lag"] >= 0) & (ann["filing_lag"] <= 400)]
        LAG = ann.groupby(["cik", "year"], as_index=False)["filing_lag"].min()
        late = R[R["form"].str.startswith("NT 10-K", na=False)].copy()
        late["year"] = late["fdate"].dt.year
        LATE = late.groupby(["cik", "year"], as_index=False).size().rename(columns={"size": "nt_10k"})
        amd = R[R["form"] == "10-K/A"].copy(); amd["year"] = amd["fdate"].dt.year
        AMD = amd.groupby(["cik", "year"], as_index=False).size().rename(columns={"size": "amended_10k"})
        M = LAG.merge(LATE, on=["cik", "year"], how="left").merge(AMD, on=["cik", "year"], how="left")
        M[["nt_10k", "amended_10k"]] = M[["nt_10k", "amended_10k"]].fillna(0)
        P = pd.DataFrame(prof).drop_duplicates("cik")
        P["sic2"] = P["sic"].str[:2]
        META_TBL = M.merge(P, on="cik", how="left")
        META_NOTE.append("FILING METADATA (all mandatory, all already cached):")
        META_NOTE.append(f"  filing lag (days, period end → filing): median={M['filing_lag'].median():.0f} · "
                         f"p10={M['filing_lag'].quantile(.1):.0f} · p90={M['filing_lag'].quantile(.9):.0f} · "
                         f"n={len(M)}")
        META_NOTE.append(f"  NT 10-K late notifications: {int(M['nt_10k'].sum())} firm-years · "
                         f"{int((M.groupby('cik')['nt_10k'].max() > 0).sum())} firms")
        META_NOTE.append(f"  10-K/A amendments:          {int(M['amended_10k'].sum())} firm-years · "
                         f"{int((M.groupby('cik')['amended_10k'].max() > 0).sum())} firms")
        META_NOTE.append(f"  distinct 4-digit SIC codes: {P['sic'].nunique()} · 2-digit groups: {P['sic2'].nunique()}"
                         + (f" (panel 'industry_raw' has {df['industry_raw'].nunique()})"
                            if G("df") is not None and "industry_raw" in df.columns else ""))
        META_NOTE.append(f"  filer categories: {dict(P['filer_category'].value_counts().head(4))}")
        META_NOTE.append(f"  fiscal year ends: {P['fye'].nunique()} distinct — the FILED value, replacing the")
        META_NOTE.append("    month heuristic used in the fiscal-year reconciliation.")
        if G("df") is not None:
            globals()["df"] = df.merge(META_TBL, on=["cik", "year"], how="left")
            for c in ["nt_10k", "amended_10k"]:
                if c in df.columns: df[c] = df[c].fillna(0)
            try:
                import statsmodels.api as sm
                W = df.sort_values(["cik", "year"]).copy()
                lgl = W[["cik", "year", "filing_lag"]].copy(); lgl["year"] += 1
                W = W.merge(lgl.rename(columns={"filing_lag": "lag_L"}), on=["cik", "year"], how="left")
                ctl = [c for c in ["size", "lev", "rnd"] if c in W.columns]
                S = W.dropna(subset=["lag_L", "roa"] + ctl)
                if len(S) > 80 and S["lag_L"].std() > 0:
                    X = S[["lag_L"] + ctl].astype(float)
                    X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y",
                                                     drop_first=True, dtype=float)], axis=1)
                    icol = "sic2" if "sic2" in S.columns and S["sic2"].nunique() > 3 else "industry_raw"
                    if icol in S.columns and S[icol].nunique() > 1:
                        X = pd.concat([X, pd.get_dummies(S[icol].astype(str), prefix="i",
                                                         drop_first=True, dtype=float)], axis=1)
                    r_ = sm.OLS(S["roa"].values, sm.add_constant(X)).fit(
                        cov_type="cluster", cov_kwds={"groups": S["cik"].values})
                    META_NOTE.append(f"  filing lag(t−1) → ROA (FE on {icol}): β={r_.params['lag_L']:+.5f} "
                                     f"p={r_.pvalues['lag_L']:.3f} n={len(S)}")
                    META_NOTE.append("    NEGATIVE ⇒ slower close predicts weaker performance. Interpret as")
                    META_NOTE.append("    operational capability, not architecture per se — the two overlap,")
                    META_NOTE.append("    and lag also reflects auditor and complexity effects.")
                for c in ["mw_it", "ev_restate_402"]:
                    if c in df.columns and df[c].std() > 0 and df["filing_lag"].notna().any():
                        META_NOTE.append(f"    corr(filing lag, {c}) = {df['filing_lag'].corr(df[c]):+.3f}")
            except Exception as e:
                META_NOTE.append("  outcome test unavailable: " + type(e).__name__)
        try: META_TBL.to_csv(f"{OUT_DIR}/filing_metadata.csv", index=False)
        except Exception: pass
    globals()["META_NOTE"] = META_NOTE
    print("\n".join(META_NOTE) if META_NOTE else "no filing metadata recovered")
````

````python
# ============ 5g · CUSTOMISATION VOLUME (XBRL capitalised software) ============
# The 2x2 turns on HOW MUCH a firm customises. The panel measures only how CLEANLY extensions
# are built (adherence), never the volume - so its central question has been untestable.
# The closest public measure is CAPITALISED INTERNAL-USE SOFTWARE (ASC 350-40): money spent
# building software rather than buying it. It is a MANDATORY disclosure when material, which is
# exactly where the protocol says to look. capsoft_intensity exists in the panel at only 13%
# coverage; XBRL company facts carry the same concept for far more firms.
CUST_NOTE, CUST_TBL = [], None
SOFT_TAGS = ["CapitalizedComputerSoftwareNet", "CapitalizedComputerSoftwareGross",
             "CapitalizedComputerSoftwareAdditions", "PaymentsToDevelopSoftware",
             "PaymentsForSoftware", "CapitalizedSoftwareDevelopmentCosts"]
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    print("SKIPPED - no cache.")
else:
    # is there a usable local xbrl store already?
    xdirs = []
    for root, dirs, files in os.walk(os.path.dirname(os.path.dirname(CACHE_DIR))):
        if os.path.basename(root).lower() == "xbrl" and files:
            xdirs.append((root, len(files)))
    if xdirs:
        CUST_NOTE.append(f"local xbrl store found: {xdirs[0][0]} ({xdirs[0][1]} files) - inspect its")
        CUST_NOTE.append("  layout to wire it in; the SEC companyfacts route below runs regardless.")
    cf_path = f"{CACHE_DIR}/capsoft_xbrl.csv"
    cx = read_cache(cf_path, required=["cik", "year", "tag", "val"], label="capsoft_xbrl.csv")
    if cx is not None:
        print(f"Reusing XBRL software facts - {len(cx)} firm-years")
    else:
        recs = []
        for cik in tqdm(FIRM_CIKS, desc="XBRL companyfacts"):
            fp = f"{CACHE_DIR}/facts_{cik}.json"
            try:
                if os.path.exists(fp):
                    j = json.load(open(fp))
                else:
                    j = json.loads(_get(f"https://data.sec.gov/api/xbrl/companyfacts/CIK{int(cik):010d}.json"))
                    json.dump({"facts": {"us-gaap": {k: v for k, v in
                               j.get("facts", {}).get("us-gaap", {}).items() if k in SOFT_TAGS}}},
                              open(fp, "w"))     # store only what we need
            except Exception:
                continue
            ug = j.get("facts", {}).get("us-gaap", {})
            for tag in SOFT_TAGS:
                for unit, rows in ug.get(tag, {}).get("units", {}).items():
                    for r in rows:
                        fy = r.get("fy")
                        if fy and YEAR_MIN <= int(fy) <= YEAR_MAX and r.get("form", "").startswith("10-K"):
                            recs.append({"cik": cik, "year": int(fy), "tag": tag, "val": float(r.get("val", 0))})
        cx = (pd.DataFrame(recs).groupby(["cik", "year", "tag"], as_index=False)["val"].max()
              if recs else pd.DataFrame())
        if len(cx): cx.to_csv(cf_path, index=False)
    if len(cx):
        wide = cx.pivot_table(index=["cik", "year"], columns="tag", values="val", aggfunc="max").reset_index()
        have = [t for t in SOFT_TAGS if t in wide.columns]
        wide["capsoft_xbrl"] = wide[have].max(axis=1) if have else np.nan
        CUST_NOTE.append("XBRL capitalised-software coverage:")
        for t in have:
            CUST_NOTE.append(f"    {t:38s} {wide[t].notna().mean():5.1%} of matched firm-years")
        CUST_NOTE.append(f"    ANY tag present: {wide['capsoft_xbrl'].notna().sum()} firm-years · "
                         f"{wide.loc[wide['capsoft_xbrl'].notna(), 'cik'].nunique()} firms")
        if G("df") is not None:
            D2 = df.merge(wide[["cik", "year", "capsoft_xbrl"]], on=["cik", "year"], how="left")
            globals()["df"] = D2
            base_cov = float(df["capsoft"].notna().mean()) if "capsoft" in df.columns else 0.0
            new_cov = float(D2["capsoft_xbrl"].notna().mean())
            CUST_NOTE.append(f"    coverage vs the panel's capsoft_intensity: {base_cov:.0%} → {new_cov:.0%}")
            if new_cov >= 0.30:
                import statsmodels.api as sm
                S = D2.dropna(subset=["capsoft_xbrl", "roa"]).copy()
                # scale by assets when available so it is an INTENSITY, not a level
                if "size" in S.columns:
                    S["cust_int"] = S["capsoft_xbrl"] / np.exp(S["size"]).replace(0, np.nan)
                else:
                    S["cust_int"] = S["capsoft_xbrl"]
                S = S.replace([np.inf, -np.inf], np.nan).dropna(subset=["cust_int"])
                S["cust_int"] = S["cust_int"].clip(S["cust_int"].quantile(.01), S["cust_int"].quantile(.99))
                ctl2 = [c for c in ["size", "lev", "rnd"] if c in S.columns]
                if len(S) > 80:
                    X = S[["cust_int"] + ctl2].astype(float)
                    X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y",
                                                     drop_first=True, dtype=float)], axis=1)
                    ic = "sic2" if "sic2" in S.columns and S["sic2"].nunique() > 3 else "industry_raw"
                    if ic in S.columns and S[ic].nunique() > 1:
                        X = pd.concat([X, pd.get_dummies(S[ic].astype(str), prefix="i",
                                                         drop_first=True, dtype=float)], axis=1)
                    r_ = sm.OLS(S["roa"].values, sm.add_constant(X)).fit(
                        cov_type="cluster", cov_kwds={"groups": S["cik"].values})
                    CUST_NOTE.append(f"  CUSTOMISATION VOLUME → ROA: β={r_.params['cust_int']:+.4f} "
                                     f"p={r_.pvalues['cust_int']:.3f} n={len(S)}")
                    # THE 2x2's core claim: volume should pay where differentiation potential is high
                    if "rnd" in S.columns:
                        S["dif"] = S.groupby("cik")["rnd"].transform("mean")
                        S["ixn"] = (S["cust_int"] - S["cust_int"].mean()) * (S["dif"] - S["dif"].mean())
                        X2 = pd.concat([X, S[["dif", "ixn"]].astype(float)], axis=1)
                        r2_ = sm.OLS(S["roa"].values, sm.add_constant(X2)).fit(
                            cov_type="cluster", cov_kwds={"groups": S["cik"].values})
                        CUST_NOTE.append(f"  VOLUME × differentiation potential: β={r2_.params['ixn']:+.4f} "
                                         f"p={r2_.pvalues['ixn']:.3f}")
                        CUST_NOTE.append("    POSITIVE ⇒ customisation pays where differentiation is high and")
                        CUST_NOTE.append("    costs where it is not — the 2x2's central claim, finally testable.")
                    CUST_TBL = S[["cik", "year", "cust_int"]]
            else:
                CUST_NOTE.append("    coverage still under 30% — report as a limitation, not a test.")
        try: wide.to_csv(f"{OUT_DIR}/capsoft_xbrl.csv", index=False)
        except Exception: pass
    else:
        CUST_NOTE.append("No XBRL software facts retrieved; customisation VOLUME remains unmeasured.")
    globals()["CUST_NOTE"] = CUST_NOTE
    print("\n".join(CUST_NOTE))
````

````python
# ============ 5c · VALIDATION ARTIFACTS (offline, from cached filings) ============
# Automates the extraction and PRE-LABELLING of both validation tasks. Human adjudication
# is reduced to the ambiguous cases; the coding sheets are written to OUT_DIR.
VALID_ARTIFACTS = None
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    print("SKIPPED — no cached text layer.")
else:
    CTX = 350
    TC_STRONG = ["charter", "met ", "meetings", "consists of", "chair of the", "chairman of the",
                 "members of the", "composition", "standing committee", "committees of the board"]
    TC_BIO = ["has served", "serves as a director of", "previously served", "prior to joining",
              "nominee", "director since", "other public compan"]

    def _ctx(txt, m, w=CTX):
        return " ".join(txt[max(0, m.start() - w): m.end() + w].split())

    # ---------- (a) TechCom: evidence extraction + rule-based triage ----------
    ROLE_PATS = {
        "exec_cio":  re.compile(r"chief information officer"),
        "exec_cto":  re.compile(r"chief technology officer"),
        "exec_cdo":  re.compile(r"chief (digital|data) officer"),
        "exec_arch": re.compile(r"chief (enterprise )?architect"),
    }
    ROLE_OWN = ["our ", "company's", "of the company", "we appointed", "serves as our"]
    tc_rows, role_rows = [], []
    for cik in tqdm(FIRM_CIKS, desc="TechCom evidence"):
        try:
            fl = list(filings_of(cik, "DEF 14A"))
        except Exception:
            continue
        for fy, acc, doc in fl:
            if not (YEAR_MIN <= fy <= YEAR_MAX): continue
            fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
            if not os.path.exists(fp): continue
            txt = open(fp, encoding="utf-8", errors="ignore").read()
            _r = {}
            for rk, rp in ROLE_PATS.items():
                _r[rk] = 0
                for mm in rp.finditer(txt):
                    if any(cue in _ctx(txt, mm, 60) for cue in ROLE_OWN):
                        _r[rk] = 1; break
            role_rows.append(dict(cik=cik, year=fy, **_r))
            for m in TC_PATT.finditer(txt):
                w = _ctx(txt, m)
                strong = sum(k in w for k in TC_STRONG)
                bio = sum(k in w for k in TC_BIO)
                auto = "confirm" if (strong >= 2 and bio == 0) else ("reject" if bio >= 1 and strong == 0 else "review")
                tc_rows.append(dict(cik=cik, year=fy, match=m.group(0), auto=auto,
                                    strong_cues=strong, bio_cues=bio, evidence=w))
    tc_ev = pd.DataFrame(tc_rows)
    if len(tc_ev):
        agg = tc_ev.groupby(["cik", "year"])["auto"].agg(
            lambda s: "confirm" if (s == "confirm").any() else ("review" if (s == "review").any() else "reject"))
        tcv = agg.reset_index().rename(columns={"auto": "verdict"})
        tcv["techcom_verified"] = (tcv["verdict"] == "confirm").astype(int)
        tc_ev.to_csv(f"{OUT_DIR}/techcom_evidence.csv", index=False)
        tcv.to_csv(f"{OUT_DIR}/techcom_verified.csv", index=False)
        n_rev = int((tcv.verdict == "review").sum())
        print(f"\nTechCom triage: {len(tc_ev)} matches over {len(tcv)} firm-years → "
              f"confirm {int((tcv.verdict=='confirm').sum())} · review {n_rev} · reject {int((tcv.verdict=='reject').sum())}")
        print(f"  human effort: {n_rev} contexts to adjudicate (not {len(tcv)} filings) → techcom_evidence.csv")
        if G("df") is not None and "techcom" in df.columns:
            chk = df.merge(tcv[["cik", "year", "techcom_verified"]], on=["cik", "year"], how="left")
            chk["techcom_verified"] = chk["techcom_verified"].fillna(0).astype(int)
            flip = int((chk["techcom"] != chk["techcom_verified"]).sum())
            print(f"  regex vs verified: {flip}/{len(chk)} firm-years change ({flip/max(1,len(chk)):.1%}); "
                  f"share {chk.techcom.mean():.1%} → {chk.techcom_verified.mean():.1%}")
            globals()["df"] = chk
    else:
        tcv = None
        print("No TechCom matches found in cached proxies.")

    # ---------- (a2) named executive technology roles — structural, non-textual ----------
    if role_rows:
        er = pd.DataFrame(role_rows).groupby(["cik", "year"], as_index=False).max()
        rc_ = [c for c in er.columns if c.startswith("exec_")]
        er["exec_tech_roles"] = er[rc_].sum(axis=1)
        er.to_csv(f"{OUT_DIR}/exec_tech_roles.csv", index=False)
        print(f"\nNamed executive tech roles ({len(er)} firm-years): "
              + " · ".join(f"{c.replace('exec_','')}={er[c].mean():.0%}" for c in rc_))
        if G("df") is not None:
            globals()["df"] = df.merge(er, on=["cik", "year"], how="left")
            for c in rc_ + ["exec_tech_roles"]:
                df[c] = df[c].fillna(0)
            print(f"  merged into df · mean roles/firm-year = {df['exec_tech_roles'].mean():.2f}")

    # ---------- (b) 40-filing lexicon validation coding sheet ----------
    BOILER = ["risk factor", "may adversely", "no assurance", "forward-looking", "could materially"]
    PRODUCT = ["our products", "our customers", "our platform", "we offer", "our solutions", "our services"]
    SUBSTANT = ["we have implemented", "we invested", "our program", "initiative", "we completed",
                "we continue to", "our roadmap", "governance framework", "we established"]
    if G("df") is not None and "eagi" in df.columns:
        base = df[["cik", "year", "eagi"]].dropna().copy()
        try:
            base["stratum"] = pd.qcut(base["eagi"], 3, labels=["low", "mid", "high"], duplicates="drop").astype(str)
        except Exception:
            base["stratum"] = "all"
        parts = []
        for _lab, _g in base.groupby("stratum", observed=True):
            parts.append(_g.sample(min(14, len(_g)), random_state=42))
        pick = pd.concat(parts, ignore_index=True).head(42) if parts else base.head(0)
        rows = []
        for _, r in tqdm(list(pick.iterrows()), desc="lexicon contexts"):
            cik, yr = int(r.cik), int(r.year)
            try:
                fl = [x for x in filings_of(cik, "10-K") if x[0] == yr]
            except Exception:
                continue
            for fy, acc, doc in fl[:1]:
                fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
                if not os.path.exists(fp): continue
                txt = open(fp, encoding="utf-8", errors="ignore").read()
                for t in LEXICON:
                    patt = r"\b" + re.escape(t).replace(r"\-", "[- ]") + r"\b"
                    for m in list(re.finditer(patt, txt))[:2]:
                        w = _ctx(txt, m, 250)
                        lab = ("boilerplate" if any(k in w for k in BOILER) else
                               "product_marketing" if any(k in w for k in PRODUCT) else
                               "substantive" if any(k in w for k in SUBSTANT) else "unclear")
                        rows.append(dict(cik=cik, year=yr, stratum=str(r.stratum), term=t,
                                         auto_label=lab, human_label="", context=w))
        lexval = pd.DataFrame(rows)
        if len(lexval):
            lexval.to_csv(f"{OUT_DIR}/lexicon_validation_sheet.csv", index=False)
            try:
                with pd.ExcelWriter(f"{OUT_DIR}/lexicon_validation_sheet.xlsx", engine="openpyxl") as xw:
                    lexval.to_excel(xw, sheet_name="coding_sheet", index=False)
            except Exception:
                pass
            mix = lexval.auto_label.value_counts(normalize=True).round(3).to_dict()
            print(f"\nLexicon validation sheet: {len(lexval)} contexts from {lexval.cik.nunique()} filings "
                  f"across 3 EAGI strata → lexicon_validation_sheet.xlsx")
            print(f"  auto-label mix: {mix}")
            print("  Fill 'human_label' (substantive / boilerplate / product_marketing) — the share of")
            print("  SUBSTANTIVE matches is the construct-validity estimate the paper needs.")
            globals()["LEXVAL_MIX"] = mix
            VALID_ARTIFACTS = True

    # ---------- (c) GOLD STANDARD + AGREEMENT SCORING ----------
    # Automation can replace the BULK of the coding, but not the validation OF the automation.
    # The irreducible human task is a small gold standard; everything else is then machine-applied
    # and reported with an agreement statistic.
    def cohen_kappa(a, b):
        a, b = pd.Series(list(a)).astype(str), pd.Series(list(b)).astype(str)
        cats = sorted(set(a) | set(b))
        if len(a) == 0 or len(cats) < 2: return float("nan")
        po = float((a.values == b.values).mean())
        pe = sum((a == c).mean() * (b == c).mean() for c in cats)
        return float((po - pe) / (1 - pe)) if pe < 1 else float("nan")

    def score_coded(path, auto_col, human_col, label):
        if not os.path.exists(path): return None
        cd = pd.read_csv(path)
        cd = cd[cd[human_col].astype(str).str.strip().ne("") & cd[human_col].notna()]
        if len(cd) < 10: return None
        k = cohen_kappa(cd[auto_col], cd[human_col])
        acc = float((cd[auto_col].astype(str) == cd[human_col].astype(str)).mean())
        print(f"  {label}: n={len(cd)} coded · agreement={acc:.0%} · Cohen κ={k:.3f} "
              + ("(κ≥0.6 → machine labels usable for the full corpus)" if k == k and k >= 0.6
                 else "(κ<0.6 → rules need revision before machine labels can be relied on)"))
        return dict(task=label, n=len(cd), accuracy=round(acc, 3), kappa=round(k, 3) if k == k else None)

    AGREE = []
    print("\nGold standard (the only irreducible human step):")
    if len(tc_ev):
        _parts = []
        for _lab, _g in tc_ev.groupby("auto", observed=True):
            _parts.append(_g.sample(min(15, len(_g)), random_state=7))
        gs_tc = pd.concat(_parts, ignore_index=True) if _parts else tc_ev.head(0)
        gs_tc = gs_tc.assign(human_label="")[["cik", "year", "auto", "evidence", "human_label"]]
        gs_tc.to_csv(f"{OUT_DIR}/techcom_goldstandard.csv", index=False)
        print(f"  techcom_goldstandard.csv — {len(gs_tc)} rows to code "
              f"(stratified across confirm/review/reject, NOT the full {len(tc_ev)})")
        r = score_coded(f"{OUT_DIR}/techcom_goldstandard_coded.csv", "auto", "human_label", "TechCom triage")
        if r: AGREE.append(r)
    if G("LEXVAL_MIX") is not None or ("lexval" in dir()):
        r = score_coded(f"{OUT_DIR}/lexicon_validation_sheet_coded.csv", "auto_label", "human_label", "Lexicon contexts")
        if r: AGREE.append(r)
    if AGREE:
        globals()["AGREE_TBL"] = pd.DataFrame(AGREE)
        pd.DataFrame(AGREE).to_csv(f"{OUT_DIR}/coding_agreement.csv", index=False)
    else:
        print("  (no *_coded.csv found yet — code the gold standard, save with '_coded' suffix, re-run)")

    # ---------- (d) SECOND CLASSIFIER: cached LLM labels (labels become DATA, not a live call) ----------
    import hashlib
    LLM_LABELS, LLM_MODE = None, "not run"
    if len(tc_ev):
        llm_path = f"{CACHE_DIR}/llm_techcom_labels.csv"
        cache = pd.read_csv(llm_path) if os.path.exists(llm_path) else pd.DataFrame(columns=["key", "llm_label"])
        tc_ev = tc_ev.copy()
        tc_ev["key"] = [hashlib.md5(str(x).encode("utf-8")).hexdigest() for x in tc_ev["evidence"]]
        have = set(cache["key"]) if len(cache) else set()
        todo = tc_ev[~tc_ev["key"].isin(have)].drop_duplicates("key").sort_values("key")
        if len(cache): print(f"\nLLM label cache: {len(cache)} labels reused")
        if USE_LLM_CODING and len(todo):
            _key = None
            try:
                from google.colab import userdata
                _key = userdata.get("ANTHROPIC_API_KEY")
            except Exception:
                _key = None
            if not _key:
                print(f"  {len(todo)} contexts unlabelled and no ANTHROPIC_API_KEY — rules remain the only classifier.")
            else:
                import requests as _rq
                got_rows, errs = [], 0
                PROMPT = ("You are classifying a passage from a US proxy statement (DEF 14A).\n"
                          "Answer with EXACTLY ONE WORD:\n"
                          "CONFIRM - the passage shows the FILING COMPANY has a board-level technology, "
                          "digital or innovation committee.\n"
                          "REJECT - it refers to another company, or to a director's role outside this board.\n"
                          "REVIEW - genuinely ambiguous.\n\nPassage:\n")
                for _, rr in tqdm(list(todo.head(MAX_LLM_CALLS).iterrows()), desc="LLM labelling"):
                    try:
                        rp = _rq.post("https://api.anthropic.com/v1/messages", timeout=45,
                                      headers={"x-api-key": _key, "anthropic-version": "2023-06-01",
                                               "content-type": "application/json"},
                                      json={"model": LLM_MODEL, "max_tokens": 8,
                                            "temperature": LLM_TEMPERATURE,
                                            "messages": [{"role": "user",
                                                          "content": PROMPT + str(rr["evidence"])[:1500]}]})
                        w = rp.json().get("content", [{}])[0].get("text", "").strip().upper()
                        lab = ("confirm" if "CONFIRM" in w else "reject" if "REJECT" in w else "review")
                    except Exception:
                        errs += 1; lab = "review"
                    got_rows.append({"key": rr["key"], "llm_label": lab})
                if got_rows:
                    cache = pd.concat([cache, pd.DataFrame(got_rows)], ignore_index=True).drop_duplicates("key")
                    cache.to_csv(llm_path, index=False)
                    print(f"  LLM labelled {len(got_rows)} new contexts ({errs} errors → 'review') · cached to {llm_path}")
        if len(cache):
            tc_ev = tc_ev.merge(cache, on="key", how="left")
            cov = tc_ev["llm_label"].notna().mean()
            LLM_MODE = f"cached labels for {cov:.0%} of contexts (model {LLM_MODEL}, temp {LLM_TEMPERATURE})"
            LLM_LABELS = cache
            tc_ev.to_csv(f"{OUT_DIR}/techcom_evidence.csv", index=False)

    # ---------- (e) THREE TECHCOM DEFINITIONS — measurement is itself a result ----------
    def _verdict(series_auto, series_llm):
        """Primary = deterministic rules; LLM upgrades only where rules said 'review'."""
        out = series_auto.copy()
        if series_llm is not None:
            m = (out == "review") & series_llm.isin(["confirm", "reject"])
            out = out.mask(m, series_llm)
        return out
    if len(tc_ev):
        llm_col = tc_ev["llm_label"] if "llm_label" in tc_ev.columns else None
        tc_ev["verdict_final"] = _verdict(tc_ev["auto"], llm_col)
        rank = {"confirm": 3, "review": 2, "reject": 1}
        fy_v = (tc_ev.assign(r=tc_ev["verdict_final"].map(rank))
                     .groupby(["cik", "year"], as_index=False)["r"].max())
        fy_v["verdict"] = fy_v["r"].map({3: "confirm", 2: "review", 1: "reject"})
        fy_v["techcom_strict"] = (fy_v["verdict"] == "confirm").astype(int)
        fy_v["techcom_incl"] = fy_v["verdict"].isin(["confirm", "review"]).astype(int)
        fy_v[["cik", "year", "verdict", "techcom_strict", "techcom_incl"]].to_csv(
            f"{OUT_DIR}/techcom_verified.csv", index=False)
        globals()["TC_REVIEW_SHARE"] = float((fy_v["verdict"] == "review").mean())
        print("\nTechCom coding scenarios — the ambiguous pile is bounded, not guessed:")
        print(f"  ambiguous ('review') firm-years: {(fy_v['verdict']=='review').sum()} "
              f"({globals()['TC_REVIEW_SHARE']:.1%}) — this width drives the bounds below")
        print(f"  raw regex        share = {df['techcom'].mean():.1%}" if G("df") is not None and "techcom" in df.columns else "")
        print(f"  LOWER bound  (review→reject)  share = {fy_v['techcom_strict'].mean():.1%}")
        print(f"  UPPER bound  (review→confirm) share = {fy_v['techcom_incl'].mean():.1%}")
        if G("df") is not None:
            globals()["df"] = df.merge(fy_v[["cik", "year", "techcom_strict", "techcom_incl"]],
                                       on=["cik", "year"], how="left")
            for c in ["techcom_strict", "techcom_incl"]: df[c] = df[c].fillna(0).astype(int)
        globals()["LLM_MODE"] = LLM_MODE
        VALID_ARTIFACTS = True

    # ---------- (f) GOLD STANDARD for the QUALITATIVE follow-up study (not needed for the empirics) ----------
    if len(tc_ev):
        _parts = []
        for _lab, _g in tc_ev.groupby("auto", observed=True):
            _parts.append(_g.sample(min(15, len(_g)), random_state=SEED))
        gs_tc = pd.concat(_parts, ignore_index=True) if _parts else tc_ev.head(0)
        cols_keep = [c for c in ["cik", "year", "auto", "llm_label", "verdict_final", "evidence"] if c in gs_tc.columns]
        gs_tc.assign(human_label="")[cols_keep + ["human_label"]].to_csv(
            f"{OUT_DIR}/techcom_goldstandard.csv", index=False)
        print(f"  gold standard for the qualitative follow-up: techcom_goldstandard.csv ({len(gs_tc)} rows)")

    def cohen_kappa(a, b):
        a, b = pd.Series(list(a)).astype(str), pd.Series(list(b)).astype(str)
        cats = sorted(set(a) | set(b))
        if len(a) == 0 or len(cats) < 2: return float("nan")
        po = float((a.values == b.values).mean())
        pe = sum((a == c).mean() * (b == c).mean() for c in cats)
        return float((po - pe) / (1 - pe)) if pe < 1 else float("nan")
    if len(tc_ev) and "llm_label" in tc_ev.columns:
        _b = tc_ev.dropna(subset=["llm_label"])
        if len(_b) > 20:
            _k = cohen_kappa(_b["auto"], _b["llm_label"])
            globals()["KAPPA_RULES_LLM"] = _k
            print(f"  classifier agreement (rules vs LLM, no human input): "
                  f"accuracy={float((_b['auto'].astype(str)==_b['llm_label'].astype(str)).mean()):.0%} · κ={_k:.3f}")
````

````python
# ============ 5d · ERP TRANSITION SIGNALS (exploratory, offline) ============
# The 2027 end-of-maintenance date has NO cross-sectional variation — a pure time shock is
# absorbed by year FE. A usable design needs firm-level EXPOSURE. What is extractable from
# cached filings is DISCLOSED transition activity, which is endogenous (firms disclose ERP
# work when it is material). Treated here as descriptive/hypothesis-generating, never as an IV.
ERP_TBL = None
if G("FIRM_CIKS") is None or G("CACHE_DIR") is None:
    print("SKIPPED — no cached text layer.")
else:
    VEND = {"sap": r"\bsap\b", "oracle": r"\boracle\b", "dynamics": r"microsoft dynamics",
            "workday": r"\bworkday\b", "infor": r"\binfor\b"}
    MIGR = r"(s/?4\s?hana|erp (implementation|migration|transformation|conversion|upgrade)|" \
           r"enterprise resource planning (system|implementation)|system conversion|go[- ]live)"
    erp_path = f"{CACHE_DIR}/erp_signals.csv"
    ERP_TBL = read_cache(erp_path, required=["cik", "year", "erp_migration"], label="erp_signals.csv")
    if ERP_TBL is not None:
        print(f"Reusing ERP signals — {len(ERP_TBL)} firm-years")
    else:
        rows = []
        for cik in tqdm(FIRM_CIKS, desc="ERP signals"):
            try: fl = list(filings_of(cik, "10-K"))
            except Exception: continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX): continue
                fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
                if not os.path.exists(fp): continue
                t = open(fp, encoding="utf-8", errors="ignore").read()
                r = {"cik": cik, "year": fy, "erp_migration": int(bool(re.search(MIGR, t)))}
                for v, pat in VEND.items(): r["vend_" + v] = int(bool(re.search(pat, t)))
                rows.append(r)
        ERP_TBL = pd.DataFrame(rows)
        if len(ERP_TBL): ERP_TBL.to_csv(erp_path, index=False)
    if ERP_TBL is not None and len(ERP_TBL):
        vs = [c for c in ERP_TBL.columns if c.startswith("vend_")]
        print("ERP disclosure prevalence: migration={:.0%} · ".format(ERP_TBL.erp_migration.mean())
              + " · ".join(f"{c.replace('vend_','')}={ERP_TBL[c].mean():.0%}" for c in vs))
        first = (ERP_TBL[ERP_TBL.erp_migration == 1].groupby("cik")["year"].min().rename("erp_year"))
        print(f"  firms disclosing a transition at least once: {first.notna().sum()}/{ERP_TBL.cik.nunique()}")
        if G("df") is not None:
            globals()["df"] = df.merge(ERP_TBL, on=["cik", "year"], how="left").merge(
                first.reset_index(), on="cik", how="left")
            for c in ["erp_migration"] + vs: df[c] = df[c].fillna(0)
            e = df.dropna(subset=["erp_year", "roa"]).copy()
            e["tau"] = e["year"] - e["erp_year"]
            if len(e) > 30:
                print("\nDescriptive ROA around FIRST disclosed ERP transition (NOT causal):")
                print(e[e.tau.between(-2, 3)].groupby("tau")["roa"].agg(["mean", "count"]).round(4).to_string())
            print("\n⚠ Endogenous by construction: disclosure is triggered by materiality.")
            print("  Valid exposure requires external install-base data; SAP-internal customer")
            print("  lists would reintroduce exactly the vendor conflict the CIAOF rebrand removed.")
````

````python
# ============ 6 · ESTIMATION — H1 & H2 (two-way FE) ============
# SE note: with T≈7 the Driscoll–Kraay asymptotics (large-T) are weak, so firm-CLUSTERED
# SEs are primary and DK is reported alongside as a cross-sectional-dependence check.
if G("df") is None:
    print("SKIPPED — no merged panel (prefetch-only or upstream incomplete).")
else:
    d = df.sort_values(["cik", "year"]).copy()
    _dk = int(d.duplicated(subset=["cik", "year"]).sum())
    if _dk:
        print(f"  ⚠ {_dk} duplicate (cik,year) rows entering estimation — collapsing by mean")
        d = d.groupby(["cik", "year"], as_index=False).mean(numeric_only=True)
    CAND = [c for c in ["eagi", "ci", "size", "lev", "rnd", "rnd_missing", "capex", "capsoft"] if c in d.columns]
    if "placebo" in d.columns: CAND = CAND + ["placebo"]

    def add_lags(frame, cols, k=1):
        """Year-aware lag: t-k must be the previous CALENDAR year, not the previous row.
        Positional shift() silently mis-lags whenever a firm-year is missing."""
        lg = frame[["cik", "year"] + cols].copy()
        lg["year"] = lg["year"] + k
        lg = lg.rename(columns={c: c + "_l" for c in cols})
        return frame.merge(lg, on=["cik", "year"], how="left")

    d = add_lags(d, CAND, 1)

    POSS = [c for c in ["size", "lev", "rnd", "rnd_missing", "capex", "capsoft"] if c + "_l" in d.columns]
    keep = []
    for c in POSS:
        cl, cov = c + "_l", d[c + "_l"].notna().mean()
        within = d.groupby("cik")[cl].transform(lambda x: x - x.mean()).abs().sum()
        reason = ""
        if cov < 0.70: reason = "coverage <70%"
        elif not (within > 1e-10): reason = "no within-firm variation (absorbed by firm FE)"
        else: keep.append(c)
        print(f"  control {c:12s} coverage {cov:.0%}" + (f"  → DROPPED ({reason})" if reason else ""))
    CTRL = [c + "_l" for c in keep]
    print("controls used:", CTRL)

    n0 = len(d)
    n1 = int(d["eagi_l"].notna().sum())
    n2 = int((d["eagi_l"].notna() & d["ci_l"].notna()).sum())
    d = d.dropna(subset=["eagi_l", "ci_l"] + CTRL).set_index(["cik", "year"])
    EST_N = len(d)
    EST_FIRMS = d.index.get_level_values(0).nunique()
    globals()["SD_EAGI_L"] = float(d["eagi_l"].std()) if EST_N else 0.0
    globals()["SD_CI_L"] = float(d["ci_l"].std()) if EST_N else 0.0
    print(f"attrition: {n0} merged → {n1} with EAGI(t-1) → {n2} with CI(t-1) → {EST_N} rows · {EST_FIRMS} firms")
    print(f"sd(EAGI_l)={SD_EAGI_L:.4f} · sd(CI_l)={SD_CI_L:.4f}")
    if EST_N < 100:
        print(f"\n*** WARNING: {EST_N} rows / {EST_FIRMS} firms — two-way FE is degenerate here. ***\n")

    # centred interaction: β_ixn is unchanged, but main effects become effects AT THE MEAN
    # of the other variable rather than at zero, and the product term is far less collinear.
    d["ci_c"] = d["ci_l"] - d["ci_l"].mean()
    d["eagi_c"] = d["eagi_l"] - d["eagi_l"].mean()
    d["ci_x_eagi_l"] = d["ci_c"] * d["eagi_c"]

    def fe_fit(dep, xs, data, se="clustered"):
        return fe_ols(dep, xs, data, se=se)     # own implementation, no estimator package

    def fe_dk(dep, xs, data):     # kept for the robustness cell
        return fe_fit(dep, xs, data, se="clustered")

    results, RESULTS_DK = {}, {}
    for y in ["roa", "q", "sga"]:
        dy = d.dropna(subset=[y])
        ny, fy_ = len(dy), dy.index.get_level_values(0).nunique()
        print(f"\n[{y.upper()}] rows: {ny} · firms: {fy_}")
        if ny < 30:
            print(f"  too few rows for {y.upper()} — skipped"); continue
        h1 = ["eagi_l"] + CTRL
        h2 = ["ci_c", "eagi_c", "ci_x_eagi_l"] + CTRL
        results[y] = (fe_fit(y, h1, dy), fe_fit(y, h2, dy))
        try:
            RESULTS_DK[y] = (fe_fit(y, h1, dy, se="kernel"), fe_fit(y, h2, dy, se="kernel"))
        except Exception as e:
            print("  DK SEs unavailable:", type(e).__name__)
        print(f"===== {y.upper()} — H1 (firm-clustered SE) =====")
        print(results[y][0].summary.tables[1])
        print(f"===== {y.upper()} — H2 (centred interaction) =====")
        print(results[y][1].summary.tables[1])
````

````python
# ============ 6b · ALTERNATIVE DESIGNS — between-firm, IV, adaptability ============
# H1/H2 as originally specified test the MEAN with a WITHIN-firm estimator. Both are
# mismatched to the theory and to the construct. These are the better-matched tests.
ALT_NOTE = []
if G("df") is None or G("CTRL") is None:
    print("SKIPPED — estimation incomplete.")
else:
    import statsmodels.api as sm

    A_ = df.copy()
    idx_col = "eagi_best" if "eagi_best" in A_.columns and A_["eagi_best"].notna().sum() > 100 else "eagi"
    ALT_NOTE.append(f"[P2] index used: {idx_col}")
    base_cols = [c for c in ["size", "lev", "rnd", "capsoft"] if c in A_.columns
                 and A_[c].notna().mean() >= 0.70]
    _drop_cov = [c for c in ["size", "lev", "rnd", "capsoft"]
                 if c in A_.columns and A_[c].notna().mean() < 0.70]
    if _drop_cov:
        ALT_NOTE.append(f"[P2] controls dropped for <70% coverage: {_drop_cov} "
                        "(keeping them would have collapsed the sample)")
    A_ = A_.sort_values(["cik", "year"])
    lg = A_[["cik", "year", idx_col, "ci"] + base_cols].copy(); lg["year"] += 1
    lg = lg.rename(columns={c: c + "_L" for c in [idx_col, "ci"] + base_cols})
    A_ = A_.merge(lg, on=["cik", "year"], how="left")
    CT = [c + "_L" for c in base_cols]

    # ---------- (1) BETWEEN-FIRM: industry x year FE (EAGI is a firm type, not an event) ----------
    if "industry_raw" in A_.columns:
        A_["iy"] = A_["industry_raw"].astype(str) + "_" + A_["year"].astype(str)
    else:
        A_["iy"] = A_["year"].astype(str)
    bx = A_.dropna(subset=[idx_col + "_L", "ci_L", "roa"] + CT).copy()
    if len(bx) > 60:
        X = pd.get_dummies(bx[["iy"]], drop_first=True, dtype=float)
        X[idx_col + "_L"] = bx[idx_col + "_L"].values
        for c in CT: X[c] = bx[c].values
        X = sm.add_constant(X)
        r_b = sm.OLS(bx["roa"].values, X).fit(cov_type="cluster",
                                              cov_kwds={"groups": bx["cik"].values})
        b, se = r_b.params[idx_col + "_L"], r_b.bse[idx_col + "_L"]
        ALT_NOTE.append(f"[P2] BETWEEN (industry x year FE, firm-clustered): beta={b:+.4f} se={se:.4f} "
                        f"p={r_b.pvalues[idx_col + '_L']:.3f} n={len(bx)}")
    # ---------- (2) MUNDLAK: separate within from between explicitly ----------
    mk = A_.dropna(subset=[idx_col + "_L", "roa"] + CT).copy()
    if len(mk) > 60:
        mk["x_bar"] = mk.groupby("cik")[idx_col + "_L"].transform("mean")
        mk["x_dev"] = mk[idx_col + "_L"] - mk["x_bar"]
        Xm = sm.add_constant(mk[["x_bar", "x_dev"] + CT].astype(float))
        r_m = sm.OLS(mk["roa"].values, Xm).fit(cov_type="cluster", cov_kwds={"groups": mk["cik"].values})
        ALT_NOTE.append(f"[P2] MUNDLAK: between beta={r_m.params['x_bar']:+.4f} (p={r_m.pvalues['x_bar']:.3f}) · "
                        f"within beta={r_m.params['x_dev']:+.4f} (p={r_m.pvalues['x_dev']:.3f})")
        ALT_NOTE.append("[P2]   → if between is meaningful and within is noise, firm FE was discarding the signal.")
    # ---------- (3) SPLIT-HALF IV: corrects attenuation AND tests the construct ----------
    if {"eagi_hA", "eagi_hB"}.issubset(A_.columns):
        iv = A_.sort_values(["cik", "year"]).copy()
        lg2 = iv[["cik", "year", "eagi_hA", "eagi_hB"]].copy(); lg2["year"] += 1
        iv = iv.merge(lg2.rename(columns={"eagi_hA": "hA_L", "eagi_hB": "hB_L"}), on=["cik", "year"], how="left")
        iv = iv.dropna(subset=["hA_L", "hB_L", "roa"] + CT)
        if len(iv) > 60:
            Xf = sm.add_constant(iv[["hA_L"] + CT].astype(float))
            fs = sm.OLS(iv["hB_L"].values, Xf).fit(cov_type="cluster", cov_kwds={"groups": iv["cik"].values})
            F = float(fs.tvalues["hA_L"] ** 2)
            iv["hB_hat"] = fs.predict(Xf)
            Xs = sm.add_constant(iv[["hB_hat"] + CT].astype(float))
            ss = sm.OLS(iv["roa"].values, Xs).fit(cov_type="cluster", cov_kwds={"groups": iv["cik"].values})
            ALT_NOTE.append(f"[P2] SPLIT-HALF IV: first-stage F={F:.2f} " +
                            ("(WEAK: halves do not predict each other → no common latent construct)"
                             if F < 10 else "(usable)") +
                            f" · 2SLS beta={ss.params['hB_hat']:+.4f} p={ss.pvalues['hB_hat']:.3f}")
    # ---------- (4) ADAPTABILITY: the theory predicts variance, not level ----------
    ad = A_.dropna(subset=[idx_col + "_L", "roa"]).copy()
    vol = (ad.groupby("cik")["roa"].std().rename("roa_vol").reset_index())
    xbar = ad.groupby("cik")[idx_col + "_L"].mean().rename("x_bar").reset_index()
    fmeans = ad.groupby("cik")[[c for c in ["size", "lev", "rnd"] if c in ad.columns]].mean().reset_index()
    cs = vol.merge(xbar, on="cik").merge(fmeans, on="cik", how="left").dropna()
    if "industry_raw" in ad.columns:
        cs = cs.merge(ad.groupby("cik")["industry_raw"].first().reset_index(), on="cik", how="left")
    if len(cs) > 40:
        Xv = sm.add_constant(cs[["x_bar"]].astype(float))
        rv = sm.OLS(cs["roa_vol"].values, Xv).fit(cov_type="HC1")
        ALT_NOTE.append(f"[P2] ADAPTABILITY raw (ROA volatility ~ mean EAGI, NO controls): "
                        f"beta={rv.params['x_bar']:+.4f} p={rv.pvalues['x_bar']:.3f} n={len(cs)}")
        cols_v = ["x_bar"] + [c for c in ["size", "lev", "rnd"] if c in cs.columns]
        Xv2 = cs[cols_v].astype(float)
        if "industry_raw" in cs.columns and cs["industry_raw"].nunique() > 1:
            Xv2 = pd.concat([Xv2, pd.get_dummies(cs["industry_raw"].astype(str), prefix="ind",
                                                 drop_first=True, dtype=float)], axis=1)
        rv2 = sm.OLS(cs["roa_vol"].values, sm.add_constant(Xv2)).fit(cov_type="HC1")
        ALT_NOTE.append(f"[P2] ADAPTABILITY controlled (+size, lev, rnd, industry): "
                        f"beta={rv2.params['x_bar']:+.4f} p={rv2.pvalues['x_bar']:.3f}"
                        + ("  → survives controls" if rv2.pvalues["x_bar"] < 0.05
                           else "  → does NOT survive controls; the raw result was composition"))
    sh = A_.dropna(subset=[idx_col + "_L", "roa"] + CT).copy()
    sh["post2020"] = (sh["year"] >= 2020).astype(float)
    sh["x_post"] = sh[idx_col + "_L"] * sh["post2020"]
    if len(sh) > 60:
        Xs2 = sm.add_constant(sh[[idx_col + "_L", "post2020", "x_post"] + CT].astype(float))
        rs = sm.OLS(sh["roa"].values, Xs2).fit(cov_type="cluster", cov_kwds={"groups": sh["cik"].values})
        _bp = float(rs.params["x_post"])
        ALT_NOTE.append(f"[P2] SHOCK INTERACTION (EAGI x post-2020): beta={_bp:+.4f} "
                        f"p={rs.pvalues['x_post']:.3f}  ⇒ higher-index firms performed "
                        + ("BETTER" if _bp > 0 else "WORSE") + " after 2020")
    # ---------- (5) CCAOF REVISITED: is Core Integrity itself a between-firm story? ----------
    ci_d = df.dropna(subset=["ci"]).copy()
    if len(ci_d) > 60:
        _w = float(ci_d.groupby("cik")["ci"].transform(lambda x: x - x.mean()).std())
        _b = float(ci_d.groupby("cik")["ci"].mean().std())
        _share = _b ** 2 / max(1e-12, _b ** 2 + _w ** 2)
        ALT_NOTE.append(f"[P1] CI variance decomposition: between-firm sd={_b:.4f} · within-firm sd={_w:.4f} "
                        f"· between share of variance={_share:.0%}")
        m2_ = A_.dropna(subset=["ci_L", "roa"] + CT).copy()
        if len(m2_) > 60:
            m2_["ci_bar"] = m2_.groupby("cik")["ci_L"].transform("mean")
            m2_["ci_dev"] = m2_["ci_L"] - m2_["ci_bar"]
            Xc = sm.add_constant(pd.concat([m2_[["ci_bar", "ci_dev"] + CT].astype(float),
                                            pd.get_dummies(m2_["year"].astype(str), prefix="yr",
                                                           drop_first=True, dtype=float)], axis=1))
            rc = sm.OLS(m2_["roa"].values, Xc).fit(cov_type="cluster", cov_kwds={"groups": m2_["cik"].values})
            ALT_NOTE.append(f"[P1] CCAOF revisited (year FE) — CI between beta={rc.params['ci_bar']:+.4f} "
                            f"(p={rc.pvalues['ci_bar']:.3f}) · within beta={rc.params['ci_dev']:+.4f} "
                            f"(p={rc.pvalues['ci_dev']:.3f})")
            # the decisive control: sector composition. A between-firm estimate without industry
            # is a comparison of industries as much as of firms.
            if "industry_raw" in m2_.columns and m2_["industry_raw"].nunique() > 1:
                Xci = sm.add_constant(pd.concat([
                    m2_[["ci_bar", "ci_dev"] + CT].astype(float),
                    pd.get_dummies(m2_["industry_raw"].astype(str) + "_" + m2_["year"].astype(str),
                                   prefix="iy", drop_first=True, dtype=float)], axis=1))
                rci = sm.OLS(m2_["roa"].values, Xci).fit(cov_type="cluster", cov_kwds={"groups": m2_["cik"].values})
                ALT_NOTE.append(f"[P1] CCAOF revisited (INDUSTRY x YEAR FE) — CI between "
                                f"beta={rci.params['ci_bar']:+.4f} (p={rci.pvalues['ci_bar']:.3f}) · "
                                f"within beta={rci.params['ci_dev']:+.4f} (p={rci.pvalues['ci_dev']:.3f}) · "
                                f"industries={m2_['industry_raw'].nunique()}")
                _b0, _b1 = float(rc.params["ci_bar"]), float(rci.params["ci_bar"])
                if abs(_b0) > 1e-9:
                    ALT_NOTE.append(f"[P1]   sector composition explains {100*(1-_b1/_b0):.0f}% of the raw "
                                    "between-firm coefficient")
                ALT_NOTE.append(f"[P1]   per-1sd(between) ROA effect with industry FE = "
                                f"{_b1 * float(ci_d.groupby('cik')['ci'].mean().std()):+.4f}")
            # the inverted-U is a SHAPE claim: it can only be traced across the full range,
            # i.e. between firms. Testing it on within-firm variation is near-impossible.
            m2_["ci_bar2"] = m2_["ci_bar"] ** 2
            Xq = sm.add_constant(m2_[["ci_bar", "ci_bar2"] + CT].astype(float))
            rq = sm.OLS(m2_["roa"].values, Xq).fit(cov_type="cluster", cov_kwds={"groups": m2_["cik"].values})
            b1, b2, p2 = rq.params["ci_bar"], rq.params["ci_bar2"], rq.pvalues["ci_bar2"]
            nt = f"inverted-U tested BETWEEN firms: b1={b1:+.4f} b2={b2:+.4f} (p={p2:.3f})"
            if b2 < 0 and p2 < 0.10:
                vx = -b1 / (2 * b2)
                lo, hi = m2_["ci_bar"].quantile([0.05, 0.95])
                inside = bool(lo <= vx <= hi)
                nt += (f" · vertex={vx:+.3f} {'INSIDE' if inside else 'OUTSIDE'} the 5–95% range "
                       f"[{lo:+.3f},{hi:+.3f}]" + ("" if inside else " → NOT an interior optimum; do not report as an inverted-U"))
            else:
                nt += " · not concave at conventional levels → no inverted-U claim"
            ALT_NOTE.append("[P1]   " + nt)
            ALT_NOTE.append("[P1]   CAUTION: between-firm estimates are descriptive. Disciplined firms differ in age,")
            ALT_NOTE.append("[P1]   M&A history, estate complexity and industry; reverse causality (profits fund")
            ALT_NOTE.append("[P1]   discipline) is unresolved. Report as association, not effect.")

    # ---------- (5b) STRUCTURAL GOVERNANCE — non-textual proxy for the same construct ----------
    st_cols = [c for c in ["techcom_verified", "techcom", "exec_cio", "exec_cto", "exec_cdo", "exec_arch"]
               if c in df.columns]
    st_cols = [c for c in st_cols if not (c == "techcom" and "techcom_verified" in st_cols)]
    if len(st_cols) >= 2:
        S_ = df.copy()
        S_["gov_struct"] = S_[st_cols].fillna(0).sum(axis=1)
        lg3 = S_[["cik", "year", "gov_struct"]].copy(); lg3["year"] += 1
        S_ = S_.merge(lg3.rename(columns={"gov_struct": "gov_struct_L"}), on=["cik", "year"], how="left")
        S_ = S_.merge(A_[["cik", "year"] + CT].drop_duplicates(["cik", "year"]), on=["cik", "year"], how="left")
        sx = S_.dropna(subset=["gov_struct_L", "roa"] + CT)
        if len(sx) > 60:
            Xg = pd.get_dummies(sx[["year"]].astype(str), drop_first=True, dtype=float)
            Xg["gov_struct_L"] = sx["gov_struct_L"].values
            for c in CT: Xg[c] = sx[c].values
            rg = sm.OLS(sx["roa"].values, sm.add_constant(Xg)).fit(
                cov_type="cluster", cov_kwds={"groups": sx["cik"].values})
            ALT_NOTE.append(f"[P2] STRUCTURAL GOVERNANCE ({'+'.join(st_cols)}) → ROA: "
                            f"beta={rg.params['gov_struct_L']:+.4f} p={rg.pvalues['gov_struct_L']:.3f} n={len(sx)}")
            if idx_col in sx.columns or (idx_col + "_L") in S_.columns:
                _c = S_.dropna(subset=["gov_struct", idx_col])
                if len(_c) > 30:
                    ALT_NOTE.append(f"[P2]   corr(structural governance, {idx_col}) = "
                                    f"{_c['gov_struct'].corr(_c[idx_col]):+.3f} "
                                    "(convergent validity for the text index)")

    # ---------- (5c) LONG HORIZON — governance payoffs are multi-year, not t-1 → t ----------
    lh = df.dropna(subset=[idx_col, "ci", "roa"]).sort_values(["cik", "year"])
    fst = lh.groupby("cik").first().reset_index(); lst = lh.groupby("cik").last().reset_index()
    span = fst[["cik", idx_col, "ci"] + [c for c in ["size", "lev", "rnd"] if c in fst.columns]].copy()
    span["d_roa"] = (lst["roa"] - fst["roa"]).values
    span["yrs"] = (lst["year"] - fst["year"]).values
    span = span[span["yrs"] >= 3].dropna()
    if "industry_raw" in fst.columns:
        span = span.merge(fst[["cik", "industry_raw"]], on="cik", how="left")
    if len(span) > 40:
        _Xl = span[[idx_col, "ci"] + [c for c in ["size", "lev", "rnd"] if c in span.columns]].astype(float)
        if "industry_raw" in span.columns and span["industry_raw"].nunique() > 1:
            _Xl = pd.concat([_Xl, pd.get_dummies(span["industry_raw"].astype(str), prefix="ind",
                                                 drop_first=True, dtype=float)], axis=1)
        Xl = sm.add_constant(_Xl)
        rl = sm.OLS(span["d_roa"].values, Xl).fit(cov_type="HC1")
        ALT_NOTE.append(f"[P2] LONG HORIZON (Δ ROA first→last, ≥3 yrs, n={len(span)}): "
                        f"initial {idx_col} beta={rl.params[idx_col]:+.4f} p={rl.pvalues[idx_col]:.3f} · "
                        f"initial CI beta={rl.params['ci']:+.4f} p={rl.pvalues['ci']:.3f}")
        ALT_NOTE.append("[P2]   (t−1→t may be too short a window for architecture payoffs)")

    # ---------- (5d) PAPER 1 robustness: the alternative CI composite ----------
    if "ci_alt" in df.columns and df["ci_alt"].notna().sum() > 100:
        a2 = A_.copy()
        lg4 = df[["cik", "year", "ci_alt"]].copy(); lg4["year"] += 1
        a2 = a2.merge(lg4.rename(columns={"ci_alt": "ci_alt_L"}), on=["cik", "year"], how="left")
        a2 = a2.dropna(subset=["ci_alt_L", "roa"] + CT)
        if len(a2) > 60:
            a2["cb"] = a2.groupby("cik")["ci_alt_L"].transform("mean")
            a2["cd"] = a2["ci_alt_L"] - a2["cb"]
            Xa = sm.add_constant(a2[["cb", "cd"] + CT].astype(float))
            ra = sm.OLS(a2["roa"].values, Xa).fit(cov_type="cluster", cov_kwds={"groups": a2["cik"].values})
            ALT_NOTE.append(f"[P1] CI robustness (ccai_pca): between beta={ra.params['cb']:+.4f} "
                            f"(p={ra.pvalues['cb']:.3f}) · within beta={ra.params['cd']:+.4f} (p={ra.pvalues['cd']:.3f})")

    # ---------- (6) 2023 disclosure-rule confound ----------
    yr_mean = df.groupby("year")["eagi"].mean().round(3)
    ALT_NOTE.append("[P2] EAGI mean by year (SEC cyber-disclosure Item 106 took effect FY2023 — watch for a jump): "
                    + " ".join(f"{int(y)}={v:+.3f}" for y, v in yr_mean.items()))
    print("\n".join(ALT_NOTE))
    globals()["ALT_NOTE"] = ALT_NOTE
````

````python
# ============ 6c · COMPLIANCE CHANNEL — is the visible vocabulary the MEANINGFUL one? ============
# Conjecture: EA governance becomes OBSERVABLE precisely where regulation mandates disclosure.
# If so, the compliance-adjacent terms are not noise — they are the measurable manifestation, and
# the payoff should appear in TAIL RISK (absorbed mandates, avoided disruption) rather than in mean ROA.
# Competing explanation a referee will raise: those terms simply proxy REGULATORY EXPOSURE.
# Both are tested below; the year-break test is what distinguishes them.
COMP_NOTE = []
if G("Z") is None or G("df") is None:
    print("SKIPPED — index or panel unavailable.")
else:
    import statsmodels.api as sm
    COMP_TERMS = [t for t in ["data governance", "master data management", "it governance",
                              "technology governance"] if t in LEXICON]
    ARCH_TERMS = [t for t in LEXICON if t not in COMP_TERMS]
    if len(COMP_TERMS) >= 2 and len(ARCH_TERMS) >= 4:
        def _rel(cols):
            X = Z[cols]
            k = X.shape[1]; tot = X.sum(axis=1).var(ddof=1)
            al = float(k / (k - 1) * (1 - X.var(ddof=1).sum() / tot)) if tot else float("nan")
            C = X.corr().to_numpy(); iu = np.triu_indices(k, 1)
            return al, float(np.nanmean(C[iu]))
        a_c, r_c = _rel(COMP_TERMS); a_a, r_a = _rel(ARCH_TERMS)
        prev_c = float((counts[COMP_TERMS] > 0).mean().mean())
        prev_a = float((counts[ARCH_TERMS] > 0).mean().mean())
        COMP_NOTE.append(f"compliance terms {COMP_TERMS}: prevalence {prev_c:.1%} · α={a_c:.3f} · mean r={r_c:.3f}")
        COMP_NOTE.append(f"architecture terms (k={len(ARCH_TERMS)}): prevalence {prev_a:.1%} · α={a_a:.3f} · mean r={r_a:.3f}")

        sub = Z[["cik", "year"]].copy()
        sub["eagi_comp"] = Z[COMP_TERMS].mean(axis=1)
        sub["eagi_arch"] = Z[ARCH_TERMS].mean(axis=1)
        D = df.merge(sub, on=["cik", "year"], how="left")

        # (1) REGIME BREAK: SEC Item 106 took effect FY2023. A jump in compliance vocabulary that is
        # absent in architecture vocabulary means the terms track MANDATES, not practice.
        for col in ["eagi_comp", "eagi_arch"]:
            dd = D.dropna(subset=[col]).copy()
            if len(dd) < 100: continue
            dd["post23"] = (dd["year"] >= 2023).astype(float)
            dd["dm"] = dd[col] - dd.groupby("cik")[col].transform("mean")     # within firm
            X = sm.add_constant(dd[["post23"]].astype(float))
            r = sm.OLS(dd["dm"].values, X).fit(cov_type="cluster", cov_kwds={"groups": dd["cik"].values})
            COMP_NOTE.append(f"regime break {col}: post-2023 shift={r.params['post23']:+.4f} "
                             f"(p={r.pvalues['post23']:.3f})")
        COMP_NOTE.append("  → a break in compliance terms only ⇒ the vocabulary tracks MANDATES; "
                         "a break in both ⇒ a general disclosure trend; neither ⇒ practice-driven.")

        # (2) OUTCOME: mean ROA vs TAIL RISK. Compliance capability is insurance, so it should show
        # up in the worst year, not the average one.
        ad = D.dropna(subset=["roa", "eagi_comp"]).sort_values(["cik", "year"])
        g_ = ad.groupby("cik")
        firm = pd.DataFrame({
            "roa_mean": g_["roa"].mean(),
            "roa_worst_drop": g_["roa"].apply(lambda x: float(x.diff().min()) if len(x) > 2 else np.nan),
            "comp": g_["eagi_comp"].mean(), "arch": g_["eagi_arch"].mean(),
        }).reset_index()
        for c in ["size", "lev", "rnd"]:
            if c in ad.columns: firm[c] = g_[c].mean().values
        if "industry_raw" in ad.columns: firm["ind"] = g_["industry_raw"].first().values
        firm = firm.dropna(subset=["roa_worst_drop", "comp", "arch"])
        if len(firm) > 40:
            base = [c for c in ["size", "lev", "rnd"] if c in firm.columns]
            for dep in ["roa_mean", "roa_worst_drop"]:
                Xc = firm[["comp", "arch"] + base].astype(float)
                if "ind" in firm.columns and firm["ind"].nunique() > 1:
                    Xc = pd.concat([Xc, pd.get_dummies(firm["ind"].astype(str), prefix="i",
                                                       drop_first=True, dtype=float)], axis=1)
                rr = sm.OLS(firm[dep].values, sm.add_constant(Xc)).fit(cov_type="HC1")
                COMP_NOTE.append(f"{dep:15s} ~ compliance β={rr.params['comp']:+.4f} (p={rr.pvalues['comp']:.3f}) · "
                                 f"architecture β={rr.params['arch']:+.4f} (p={rr.pvalues['arch']:.3f}) · n={len(firm)}")
            COMP_NOTE.append("  → 'roa_worst_drop' is the largest one-year ROA fall: LESS NEGATIVE is better.")
            COMP_NOTE.append("  → compliance significant on tail risk but not on the mean would support the")
            COMP_NOTE.append("    insurance/absorption reading over the profitability reading.")
        globals()["COMP_NOTE"] = COMP_NOTE
        print("\n".join(COMP_NOTE))
    else:
        print("Not enough terms in either subset to split the lexicon.")
````

````python
# ============ 6d · H2 WITH A TESTABLE MODERATOR + FORMATIVE-vs-REFLECTIVE ============
# H2 died because EAGI has no reliability. But the moderator does not have to be text.
# The STRUCTURAL governance index (board technology committee + named CIO/CTO/CDO/Chief Architect)
# has real prevalence and is not word-frequency — so H2 becomes testable.
#
# MEASUREMENT-THEORY POINT that matters for BOTH papers:
#   REFLECTIVE construct → a latent trait causes the indicators → they must correlate → α applies.
#   FORMATIVE construct  → the indicators CONSTITUTE the construct → correlation is NOT required
#                          → α is the WRONG statistic.
# EAGI is reflective by design (every term should reflect one governance emphasis) → α=-0.06 is a
# genuine failure. The CIAOF dimensions and the structural index are FORMATIVE (a firm can be high
# on API integration and low on data harmonisation and still be "clean core") → α=0.299 is not a
# defect, it is expected. Formative constructs are validated by criterion/nomological validity and
# must ALSO be reported dimension by dimension.
H2S_NOTE = []
if G("df") is None:
    print("SKIPPED — no panel.")
else:
    import statsmodels.api as sm
    D = df.copy()
    st_cols = [c for c in ["techcom_strict", "techcom", "exec_cio", "exec_cto", "exec_cdo", "exec_arch"]
               if c in D.columns]
    st_cols = [c for c in st_cols if not (c == "techcom" and "techcom_strict" in st_cols)]
    if len(st_cols) >= 2 and "ci" in D.columns:
        D["gov_s"] = D[st_cols].fillna(0).sum(axis=1)
        D = D.sort_values(["cik", "year"])
        lg = D[["cik", "year", "gov_s", "ci"] + [c for c in ["size", "lev", "rnd"] if c in D.columns]].copy()
        lg["year"] += 1
        D = D.merge(lg.rename(columns={c: c + "_L" for c in ["gov_s", "ci"] +
                                       [x for x in ["size", "lev", "rnd"] if x in D.columns]}),
                    on=["cik", "year"], how="left")
        CT2 = [c + "_L" for c in ["size", "lev", "rnd"] if c + "_L" in D.columns]
        H = D.dropna(subset=["gov_s_L", "ci_L", "roa"] + CT2).copy()
        H2S_NOTE.append(f"structural moderator: {'+'.join(st_cols)} · mean={H['gov_s_L'].mean():.2f} "
                        f"· sd={H['gov_s_L'].std():.2f} · n={len(H)}")
        if len(H) > 80:
            H["ci_c"] = H["ci_L"] - H["ci_L"].mean()
            H["gv_c"] = H["gov_s_L"] - H["gov_s_L"].mean()
            H["ixn"] = H["ci_c"] * H["gv_c"]
            for lab, use_ind in [("year FE", False), ("industry+year FE", True)]:
                X = H[["ci_c", "gv_c", "ixn"] + CT2].astype(float)
                X = pd.concat([X, pd.get_dummies(H["year"].astype(str), prefix="y",
                                                 drop_first=True, dtype=float)], axis=1)
                if use_ind and "industry_raw" in H.columns and H["industry_raw"].nunique() > 1:
                    X = pd.concat([X, pd.get_dummies(H["industry_raw"].astype(str), prefix="i",
                                                     drop_first=True, dtype=float)], axis=1)
                r = sm.OLS(H["roa"].values, sm.add_constant(X)).fit(
                    cov_type="cluster", cov_kwds={"groups": H["cik"].values})
                H2S_NOTE.append(f"H2-structural ({lab}): β_ixn={r.params['ixn']:+.4f} p={r.pvalues['ixn']:.3f}"
                                f" | β_CI={r.params['ci_c']:+.4f} (p={r.pvalues['ci_c']:.3f})"
                                f" | β_gov={r.params['gv_c']:+.4f} (p={r.pvalues['gv_c']:.3f})")
            H2S_NOTE.append("  reading: β_ixn > 0 ⇒ adherence pays MORE where governance capacity exists"
                            " (the 'governed capacity to choose' proposition).")
            H2S_NOTE.append("  NOTE: the structural index is FORMATIVE — internal consistency (α) is not")
            H2S_NOTE.append("  the relevant validity test; criterion validity and dimension-level results are.")

    # ---- CIAOF dimension-by-dimension (required for a formative composite) ----
    kpis = [c for c in df.columns if str(c).startswith("kpi_") or str(c).endswith("_kpi")]
    if not kpis and G("panel") is not None:
        kpis = [c for c in panel.columns if str(c).endswith("_kpi")]
    if kpis:
        rows = []
        B = df.dropna(subset=["roa"]).copy()
        for k in kpis:
            if k not in B.columns: continue
            sub = B.dropna(subset=[k])
            if len(sub) < 80: continue
            sub = sub.copy()
            sub["kb"] = sub.groupby("cik")[k].transform("mean")
            X = sub[["kb"] + [c for c in ["size", "lev", "rnd"] if c in sub.columns]].astype(float)
            X = pd.concat([X, pd.get_dummies(sub["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
            if "industry_raw" in sub.columns and sub["industry_raw"].nunique() > 1:
                X = pd.concat([X, pd.get_dummies(sub["industry_raw"].astype(str), prefix="i",
                                                 drop_first=True, dtype=float)], axis=1)
            r = sm.OLS(sub["roa"].values, sm.add_constant(X)).fit(
                cov_type="cluster", cov_kwds={"groups": sub["cik"].values})
            # Raw-scale coefficients across dimensions are not comparable to each other or to the
            # composite: each KPI has its own dispersion. Report per-1sd(between) alongside the raw
            # coefficient so magnitudes are interpretable and implausible values are visible.
            _sdb = float(sub.groupby("cik")[k].mean().std())
            rows.append(dict(dimension=k, beta_between=round(float(r.params["kb"]), 4),
                             sd_between=round(_sdb, 4),
                             per_1sd=round(float(r.params["kb"]) * _sdb, 4),
                             p=round(float(r.pvalues["kb"]), 3), n=len(sub)))
        if rows:
            DIMS = pd.DataFrame(rows).sort_values("p")
            globals()["DIMS"] = DIMS
            H2S_NOTE.append("CIAOF dimension-level (between-firm, industry+year FE) — per_1sd is the")
            H2S_NOTE.append("interpretable magnitude; raw betas are on each dimension's own scale — required because")
            H2S_NOTE.append("the composite is formative and may average out opposing dimension effects:")
            for ln in DIMS.to_string(index=False).splitlines(): H2S_NOTE.append("  " + ln)
            try: DIMS.to_csv(f"{OUT_DIR}/ciaof_dimension_level.csv", index=False)
            except Exception: pass
    globals()["H2S_NOTE"] = H2S_NOTE
    print("\n".join(H2S_NOTE) if H2S_NOTE else "No structural moderator or KPI dimensions available.")
````

````python
# ============ 6e · VARIANCE BATTERY · PRE-SHOCK DESIGN · 2x2 HETEROGENEITY ============
# (a) EA value is an OPTION, not a margin: brakes do not raise average speed, they make speed
#     survivable. So the outcome set must be dispersion, depth and RECOVERY - not just the mean.
# (b) The strongest design available without external data: measure in 2019, observe the 2020
#     shock. Timing rules out reverse causality; selection on unobservables remains.
# (c) The 2x2 makes an empirical claim: standardisation should pay MORE where differentiation
#     intensity is low. That is testable as a heterogeneity interaction.
VAR_NOTE = []
if G("df") is None:
    print("SKIPPED — no panel.")
else:
    import statsmodels.api as sm
    D = df.sort_values(["cik", "year"]).copy()
    idxc = "eagi_best" if "eagi_best" in D.columns and D["eagi_best"].notna().sum() > 100 else "eagi"

    # ---------- (a) RICHER VARIANCE BATTERY ----------
    def firm_risk(x):
        r = x["roa"].astype(float).values
        if len(r) < 4: return pd.Series(dict(vol=np.nan, semidev=np.nan, mdd=np.nan, recov=np.nan))
        d_ = np.diff(r)
        neg = d_[d_ < 0]
        peak = np.maximum.accumulate(r)
        mdd = float(np.min(r - peak))                      # deepest fall below running peak
        # recovery: years from the worst trough back to (or above) the prior peak
        t = int(np.argmin(r - peak)); pk = float(peak[t]); rec = np.nan
        for j in range(t + 1, len(r)):
            if r[j] >= pk: rec = float(j - t); break
        if np.isnan(rec) and t < len(r) - 1: rec = float(len(r) - 1 - t)   # censored: not recovered
        return pd.Series(dict(vol=float(np.std(r, ddof=1)),
                              semidev=float(np.std(neg, ddof=1)) if len(neg) > 1 else np.nan,
                              mdd=mdd, recov=rec))
    R = D.dropna(subset=["roa"]).groupby("cik").apply(firm_risk).reset_index()
    X0 = D.groupby("cik").agg({idxc: "mean", "ci": "mean",
                               **{c: "mean" for c in ["ai_idx", "placebo"] if c in D.columns},
                               **{c: "mean" for c in ["size", "lev", "rnd"] if c in D.columns}}).reset_index()
    if "industry_raw" in D.columns:
        X0 = X0.merge(D.groupby("cik")["industry_raw"].first().reset_index(), on="cik", how="left")
    RB = R.merge(X0, on="cik", how="inner").dropna(subset=[idxc, "ci"])
    ctl = [c for c in ["size", "lev", "rnd"] if c in RB.columns]
    rows = []
    xvars = [idxc, "ci"] + [c for c in ["ai_idx", "placebo"] if c in RB.columns]
    for dep in ["vol", "semidev", "mdd", "recov"]:
        S = RB.dropna(subset=[dep])
        if len(S) < 40: continue
        for xv in [v for v in xvars if v in S.columns]:
            X = S[[xv] + ctl].astype(float)
            if "industry_raw" in S.columns and S["industry_raw"].nunique() > 1:
                X = pd.concat([X, pd.get_dummies(S["industry_raw"].astype(str), prefix="i",
                                                 drop_first=True, dtype=float)], axis=1)
            r_ = sm.OLS(S[dep].values, sm.add_constant(X)).fit(cov_type="HC1")
            rows.append(dict(risk_measure=dep, regressor=xv, beta=round(float(r_.params[xv]), 4),
                             p=round(float(r_.pvalues[xv]), 3), n=len(S)))
    if rows:
        VARTBL = pd.DataFrame(rows)
        globals()["VARTBL"] = VARTBL
        VAR_NOTE.append("Variance battery (firm level, +controls +industry FE):")
        for ln in VARTBL.to_string(index=False).splitlines(): VAR_NOTE.append("  " + ln)
        VAR_NOTE.append("  vol=SD of ROA · semidev=downside-only SD · mdd=deepest fall below running peak")
        VAR_NOTE.append("  recov=years back to prior peak (higher = slower recovery = worse)")
        try: VARTBL.to_csv(f"{OUT_DIR}/variance_battery.csv", index=False)
        except Exception: pass

    # ---------- (a2) RANDOMISATION INFERENCE on the headline variance result ----------
    # With α below zero, a significant coefficient deserves a non-parametric check: shuffle the
    # index across firms 2000x, breaking any firm-specific link while preserving its distribution.
    S0 = RB.dropna(subset=["vol", idxc])
    if len(S0) > 40:
        Xb = S0[ctl].astype(float)
        if "industry_raw" in S0.columns and S0["industry_raw"].nunique() > 1:
            Xb = pd.concat([Xb, pd.get_dummies(S0["industry_raw"].astype(str), prefix="i",
                                               drop_first=True, dtype=float)], axis=1)
        Xb = sm.add_constant(Xb)
        yv = S0["vol"].values
        def _b(xcol):
            XX = pd.concat([pd.Series(xcol, index=Xb.index, name="x"), Xb], axis=1)
            return float(np.linalg.pinv(XX.values.T @ XX.values) @ XX.values.T @ yv)[0] \
                if False else float((np.linalg.pinv(XX.values.T @ XX.values) @ XX.values.T @ yv)[0])
        obs = _b(S0[idxc].values)
        rs = np.random.default_rng(SEED if G("SEED") else 42)
        draws = np.array([_b(rs.permutation(S0[idxc].values)) for _ in range(2000)])
        p_ri = float((np.abs(draws) >= abs(obs)).mean())
        VAR_NOTE.append(f"RANDOMISATION INFERENCE (vol ~ {idxc}, 2000 permutations): "
                        f"observed β={obs:+.4f} · permutation p={p_ri:.3f}")
        VAR_NOTE.append("  a permutation p close to the analytic p means the result is not an artefact")
        VAR_NOTE.append("  of distributional assumptions; it does NOT address confounding.")

    # ---------- (b) PRE-SHOCK DESIGN: 2019 measure -> 2020 outcome ----------
    pre = D[D.year == 2019][["cik", idxc, "ci"] + ctl].dropna()
    if "industry_raw" in D.columns:
        pre = pre.merge(D.groupby("cik")["industry_raw"].first().reset_index(), on="cik", how="left")
    post = D[D.year == 2020][["cik", "roa"]].rename(columns={"roa": "roa20"})
    base19 = D[D.year == 2019][["cik", "roa"]].rename(columns={"roa": "roa19"})
    PS = pre.merge(post, on="cik").merge(base19, on="cik")
    PS["d_roa"] = PS["roa20"] - PS["roa19"]
    if len(PS) > 40:
        for xv in [idxc, "ci"]:
            X = PS[[xv, "roa19"] + ctl].astype(float)
            if "industry_raw" in PS.columns and PS["industry_raw"].nunique() > 1:
                X = pd.concat([X, pd.get_dummies(PS["industry_raw"].astype(str), prefix="i",
                                                 drop_first=True, dtype=float)], axis=1)
            r_ = sm.OLS(PS["d_roa"].values, sm.add_constant(X)).fit(cov_type="HC1")
            VAR_NOTE.append(f"PRE-SHOCK 2019→2020 (Δ ROA, industry FE, n={len(PS)}): {xv} "
                            f"β={r_.params[xv]:+.4f} p={r_.pvalues[xv]:.3f}")
        VAR_NOTE.append("  timing rules out reverse causality (a 2020 outcome cannot cause a 2019 measure);")
        VAR_NOTE.append("  selection on unobservables remains — this is stronger association, NOT causal.")

    # ---------- (c) 2x2 HETEROGENEITY: does standardisation pay more where differentiation is low? ----------
    if "rnd" in D.columns:
        H = D.dropna(subset=["ci", "roa", "rnd"]).copy()
        H["diff_int"] = H.groupby("cik")["rnd"].transform("mean")       # differentiation intensity proxy
        H["hi_diff"] = (H["diff_int"] > H["diff_int"].median()).astype(float)
        H["ci_b"] = H.groupby("cik")["ci"].transform("mean")
        H["ci_x_diff"] = (H["ci_b"] - H["ci_b"].mean()) * (H["hi_diff"] - H["hi_diff"].mean())
        Xh = H[["ci_b", "hi_diff", "ci_x_diff"] + [c for c in ctl if c in H.columns]].astype(float)
        Xh = pd.concat([Xh, pd.get_dummies(H["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
        if "industry_raw" in H.columns and H["industry_raw"].nunique() > 1:
            Xh = pd.concat([Xh, pd.get_dummies(H["industry_raw"].astype(str), prefix="i",
                                               drop_first=True, dtype=float)], axis=1)
        r_ = sm.OLS(H["roa"].values, sm.add_constant(Xh)).fit(cov_type="cluster",
                                                              cov_kwds={"groups": H["cik"].values})
        VAR_NOTE.append(f"2x2 HETEROGENEITY (CI × high-differentiation, R&D-intensity split, n={len(H)}): "
                        f"β_ixn={r_.params['ci_x_diff']:+.4f} p={r_.pvalues['ci_x_diff']:.3f} | "
                        f"β_CI(low-diff)={r_.params['ci_b']:+.4f} (p={r_.pvalues['ci_b']:.3f})")
        VAR_NOTE.append("  the grid predicts β_ixn < 0: standardisation should pay LESS where")
        VAR_NOTE.append("  differentiation intensity is high. This is the 2x2's testable content.")
    globals()["VAR_NOTE"] = VAR_NOTE
    print("\n".join(VAR_NOTE))
````

````python
# ============ 6f · AI READINESS — POSITIVE CONTROL AND PAPER 3 PROBE ============
# Two jobs, from the SAME cached filings, the same firms, the same method:
#
# (1) POSITIVE CONTROL — the strongest available defence of Paper 2. If an AI lexicon built
#     identically reaches respectable prevalence and reliability while the EA lexicon does not,
#     the failure is CONSTRUCT-SPECIFIC, not method-specific. No referee can then argue
#     "you used bad text methods": the same methods worked on the neighbouring construct.
#
# (2) PAPER 3 PROBE — does an EA / core-integrity baseline predict AI adoption, and does it
#     moderate whether AI adoption pays? "Strong architecture is the prerequisite for
#     capitalising on AI" is a moderation claim and it is testable here.
AI_NOTE = []
AI_LEX = ["artificial intelligence", "machine learning", "generative ai", "large language model",
          "foundation model", "predictive analytics", "natural language processing", "automation platform",
          "data lakehouse", "data platform", "cloud data warehouse", "algorithmic decision"]
if G("counts") is None or G("CACHE_DIR") is None or G("df") is None:
    print("SKIPPED — text layer or panel unavailable.")
else:
    import statsmodels.api as sm
    ai_path = f"{CACHE_DIR}/ai_counts.csv"
    if os.path.exists(ai_path):
        aic = pd.read_csv(ai_path); print(f"Reusing AI counts — {len(aic)} firm-years")
    else:
        recs = []
        for cik in tqdm(FIRM_CIKS, desc="AI lexicon"):
            try: fl = list(filings_of(cik, "10-K"))
            except Exception: continue
            for fy, acc, doc in fl:
                if not (YEAR_MIN <= fy <= YEAR_MAX): continue
                fp = f"{CACHE_DIR}/txt_{cik}_{acc}.txt"
                if not os.path.exists(fp): continue
                t = open(fp, encoding="utf-8", errors="ignore").read()
                c = count_terms(t, AI_LEX); c.update({"cik": cik, "year": fy}); recs.append(c)
        aic = pd.DataFrame(recs).groupby(["cik", "year"], as_index=False).first() if recs else pd.DataFrame()
        if len(aic): aic.to_csv(ai_path, index=False)
    if len(aic):
        prev_ai = float((aic[AI_LEX] > 0).mean().mean())
        prev_ea = float((counts[LEXICON] > 0).mean().mean())
        Xa = np.log1p(aic[AI_LEX].div(aic["_words"], axis=0) * 10000)
        Za = (Xa - Xa.mean()) / Xa.std().replace(0, 1)
        k = Za.shape[1]; tot = Za.sum(axis=1).var(ddof=1)
        al = float(k / (k - 1) * (1 - Za.var(ddof=1).sum() / tot)) if tot else np.nan
        Cm = Za.corr().to_numpy(); iu = np.triu_indices(k, 1); mr = float(np.nanmean(Cm[iu]))
        h1 = Za.iloc[:, 0::2].mean(axis=1); h2 = Za.iloc[:, 1::2].mean(axis=1)
        sh = float(h1.corr(h2)); sb = 2 * sh / (1 + sh) if sh == sh and sh > -1 else np.nan
        AI_NOTE.append(f"POSITIVE CONTROL — same corpus, same method, different construct:")
        AI_NOTE.append(f"  AI lexicon (k={k}): mean term prevalence {prev_ai:.1%} · α={al:.3f} · "
                       f"mean r={mr:.3f} · Spearman-Brown={sb:.3f}")
        AI_NOTE.append(f"  EA lexicon (k={len(LEXICON)}): mean term prevalence {prev_ea:.1%} · "
                       f"α={(G('LEX_ALPHA') if G('LEX_ALPHA') is not None else float('nan')):.3f}")
        if sb == sb and sb > 0.5 and (G("LEX_ALPHA") or 0) < 0.2:
            AI_NOTE.append("  ⇒ THE METHOD WORKS. The EA null is construct-specific, not method-specific.")
            AI_NOTE.append("    This is the answer to 'you just used bad text analysis'.")
        aidx = aic[["cik", "year"]].copy(); aidx["ai_idx"] = Za.mean(axis=1).values
        A2 = df.merge(aidx, on=["cik", "year"], how="left")
        globals()["df"] = A2
        # ---- PAPER 3 PROBE ----
        W = A2.sort_values(["cik", "year"]).copy()
        gcol = "gov_s" if "gov_s" in W.columns else None
        if gcol is None:
            st = [c for c in ["techcom_strict", "exec_cio", "exec_cto", "exec_cdo", "exec_arch"] if c in W.columns]
            if len(st) >= 2: W["gov_s"] = W[st].fillna(0).sum(axis=1); gcol = "gov_s"
        keep = [c for c in ["ci", gcol, "size", "lev", "rnd"] if c]
        lg = W[["cik", "year"] + keep].copy(); lg["year"] += 1
        W = W.merge(lg.rename(columns={c: c + "_L" for c in keep}), on=["cik", "year"], how="left")
        ctl = [c + "_L" for c in ["size", "lev", "rnd"] if c + "_L" in W.columns]
        # EAGI lag must exist BEFORE the first-stage loop: it was created between the two loops,
        # so `eagi_mod_L in W.columns` was False and the EAGI arm was skipped in silence.
        if "eagi_best" in W.columns or "eagi" in W.columns:
            _e = "eagi_best" if "eagi_best" in W.columns else "eagi"
            _lg = W[["cik", "year", _e]].copy(); _lg["year"] += 1
            W = W.merge(_lg.rename(columns={_e: "eagi_mod_L"}), on=["cik", "year"], how="left")
        _arms = [c for c in ["ci_L", (gcol + "_L") if gcol else None, "eagi_mod_L"]
                 if c and c in W.columns]
        AI_NOTE.append(f"  arms available: {_arms}")
        # (i) does an architecture baseline predict later AI adoption?
        for xv in _arms:
            S = W.dropna(subset=[xv, "ai_idx"] + ctl)
            if len(S) < 80: continue
            X = S[[xv] + ctl].astype(float)
            X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
            if "industry_raw" in S.columns and S["industry_raw"].nunique() > 1:
                X = pd.concat([X, pd.get_dummies(S["industry_raw"].astype(str), prefix="i",
                                                 drop_first=True, dtype=float)], axis=1)
            r = sm.OLS(S["ai_idx"].values, sm.add_constant(X)).fit(
                cov_type="cluster", cov_kwds={"groups": S["cik"].values})
            AI_NOTE.append(f"PAPER 3 (i) {xv} → AI adoption: β={r.params[xv]:+.4f} p={r.pvalues[xv]:.3f} n={len(S)}")
        # (ii) does an architecture baseline MODERATE whether AI adoption pays?
        for mod in _arms:
            S = W.dropna(subset=["ai_idx", mod, "roa"] + ctl).copy()
            if len(S) < 80: continue
            S["ai_c"] = S["ai_idx"] - S["ai_idx"].mean()
            S["m_c"] = S[mod] - S[mod].mean()
            S["ixn"] = S["ai_c"] * S["m_c"]
            X = S[["ai_c", "m_c", "ixn"] + ctl].astype(float)
            X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
            if "industry_raw" in S.columns and S["industry_raw"].nunique() > 1:
                X = pd.concat([X, pd.get_dummies(S["industry_raw"].astype(str), prefix="i",
                                                 drop_first=True, dtype=float)], axis=1)
            r = sm.OLS(S["roa"].values, sm.add_constant(X)).fit(
                cov_type="cluster", cov_kwds={"groups": S["cik"].values})
            AI_NOTE.append(f"PAPER 3 (ii) AI × {mod} → ROA: β_ixn={r.params['ixn']:+.4f} "
                           f"p={r.pvalues['ixn']:.3f} | β_AI={r.params['ai_c']:+.4f} (p={r.pvalues['ai_c']:.3f})")
        AI_NOTE.append("  β_ixn > 0 ⇒ AI adoption pays MORE where the architecture baseline is stronger —")
        AI_NOTE.append("  i.e. architecture as the prerequisite for capitalising on AI. β_AI alone is adoption talk.")
        globals()["AI_NOTE"] = AI_NOTE
        print("\n".join(AI_NOTE))
````

````python
# ============ 6g · P1-H3 · ALLOCATION, NOT LEVEL ============
# P1-H2 (inverted-U) asked whether there is an optimal LEVEL of adherence. Rejected.
# P1-H3 asks a different question: at a GIVEN level, does SELECTIVE adherence beat UNIFORM
# adherence? The practitioner claim behind the 2x2 is that standard-everywhere and
# custom-everywhere are both corner solutions, and that the payoff comes from standardising
# supporting capabilities while differentiating revenue-critical ones.
#
#   level      = mean across the five CIAOF dimensions   (how much adherence)
#   dispersion = SD across the five dimensions           (how SELECTIVELY it is applied)
#
# MECHANICAL CAVEAT: with bounded indicators, dispersion is constrained by the level (a firm at
# 0.95 cannot be as dispersed as one at 0.50). Dispersion is therefore residualised on level and
# level-squared before use, so the coefficient is not a repackaged level effect.
H3_NOTE = []
kpic = [c for c in (df.columns if G("df") is not None else []) if str(c).startswith("kpi_")]
if G("df") is None or len(kpic) < 3:
    print("SKIPPED — fewer than three CIAOF dimensions in the panel.")
else:
    import statsmodels.api as sm
    D = df.copy()
    K = D[kpic].astype(float)
    Kz = (K - K.mean()) / K.std().replace(0, 1)          # comparable units across dimensions
    D["ci_level"] = Kz.mean(axis=1)
    D["ci_disp_raw"] = Kz.std(axis=1)
    S = D.dropna(subset=["ci_level", "ci_disp_raw", "roa"]).copy()
    H3_NOTE.append(f"dimensions used (k={len(kpic)}): {[c.replace('kpi_','') for c in kpic]}")
    if len(S) > 100:
        Xr = sm.add_constant(pd.DataFrame({"l": S["ci_level"], "l2": S["ci_level"] ** 2}))
        S["ci_disp"] = S["ci_disp_raw"].values - sm.OLS(S["ci_disp_raw"].values, Xr).fit().predict(Xr)
        H3_NOTE.append(f"dispersion residualised on level and level² · "
                       f"corr(level, raw disp)={S['ci_level'].corr(S['ci_disp_raw']):+.3f} → "
                       f"corr(level, residual disp)={S['ci_level'].corr(S['ci_disp']):+.3f}")
        ctl = [c for c in ["size", "lev", "rnd"] if c in S.columns]
        def blk(SS, cols):
            X = SS[cols + ctl].astype(float)
            X = pd.concat([X, pd.get_dummies(SS["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
            if "industry_raw" in SS.columns and SS["industry_raw"].nunique() > 1:
                X = pd.concat([X, pd.get_dummies(SS["industry_raw"].astype(str), prefix="i",
                                                 drop_first=True, dtype=float)], axis=1)
            return sm.OLS(SS["roa"].values, sm.add_constant(X)).fit(
                cov_type="cluster", cov_kwds={"groups": SS["cik"].values})
        # between-firm (the theoretically primary estimator for a firm characteristic)
        B = S.copy()
        B["lvl_b"] = B.groupby("cik")["ci_level"].transform("mean")
        B["dsp_b"] = B.groupby("cik")["ci_disp"].transform("mean")
        r = blk(B, ["lvl_b", "dsp_b"])
        H3_NOTE.append(f"P1-H3 between-firm: β_level={r.params['lvl_b']:+.4f} (p={r.pvalues['lvl_b']:.3f}) · "
                       f"β_DISPERSION={r.params['dsp_b']:+.4f} (p={r.pvalues['dsp_b']:.3f}) n={len(B)}")
        # is there an interior optimum in SELECTIVITY (the shape claim, moved to the right variable)?
        B["dsp_b2"] = B["dsp_b"] ** 2
        r2 = blk(B, ["lvl_b", "dsp_b", "dsp_b2"])
        b1, b2, p2 = r2.params["dsp_b"], r2.params["dsp_b2"], r2.pvalues["dsp_b2"]
        nt = f"P1-H3 shape in dispersion: b1={b1:+.4f} b2={b2:+.4f} (p={p2:.3f})"
        if b2 < 0 and p2 < 0.10:
            vx = -b1 / (2 * b2); lo, hi = B["dsp_b"].quantile([0.05, 0.95])
            nt += (f" · vertex={vx:+.3f} {'INSIDE' if lo <= vx <= hi else 'OUTSIDE'} the 5–95% range"
                   + ("" if lo <= vx <= hi else " → not an interior optimum"))
        else:
            nt += " · not concave → no interior optimum in selectivity"
        H3_NOTE.append(nt)
        # corner comparison: the 2x2 in its own terms
        B["hl"] = (B["lvl_b"] > B["lvl_b"].median()).map({True: "high-adherence", False: "low-adherence"})
        B["hd"] = (B["dsp_b"] > B["dsp_b"].median()).map({True: "SELECTIVE", False: "uniform"})
        cell_ = B.groupby(["hl", "hd"])["roa"].agg(["mean", "count"]).round(4)
        H3_NOTE.append("corner comparison (raw means — descriptive only, no controls):")
        for ln in cell_.to_string().splitlines(): H3_NOTE.append("  " + ln)
        # does selectivity pay MORE where there is more to differentiate on?
        if "rnd" in B.columns:
            B["dif"] = B.groupby("cik")["rnd"].transform("mean")
            B["dsp_x_dif"] = (B["dsp_b"] - B["dsp_b"].mean()) * (B["dif"] - B["dif"].mean())
            r3 = blk(B, ["lvl_b", "dsp_b", "dif", "dsp_x_dif"])
            H3_NOTE.append(f"P1-H3 interaction (dispersion × differentiation potential): "
                           f"β={r3.params['dsp_x_dif']:+.4f} (p={r3.pvalues['dsp_x_dif']:.3f})")
            H3_NOTE.append("  positive ⇒ selectivity pays more where there is more to differentiate on —")
            H3_NOTE.append("  the empirical content of the matrix, and the case for an architect in the loop.")
        # ---------- P1-H3c · IS MORE ALWAYS BETTER? A BIN TEST WITH NO FUNCTIONAL FORM ----------
        # The inverted-U asked this through a quadratic, and a quadratic vertex is an algebraic
        # artefact as often as an optimum. Bins answer it directly: estimate an adjusted mean per
        # adherence quintile and see whether the TOP quintile is actually the best one.
        # NOTE ON REACH: this can only speak to the range the data contains. If no firm sits near
        # the ceiling, nothing here licenses a claim about 100%.
        try:
            # qcut assigns Q1 to the LOWEST quintile. Label by direction so the reference
            # group cannot be misread as "top adherence".
            _ql = ["Q1_lowest_adh", "Q2", "Q3", "Q4", "Q5_highest_adh"]
            B["q5"] = pd.qcut(B["lvl_b"], 5, labels=_ql, duplicates="drop").astype(str)
            if B["q5"].nunique() >= 4:
                Xq = pd.get_dummies(B["q5"], prefix="q", drop_first=True, dtype=float)
                Xq = pd.concat([Xq, B[ctl].astype(float)], axis=1)
                Xq = pd.concat([Xq, pd.get_dummies(B["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
                if "industry_raw" in B.columns and B["industry_raw"].nunique() > 1:
                    Xq = pd.concat([Xq, pd.get_dummies(B["industry_raw"].astype(str), prefix="i",
                                                       drop_first=True, dtype=float)], axis=1)
                rq = sm.OLS(B["roa"].values, sm.add_constant(Xq)).fit(
                    cov_type="cluster", cov_kwds={"groups": B["cik"].values})
                base_q = sorted(B["q5"].unique())[0]
                eff = {base_q: 0.0}
                for qq in sorted(B["q5"].unique())[1:]:
                    kx = "q_" + qq
                    if kx in rq.params.index: eff[qq] = float(rq.params[kx])
                best = max(eff, key=eff.get); top = sorted(eff)[-1]
                H3_NOTE.append("P1-H3c adherence QUINTILES (adjusted ROA vs " + base_q +
                               ", industry+year FE) — Q1 = LOWEST adherence, Q5 = HIGHEST:")
                for qq in sorted(eff):
                    pv = rq.pvalues.get("q_" + qq, float("nan"))
                    H3_NOTE.append(f"    {qq}: {eff[qq]:+.4f}" + (f" (p={pv:.3f})" if pv == pv else " (reference)")
                                   + ("   ← highest" if qq == best else ""))
                H3_NOTE.append(f"  observed adherence range: {B['lvl_b'].min():+.2f} to {B['lvl_b'].max():+.2f} "
                               f"(z units) — claims are limited to this range.")
                _nsig = sum(1 for qq in eff if qq != base_q and rq.pvalues.get("q_" + qq, 1) < 0.0125)
                H3_NOTE.append(f"  multiplicity: 4 contrasts vs the reference → Bonferroni 5% is p<0.0125; "
                               f"{_nsig}/4 survive.")
                if _nsig == 0:
                    H3_NOTE.append("  ⇒ NO contrast survives adjustment: quintile differences are small and")
                    H3_NOTE.append("    non-monotonic. Report as 'no reliable gradient', not as a shape claim.")
                if best != top:
                    H3_NOTE.append(f"  ⇒ the HIGHEST-adherence quintile is not the best ({best} is). Within the")
                    H3_NOTE.append("    more adherence is not monotonically better — and this is shown WITHOUT")
                    H3_NOTE.append("    imposing a quadratic, so it is not a vertex artefact.")
                else:
                    H3_NOTE.append("  ⇒ the top quintile IS the best: no evidence against 'more is better' here.")
        except Exception as e:
            H3_NOTE.append("quintile test unavailable: " + type(e).__name__)

        # ---------- P1-H3b · DIRECTIONAL pattern, not just dispersion ----------
        # Clean-core logic does not say "deviate somewhere". It says deviation is LEGITIMATE on the
        # process dimension (that is where differentiation lives) and DEBT on the four technical
        # hygiene dimensions. So the predicted profile is: high technical hygiene, selectively lower
        # process standardisation. PRE-SPECIFIED, and sensitivity-tested by swapping the roles.
        PROC_DIM = [c for c in kpic if "standard" in c.lower()]
        HYG_DIM = [c for c in kpic if c not in PROC_DIM]
        if len(PROC_DIM) == 1 and len(HYG_DIM) >= 3:
            Z2 = Kz.copy()
            B["hyg"] = Z2[HYG_DIM].mean(axis=1).groupby(D["cik"]).transform("mean").reindex(B.index)
            B["prc"] = Z2[PROC_DIM[0]].groupby(D["cik"]).transform("mean").reindex(B.index)
            B2 = B.dropna(subset=["hyg", "prc"]).copy()
            if len(B2) > 100:
                B2["pattern"] = B2["hyg"] - B2["prc"]          # high hygiene, low process lock-in
                r4 = blk(B2, ["lvl_b", "pattern"])
                H3_NOTE.append(f"P1-H3b DIRECTIONAL pattern (technical hygiene minus process standardisation): "
                               f"β={r4.params['pattern']:+.4f} (p={r4.pvalues['pattern']:.3f}) n={len(B2)}")
                r5 = blk(B2, ["hyg", "prc"])
                H3_NOTE.append(f"  decomposed: β_hygiene={r5.params['hyg']:+.4f} (p={r5.pvalues['hyg']:.3f}) · "
                               f"β_process_std={r5.params['prc']:+.4f} (p={r5.pvalues['prc']:.3f})")
                H3_NOTE.append("  predicted: β_hygiene > 0 (technical deviation is debt) and β_process_std ≤ 0")
                H3_NOTE.append("  (process deviation is where differentiation legitimately lives).")
                # SENSITIVITY: swap the roles. If the reversed pattern performs just as well, the
                # dimension-to-role assignment is doing no work and the test is uninformative.
                B2["pattern_rev"] = B2["prc"] - B2["hyg"]
                r6 = blk(B2, ["lvl_b", "pattern_rev"])
                H3_NOTE.append(f"  SENSITIVITY (roles reversed): β={r6.params['pattern_rev']:+.4f} "
                               f"(p={r6.pvalues['pattern_rev']:.3f}) — by construction the mirror image; "
                               "report both so the theoretical assignment is visible, not hidden.")
        # ---------- P1-H3d · EACH FIRM'S OWN 100% — FIT TO POTENTIAL, NOT ABSOLUTE LEVEL ----------
        # An absolute optimum assumes every firm faces the same feasible maximum. They do not:
        # a firm whose portfolio is 40% genuinely differentiating cannot and should not reach the
        # adherence of one that is 5% differentiating. Both may sit exactly at THEIR optimum while
        # occupying different absolute levels — and pooling them MECHANICALLY FLATTENS the curve.
        # So the variable of interest is the GAP between actual adherence and the level the firm's
        # own business model implies. Predicted optimum: gap = 0, not level = maximum.
        try:
            P = B.dropna(subset=["lvl_b"]).copy()
            prox = [c for c in ["rnd", "size", "lev"] if c in P.columns]     # differentiation potential
            if len(prox) >= 2 and len(P) > 100:
                Xp = P[prox].astype(float)
                if "industry_raw" in P.columns and P["industry_raw"].nunique() > 1:
                    Xp = pd.concat([Xp, pd.get_dummies(P["industry_raw"].astype(str), prefix="p",
                                                       drop_first=True, dtype=float)], axis=1)
                Xp = sm.add_constant(Xp)
                fitm = sm.OLS(P["lvl_b"].values, Xp).fit()
                P["expected_adh"] = fitm.predict(Xp)          # what this firm's model implies
                P["gap"] = P["lvl_b"] - P["expected_adh"]      # + = over-adhering, - = under-adhering
                P["gap2"] = P["gap"] ** 2
                P["absgap"] = P["gap"].abs()
                H3_NOTE.append(f"P1-H3d FIT-TO-POTENTIAL · expected adherence modelled from "
                               f"{prox} + industry (R²={fitm.rsquared:.3f})")
                rg = blk(P, ["gap", "gap2"])
                b1g, b2g, p2g = rg.params["gap"], rg.params["gap2"], rg.pvalues["gap2"]
                nt2 = f"  quadratic in the GAP: b1={b1g:+.4f} b2={b2g:+.4f} (p={p2g:.3f})"
                if b2g < 0 and p2g < 0.10:
                    vx = -b1g / (2 * b2g)
                    nt2 += f" · peak at gap={vx:+.3f} (predicted ≈ 0 ⇒ being at YOUR OWN level is what pays)"
                else:
                    nt2 += " · not concave → no firm-specific optimum detected"
                H3_NOTE.append(nt2)
                ra = blk(P, ["absgap"])
                H3_NOTE.append(f"  distance from own level: β_|gap|={ra.params['absgap']:+.4f} "
                               f"(p={ra.pvalues['absgap']:.3f}) — NEGATIVE ⇒ deviating in EITHER direction costs")
                P["gq"] = pd.qcut(P["gap"], 5, labels=["most under", "under", "at level", "over", "most over"],
                                  duplicates="drop").astype(str)
                gm = P.groupby("gq")["roa"].agg(["mean", "count"]).round(4)
                H3_NOTE.append("  raw ROA by gap quintile (descriptive, no controls):")
                for ln in gm.to_string().splitlines(): H3_NOTE.append("    " + ln)
                # ATYPICALITY PLACEBO: |gap| is a residual, so firms that observables predict
                # POORLY have large |gap| by construction — and atypical firms (conglomerates,
                # restructurings) may perform differently for reasons unrelated to architecture.
                # Build the identical gap from a measure that should carry no information.
                if "placebo" in B.columns:
                    P["plc_b"] = B["placebo"].groupby(B["cik"]).transform("mean").reindex(P.index)
                    Pp = P.dropna(subset=["plc_b"]).copy()
                    if len(Pp) > 100:
                        fitp = sm.OLS(Pp["plc_b"].values, sm.add_constant(Pp[prox].astype(float))).fit()
                        Pp["gap_plc"] = Pp["plc_b"] - fitp.predict(sm.add_constant(Pp[prox].astype(float)))
                        Pp["absgap_plc"] = Pp["gap_plc"].abs()
                        rp = blk(Pp, ["absgap_plc"])
                        H3_NOTE.append(f"  ATYPICALITY PLACEBO: β_|placebo gap|={rp.params['absgap_plc']:+.4f} "
                                       f"(p={rp.pvalues['absgap_plc']:.3f})")
                        if rp.pvalues["absgap_plc"] < 0.10:
                            H3_NOTE.append("    ⚠ a meaningless measure's gap ALSO predicts performance ⇒ |gap| is")
                            H3_NOTE.append("      capturing ATYPICALITY, not architectural misfit. Do not interpret H3d.")
                        else:
                            H3_NOTE.append("    ✓ placebo gap is null ⇒ the misfit reading survives the atypicality critique.")
                H3_NOTE.append("  NOTE: an absolute-scale null is NOT evidence against a firm-specific optimum;")
                H3_NOTE.append("  pooling firms with different feasible maxima flattens the curve by construction.")
                H3_NOTE.append("  CIRCULARITY WARNING: 'expected' adherence is fitted from observables, so the gap")
                H3_NOTE.append("  is a residual, not a normative target. It shows dispersion around peer-typical")
                H3_NOTE.append("  behaviour — it cannot say what a firm SHOULD do. Only the workshops can.")
        except Exception as e:
            H3_NOTE.append("fit-to-potential test unavailable: " + type(e).__name__)

        H3_NOTE.append("CAVEAT: even H3b is a DIMENSION-level proxy. 'Differentiating on the right PROCESSES'")
        H3_NOTE.append("needs firm × process × treatment data, which no public source provides. That is the")
        H3_NOTE.append("data the TRVA workshops generate — the tool is the missing instrument, not a by-product.")
    globals()["H3_NOTE"] = H3_NOTE
    print("\n".join(H3_NOTE))
````

````python
# ============ 6h · STRUCTURE OF THE COMPOSITE — four questions H3 does not answer ============
# H3 asked whether SPREAD pays. Four further structural questions, each with a different
# practitioner consequence, and two of them make OPPOSITE predictions to H3:
#   A WEAKEST LINK   is the binding constraint the MINIMUM dimension rather than the mean?
#                    → justifies a floor guardrail ("no dimension below X"), not an aggregate target
#   B COMPLEMENTARITY do dimensions REINFORCE each other? If so, spread HURTS — the direct
#                    competitor to H3. Running both is a horse race, not a confirmation exercise.
#   C IMPORTANCE     Shapley decomposition of R² gives each dimension a defensible % weight,
#                    replacing equal weighting. Gated: decomposing a null model decomposes noise.
#   D NECESSITY      is a dimension a PREREQUISITE (high outcome impossible without it) rather
#                    than a contributor? Necessity and sufficiency are different claims.
STRUCT_NOTE = []
kpic2 = [c for c in (df.columns if G("df") is not None else []) if str(c).startswith("kpi_")]
if G("df") is None or len(kpic2) < 3:
    print("SKIPPED — dimensions unavailable.")
else:
    import statsmodels.api as sm
    from itertools import combinations
    from math import factorial   # np.math was removed in NumPy 2
    D = df.copy()
    K = D[kpic2].astype(float); Kz = (K - K.mean()) / K.std().replace(0, 1)
    for c in kpic2: D["z_" + c] = Kz[c]
    zc = ["z_" + c for c in kpic2]
    F = D.groupby("cik")[zc + ["roa"] + [x for x in ["size", "lev", "rnd"] if x in D.columns]].mean().dropna()
    if "industry_raw" in D.columns:
        F = F.join(D.groupby("cik")["industry_raw"].first())
    ctl = [c for c in ["size", "lev", "rnd"] if c in F.columns]
    def design(S, cols):
        X = S[cols + ctl].astype(float)
        if "industry_raw" in S.columns and S["industry_raw"].nunique() > 1:
            X = pd.concat([X, pd.get_dummies(S["industry_raw"].astype(str), prefix="i",
                                             drop_first=True, dtype=float)], axis=1)
        return sm.add_constant(X)
    def r2(S, cols):
        X = design(S, cols).values; y = S["roa"].values
        b = np.linalg.pinv(X.T @ X) @ X.T @ y; e = y - X @ b
        return 1 - float(e @ e) / float(((y - y.mean()) ** 2).sum())
    if len(F) > 60:
        F["mean_d"] = F[zc].mean(axis=1); F["min_d"] = F[zc].min(axis=1)
        F["max_d"] = F[zc].max(axis=1)      # CEILING: is performance driven by the BEST dimension?
        # A · weakest link vs mean
        rm = sm.OLS(F["roa"].values, design(F, ["mean_d"])).fit(cov_type="HC1")
        rn = sm.OLS(F["roa"].values, design(F, ["min_d"])).fit(cov_type="HC1")
        rb = sm.OLS(F["roa"].values, design(F, ["mean_d", "min_d"])).fit(cov_type="HC1")
        STRUCT_NOTE.append(f"A WEAKEST LINK · mean-only β={rm.params['mean_d']:+.4f} (p={rm.pvalues['mean_d']:.3f}) · "
                           f"min-only β={rn.params['min_d']:+.4f} (p={rn.pvalues['min_d']:.3f})")
        STRUCT_NOTE.append(f"   both entered: β_mean={rb.params['mean_d']:+.4f} (p={rb.pvalues['mean_d']:.3f}) · "
                           f"β_min={rb.params['min_d']:+.4f} (p={rb.pvalues['min_d']:.3f})")
        rx = sm.OLS(F["roa"].values, design(F, ["max_d"])).fit(cov_type="HC1")
        rt = sm.OLS(F["roa"].values, design(F, ["mean_d", "min_d", "max_d"])).fit(cov_type="HC1")
        STRUCT_NOTE.append(f"A2 CEILING · max-only β={rx.params['max_d']:+.4f} (p={rx.pvalues['max_d']:.3f})")
        STRUCT_NOTE.append(f"   all three entered: mean={rt.params['mean_d']:+.4f} (p={rt.pvalues['mean_d']:.3f}) · "
                           f"min={rt.params['min_d']:+.4f} (p={rt.pvalues['min_d']:.3f}) · "
                           f"max={rt.params['max_d']:+.4f} (p={rt.pvalues['max_d']:.3f})")
        _c = F[["mean_d", "min_d", "max_d"]].corr()
        STRUCT_NOTE.append(f"   ⚠ mean/min/max are ORDER STATISTICS OF THE SAME FIVE NUMBERS "
                           f"(corr mean-min={_c.loc['mean_d','min_d']:+.2f}, mean-max={_c.loc['mean_d','max_d']:+.2f}). "
                           "Entering all three")
        STRUCT_NOTE.append("     together is near-collinear: signs can flip between the single-regressor and")
        STRUCT_NOTE.append("     joint specifications. Read the SINGLE-regressor results as primary and treat the")
        STRUCT_NOTE.append("     joint decomposition as indicative only.")
        STRUCT_NOTE.append("   MIN wins  → a FLOOR binds: enforce a minimum per dimension, not an aggregate target.")
        STRUCT_NOTE.append("   MAX wins  → a CEILING logic: one strong dimension carries the firm; the rest are")
        STRUCT_NOTE.append("               hygiene and need only be adequate.")
        STRUCT_NOTE.append("   MEAN wins → dimensions are substitutable and an aggregate score is defensible.")
        STRUCT_NOTE.append("   all null  → the composite has no internal structure worth encoding in the tool.")
        # B · complementarity horse race against H3
        pairs = list(combinations(zc, 2))[:10]
        for a_, b_ in pairs:
            F[f"x_{a_}_{b_}"] = (F[a_] - F[a_].mean()) * (F[b_] - F[b_].mean())
        ixn = [f"x_{a_}_{b_}" for a_, b_ in pairs]
        rc = sm.OLS(F["roa"].values, design(F, zc + ixn)).fit(cov_type="HC1")
        pos = int(sum(1 for k in ixn if rc.params[k] > 0)); sig = int(sum(1 for k in ixn if rc.pvalues[k] < 0.05))
        STRUCT_NOTE.append(f"B COMPLEMENTARITY · {len(ixn)} pairwise interactions: {pos} positive, {sig} significant at 5%")
        STRUCT_NOTE.append("   mostly POSITIVE ⇒ dimensions reinforce each other ⇒ spread HURTS (contradicts H3);")
        STRUCT_NOTE.append("   mostly null ⇒ dimensions are separable ⇒ selective allocation is coherent.")
        # C · Shapley importance weights (gated on the joint model being informative)
        full = r2(F, zc); basec = r2(F, [])
        if full - basec > 0.01:
            sh = {c: 0.0 for c in zc}
            for c in zc:
                others = [x for x in zc if x != c]
                for k in range(len(others) + 1):
                    for sub in combinations(others, k):
                        w = factorial(k) * factorial(len(zc) - k - 1) / factorial(len(zc))
                        sh[c] += w * (r2(F, list(sub) + [c]) - r2(F, list(sub)))
            tot = sum(max(v, 0) for v in sh.values()) or 1.0
            STRUCT_NOTE.append(f"C IMPORTANCE (Shapley share of explained variance; model R²={full:.3f}):")
            for c, v in sorted(sh.items(), key=lambda t: -t[1]):
                STRUCT_NOTE.append(f"     {c.replace('z_kpi_','').replace('_kpi',''):26s} {max(v,0)/tot:6.1%}")
            STRUCT_NOTE.append("   these are DEFENSIBLE weights to replace equal weighting — but they are")
            STRUCT_NOTE.append("   sample-specific and must be reported as such, never as universal constants.")
        else:
            STRUCT_NOTE.append(f"C IMPORTANCE · skipped: dimensions jointly explain only {full - basec:.3f} of "
                               "variance. Decomposing a null model decomposes noise.")
        # D · necessity (prerequisite) vs contribution
        hi = F[F["roa"] >= F["roa"].quantile(0.75)]
        STRUCT_NOTE.append("D NECESSITY · 5th percentile of each dimension, all firms vs HIGH-ROA firms:")
        for c in zc:
            a_all = float(F[c].quantile(0.05)); a_hi = float(hi[c].quantile(0.05))
            flag = "  ← acts as a FLOOR" if (a_hi - a_all) > 0.5 else ""
            STRUCT_NOTE.append(f"     {c.replace('z_kpi_','').replace('_kpi',''):26s} all={a_all:+.2f} "
                               f"high-ROA={a_hi:+.2f} lift={a_hi - a_all:+.2f}{flag}")
        STRUCT_NOTE.append("   a large lift means high performers are never found below that level ⇒ necessary")
        STRUCT_NOTE.append("   condition (prerequisite), which is a different claim from average contribution.")
    # ---- transfer the same battery to the STRUCTURAL governance index ----
    st = [c for c in ["techcom_strict", "exec_cio", "exec_cto", "exec_cdo", "exec_arch"] if c in D.columns]
    if len(st) >= 3:
        G2 = D.groupby("cik")[st + ["roa"] + ctl].mean().dropna()
        if "industry_raw" in D.columns: G2 = G2.join(D.groupby("cik")["industry_raw"].first())
        if len(G2) > 60:
            G2["mean_g"] = G2[st].mean(axis=1); G2["min_g"] = G2[st].min(axis=1)
            ra = sm.OLS(G2["roa"].values, design(G2, ["mean_g"])).fit(cov_type="HC1")
            rb2 = sm.OLS(G2["roa"].values, design(G2, ["min_g"])).fit(cov_type="HC1")
            STRUCT_NOTE.append(f"TRANSFER to structural governance · mean β={ra.params['mean_g']:+.4f} "
                               f"(p={ra.pvalues['mean_g']:.3f}) · min β={rb2.params['min_g']:+.4f} "
                               f"(p={rb2.pvalues['min_g']:.3f})")
            # FIT-TO-POTENTIAL for governance: the claim is NOT "more governance is better".
            # It is that governance should be PROPORTIONATE TO ARCHITECTURAL COMPLEXITY.
            # Over-governing a simple firm is bureaucracy; under-governing a complex one is chaos.
            cx = [c for c in ["size", "lev", "rnd", "aux_n_tokens"] if c in D.columns]
            if len(cx) >= 2:
                CX = D.groupby("cik")[cx].mean()
                G3 = G2.join(CX, how="inner", rsuffix="_cx").dropna()
                if len(G3) > 60:
                    xcols = [c for c in G3.columns if c in cx or c.endswith("_cx")]
                    Xc = sm.add_constant(G3[xcols].astype(float))
                    fg = sm.OLS(G3["mean_g"].values, Xc).fit()
                    G3["gap_g"] = G3["mean_g"] - fg.predict(Xc)
                    G3["absgap_g"] = G3["gap_g"].abs()
                    rg1 = sm.OLS(G3["roa"].values, design(G3, ["absgap_g"])).fit(cov_type="HC1")
                    STRUCT_NOTE.append(f"   FIT-TO-POTENTIAL on governance (complexity proxies {xcols}, "
                                       f"R²={fg.rsquared:.3f}): β_|gap|={rg1.params['absgap_g']:+.4f} "
                                       f"(p={rg1.pvalues['absgap_g']:.3f})")
                    STRUCT_NOTE.append("   NEGATIVE ⇒ governance mismatched to complexity costs in BOTH directions:")
                    STRUCT_NOTE.append("   bureaucracy in simple firms, chaos in complex ones. That is a sharper claim")
                    STRUCT_NOTE.append("   than 'more governance is better' and it is what the tool should encode.")
            STRUCT_NOTE.append("   the EAGI TEXT index cannot carry this framing: a gap between two noise draws")
            STRUCT_NOTE.append("   is still noise (α=-0.06). Fit-to-potential transfers to STRUCTURE, not to text.")
            STRUCT_NOTE.append("   the text index cannot support dimension analysis (terms do not co-move);")
            STRUCT_NOTE.append("   the structural index can, and that is where the EAGI transfer lands.")
    globals()["STRUCT_NOTE"] = STRUCT_NOTE
    print("\n".join(STRUCT_NOTE))
````

````python
# ============ 6i · ITEM 106 STAGGERED-DEADLINE DESIGN (feasibility-gated) ============
# THE RULE: Item 106 governance disclosure is required for FISCAL YEARS ENDING ON OR AFTER
# 15 December 2023. The smaller-reporting-company +180-day extension applies to Item 1.05
# INCIDENT reporting, not to the annual governance description — so the SRC stagger is the
# wrong lever for this panel, and probably absent from it anyway.
#
# THE USABLE STAGGER IS FISCAL YEAR END. A December-FYE firm first complies in FY2023; a
# June-FYE firm not until its FY ending mid-2024 — up to ~11 months apart, driven by a fiscal
# calendar fixed decades ago and plausibly exogenous to current performance.
#
# This cell REFUSES TO RUN a design that has no variation. It reports feasibility first and
# states exactly what would be needed, rather than silently estimating on zero treated units.
I106_NOTE = []
if G("df") is None or "fye" not in df.columns:
    print("SKIPPED — filing metadata (cell 5f) has not run.")
else:
    import statsmodels.api as sm
    D = df.copy()
    D["fye"] = D["fye"].astype(str).str.zfill(4)
    D["fye_month"] = pd.to_numeric(D["fye"].str[:2], errors="coerce")
    D["fye_day"] = pd.to_numeric(D["fye"].str[2:], errors="coerce")
    # first fiscal year (labelled by our panel's convention) ending on/after 15 Dec 2023
    dec_cut = (D["fye_month"] == 12) & (D["fye_day"] >= 15)
    D["first_req_fy"] = np.where(dec_cut, 2023, 2024)
    D["post106"] = (D["year"] >= D["first_req_fy"]).astype(float)
    firms = D.drop_duplicates("cik")
    I106_NOTE.append("FEASIBILITY OF THE STAGGERED DESIGN:")
    I106_NOTE.append(f"  filer categories in panel: {dict(firms['filer_category'].value_counts().head(4))}")
    src_share = float(firms["filer_category"].astype(str).str.contains("Non-accelerated|Smaller",
                                                                       case=False, na=False).mean())
    I106_NOTE.append(f"  smaller / non-accelerated share: {src_share:.1%} → "
                     + ("SRC stagger unusable (needs ≥10%)" if src_share < 0.10 else "SRC stagger possible"))
    fye_mix = firms["first_req_fy"].value_counts().to_dict()
    early = int(fye_mix.get(2023, 0)); late = int(fye_mix.get(2024, 0))
    I106_NOTE.append(f"  FYE stagger: {early} firms first required in FY2023 · {late} in FY2024")
    I106_NOTE.append(f"  distinct fiscal year ends: {firms['fye'].nunique()} "
                     f"(month distribution: {dict(firms['fye_month'].value_counts().head(4))})")
    post_years = int((D["year"] >= 2023).sum() > 0) and int(D.loc[D.year >= 2023, "year"].nunique())
    I106_NOTE.append(f"  post-period years available in the panel: {post_years}")
    viable = (min(early, late) >= 20) and post_years >= 2
    if not viable:
        I106_NOTE.append("  ⇒ DESIGN NOT VIABLE HERE. Needed: ≥20 firms on each side of the FYE cut AND")
        I106_NOTE.append("    ≥2 post-period years. Report as an identification ROADMAP, not a result —")
        I106_NOTE.append("    extending the panel through FY2026 would make it estimable.")
    else:
        I106_NOTE.append("  ⇒ DESIGN VIABLE. Estimating staggered DiD on the FYE cut.")
        ctl = [c for c in ["size", "lev", "rnd"] if c in D.columns]
        # validation first: does the mandated disclosure actually appear when required?
        if "item106_present" in D.columns:
            chk = D.groupby("post106")["item106_present"].mean()
            I106_NOTE.append(f"  timing-map validation — Item 106 present: "
                             f"pre={chk.get(0.0, float('nan')):.1%} vs post={chk.get(1.0, float('nan')):.1%}")
            I106_NOTE.append("    a sharp jump confirms the map; a flat line means the map is wrong.")
        for dep in [c for c in ["roa", "q"] if c in D.columns]:
            S = D.dropna(subset=[dep] + ctl)
            if len(S) < 100: continue
            X = S[["post106"] + ctl].astype(float)
            X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y", drop_first=True, dtype=float),
                           pd.get_dummies(S["cik"].astype(str), prefix="f", drop_first=True, dtype=float)], axis=1)
            r_ = sm.OLS(S[dep].values, sm.add_constant(X)).fit(
                cov_type="cluster", cov_kwds={"groups": S["cik"].values})
            I106_NOTE.append(f"  staggered DiD ({dep.upper()}, firm+year FE): β_post106={r_.params['post106']:+.4f} "
                             f"p={r_.pvalues['post106']:.3f} n={len(S)}")
        I106_NOTE.append("  CAUTION: with staggered adoption and firm+year FE, already-treated units act as")
        I106_NOTE.append("  controls (Goodman-Bacon). With only two cohorts and a short post-period the bias")
        I106_NOTE.append("  is limited, but state it — and prefer a Callaway–Sant'Anna estimator if extended.")
    # ---------- FILER STATUS: sample composition and external validity, NOT a control ----------
    # Filer status is near-collinear with log assets, so adding it alongside `size` buys nothing.
    # Its real use is describing WHO the panel represents, and testing whether a finding holds
    # outside the very largest firms. If the panel is homogeneous, that is itself the external-
    # validity paragraph — state the limit rather than implying breadth the sample does not have.
    comp = firms["filer_category"].astype(str).replace("", "unreported").value_counts()
    I106_NOTE.append("SAMPLE COMPOSITION by filer status (for the external-validity paragraph):")
    for k_, v_ in comp.items():
        I106_NOTE.append(f"    {str(k_)[:38]:38s} {int(v_):4d} firms  ({v_/max(1,len(firms)):.0%})")
    biggest = comp.index[0] if len(comp) else ""
    other_n = int(len(firms) - comp.iloc[0]) if len(comp) else 0
    if other_n >= 20 and "roa" in D.columns:
        D["not_largest"] = (D["filer_category"].astype(str) != str(biggest)).astype(float)
        base = [c for c in ["size", "lev", "rnd"] if c in D.columns]
        xv = "ci" if "ci" in D.columns else None
        if xv:
            S = D.dropna(subset=[xv, "roa"] + base).copy()
            S["xb"] = S.groupby("cik")[xv].transform("mean")
            S["ixn"] = (S["xb"] - S["xb"].mean()) * (S["not_largest"] - S["not_largest"].mean())
            X = S[["xb", "not_largest", "ixn"] + base].astype(float)
            X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
            r_ = sm.OLS(S["roa"].values, sm.add_constant(X)).fit(
                cov_type="cluster", cov_kwds={"groups": S["cik"].values})
            I106_NOTE.append(f"  heterogeneity by filer status: β_ixn={r_.params['ixn']:+.4f} "
                             f"p={r_.pvalues['ixn']:.3f} (n={len(S)}, {other_n} smaller firms)")
            I106_NOTE.append("    null ⇒ the finding is not confined to the largest filers.")
    else:
        I106_NOTE.append(f"  only {other_n} firms outside '{str(biggest)[:34]}' — no usable heterogeneity split.")
        I106_NOTE.append("  ⇒ EXTERNAL VALIDITY LIMIT: results describe large accelerated filers. State this")
        I106_NOTE.append("    explicitly; do not generalise to mid-market or non-US firms, which is precisely")
        I106_NOTE.append("    the gap the qualitative strand is designed to address.")
    globals()["I106_NOTE"] = I106_NOTE
    print("\n".join(I106_NOTE))
````

````python
# ============ 7 · OSTER (2019) BOUNDS ============
if G("df") is None or G("d") is None:
    print("SKIPPED — estimation incomplete.")
else:
    import statsmodels.api as sm

    def oster(dep, treat, controls, data, rmax_mult=1.3):
        dd = data.reset_index().copy()
        # Alternating projections: a single firm-then-year pass is NOT the two-way within
        # transformation on an unbalanced panel. Iterate to convergence (Frisch–Waugh).
        cols_ = [dep, treat] + controls
        for _it in range(50):
            prev = dd[cols_].to_numpy(copy=True)
            for gcol in ("cik", "year"):
                for c in cols_:
                    dd[c] = dd[c] - dd.groupby(gcol)[c].transform("mean")
            if np.nanmax(np.abs(dd[cols_].to_numpy() - prev)) < 1e-10:
                break
        short = sm.OLS(dd[dep], sm.add_constant(dd[[treat]])).fit()
        long_ = sm.OLS(dd[dep], sm.add_constant(dd[[treat] + controls])).fit()
        b0, r0 = short.params[treat], short.rsquared
        b1, r1 = long_.params[treat], long_.rsquared
        rmax = min(1.0, max(rmax_mult * r1, r1 + 1e-6))   # Rmax must exceed R̃
        denom = (r1 - r0) or 1e-9
        beta_star = b1 - (b0 - b1) * (rmax - r1) / denom
        delta_for_zero = b1 * denom / (((b0 - b1) * (rmax - r1)) or 1e-9)
        return dict(beta_ctrl=b1, beta_star_d1=beta_star, delta_beta0=delta_for_zero, r_ctrl=r1, rmax=rmax)

    oster_res = {}
    for y in [k for k in ["roa", "q", "sga"] if k in (G("results") or {})]:
        o = oster(y, "eagi_l", CTRL, d.dropna(subset=[y]))
        oster_res[y] = o
        print(f"{y.upper():4s}  β̃={o['beta_ctrl']:+.4f}  β*(δ=1)={o['beta_star_d1']:+.4f}  δ(β=0)={o['delta_beta0']:+.2f}  Rmax={o['rmax']:.2f}")
    print("\nRead: |δ(β=0)| > 1 ⇒ unobservables would need to matter more than observables to null the estimate.")
    print("CAUTION: Oster bounds are only meaningful for a controlled estimate that is itself")
    print("distinguishable from zero. Where β̃ ≈ 0, δ is arithmetic noise — do not report it.")
    for _y, _o in oster_res.items():
        _p = results[_y][0].pvalues["eagi_l"]
        if _p > 0.10:
            print(f"  {_y.upper()}: β̃ not distinguishable from zero (p={_p:.2f}) → δ suppressed from interpretation")
````

````python
# ============ 8 · EVENT STUDY — TechCom formation (gated C&S) ============
if G("df") is None:
    print("SKIPPED — no merged panel.")
else:
    es = df.sort_values(["cik", "year"]).copy()
    first = es[es.techcom == 1].groupby("cik")["year"].min().rename("g")
    es = es.merge(first, on="cik", how="left")
    clean = es.groupby("cik").apply(lambda x: x["g"].notna().any() and x["g"].iloc[0] > YEAR_MIN).rename("clean").reset_index()
    switchers = int(clean["clean"].sum())
    print(f"Clean switchers (first adoption after {YEAR_MIN}): {switchers}")

    es_note = ""

    def event_study_fe():
        """Transparent event-study: relative-time dummies (τ=−3..+3, τ=−1 omitted),
        clean switchers vs never-treated, firm+year FE, firm-clustered SE.
        Replaces the external C&S package so the gate result is always estimable."""
        e = df.copy()
        g_ = df[df.techcom == 1].groupby("cik")["year"].min()
        e["g"] = e["cik"].map(g_)
        e = e[e["g"].isna() | (e["g"] > YEAR_MIN)].copy()      # drop always-treated
        e["tau"] = e["year"] - e["g"]
        dums = []
        for t in range(-3, 4):
            if t == -1: continue                                # omitted reference period
            nm = f"tau_{'m' if t < 0 else 'p'}{abs(t)}"
            e[nm] = ((e["tau"] == t) & e["g"].notna()).astype(float)
            dums.append(nm)
        ctrls = [c for c in (G("CTRL") or []) if c in e.columns]
        e = e.dropna(subset=["roa"] + ctrls).set_index(["cik", "year"])
        if len(e) < 50 or e[dums].to_numpy().sum() < 20:
            return None
        r = fe_ols("roa", dums + ctrls, e, se="clustered")
        out = ["Event-study FE (ROA), reference τ=−1, firm-clustered SE:"]
        for nm in dums:
            out.append(f"    {nm:8s} β={r.params[nm]:+.4f}  se={r.std_errors[nm]:.4f}  p={r.pvalues[nm]:.3f}")
        pre = [nm for nm in dums if nm.startswith("tau_m")]
        worst = max((r.pvalues[nm] for nm in pre), default=1.0)
        out.append(f"    pre-trend check: smallest pre-period p-value = {min((r.pvalues[nm] for nm in pre), default=float('nan')):.3f}"
                   " (want all pre-period coefficients indistinguishable from zero)")
        return "\n".join(out)

    if switchers >= 20:
        # No external package here. Cell 8c implements Callaway-Santanna directly; this cell is
        # the deterministic FE event study, which carries no dependency and always runs.
        _fe = None
        try:
            _fe = event_study_fe()
        except Exception as e2:
            print("event-study FE failed:", type(e2).__name__, e2)
        es["tau"] = es["year"] - es["g"]
        if _fe:
            es_note = _fe
            print(es_note)
        else:
            es_note = ("FE event study not estimable -> descriptive event-time ROA means (no inference):\n"
                       + es[es.g.notna() & es.tau.between(-3, 3)].groupby("tau")["roa"].mean().round(4).to_string())
            print(es_note)
    else:
        es["tau"] = es["year"] - es["g"]
        es_note = (f"GATE HOLDS: {switchers} switchers < 20 → descriptive only, excluded from inference (pre-registered).\n"
                   + es[es.g.notna() & es.tau.between(-3, 3)].groupby("tau")["roa"].mean().round(4).to_string())
        print(es_note)
````

````python
# ============ 8b · TECHCOM MEASUREMENT SENSITIVITY (additive) ============
# EXTREME-CODING BOUNDS (Manski-style). The ambiguous 'review' cases are not guessed: they are
# coded both ways, giving a lower bound (review→reject) and an upper bound (review→confirm).
# If the conclusion is identical at both extremes, the ambiguity is irrelevant to inference and
# no human or LLM coding is needed. If it differs, the bounds quantify exactly what is at stake.
TC_SENS = None
if G("df") is None or G("CTRL") is None:
    print("SKIPPED — estimation incomplete.")
else:
    import statsmodels.api as sm
    defs = [c for c in ["techcom", "techcom_strict", "techcom_incl"] if c in df.columns]
    if len(defs) >= 2:
        S = df.sort_values(["cik", "year"]).copy()
        lgs = S[["cik", "year"] + defs].copy(); lgs["year"] += 1
        S = S.merge(lgs.rename(columns={c: c + "_L" for c in defs}), on=["cik", "year"], how="left")
        S = S.merge(df[["cik", "year"]].drop_duplicates(), on=["cik", "year"], how="left")
        lgc = df.sort_values(["cik", "year"])[["cik", "year"] + [c for c in ["size", "lev", "rnd"] if c in df.columns]].copy()
        lgc["year"] += 1
        S = S.merge(lgc.rename(columns={c: c + "_l" for c in lgc.columns if c not in ("cik", "year")}),
                    on=["cik", "year"], how="left", suffixes=("", "_dup"))
        ctl = [c for c in ["size_l", "lev_l", "rnd_l"] if c in S.columns]
        rows = []
        for dcol in defs:
            sub = S.dropna(subset=[dcol + "_L", "roa"] + ctl)
            if len(sub) < 60: continue
            first = sub[sub[dcol] == 1].groupby("cik")["year"].min()
            sw = int(((first > YEAR_MIN).sum()) if len(first) else 0)
            X = pd.get_dummies(sub[["year"]].astype(str), drop_first=True, dtype=float)
            X[dcol + "_L"] = sub[dcol + "_L"].values
            for c in ctl: X[c] = sub[c].values
            r = sm.OLS(sub["roa"].values, sm.add_constant(X)).fit(
                cov_type="cluster", cov_kwds={"groups": sub["cik"].values})
            lbl = {"techcom": "raw regex (no triage)",
                   "techcom_strict": "LOWER bound: review→reject",
                   "techcom_incl": "UPPER bound: review→confirm"}.get(dcol, dcol)
            rows.append(dict(definition=dcol, scenario=lbl, share=round(float(sub[dcol].mean()), 3),
                             switchers=sw, n=len(sub),
                             beta_roa=round(float(r.params[dcol + "_L"]), 4),
                             se=round(float(r.bse[dcol + "_L"]), 4),
                             p=round(float(r.pvalues[dcol + "_L"]), 3)))
        if rows:
            TC_SENS = pd.DataFrame(rows)
            print("TechCom measurement sensitivity (year FE, firm-clustered SE):")
            print(TC_SENS.to_string(index=False))
            bnd = TC_SENS[TC_SENS.definition.isin(["techcom_strict", "techcom_incl"])]
            if len(bnd) == 2:
                lo, hi = float(bnd.beta_roa.min()), float(bnd.beta_roa.max())
                p_lo, p_hi = float(bnd.p.min()), float(bnd.p.max())
                same_sign = (lo > 0) == (hi > 0)
                both_ns, both_sig = (p_lo > .05 and p_hi > .05), (p_lo < .05 and p_hi < .05)
                print(f"\nEXTREME-CODING BOUNDS on β(TechCom→ROA): [{lo:+.4f}, {hi:+.4f}]")
                print(f"  ambiguous share = {(G('TC_REVIEW_SHARE') or 0):.1%} of firm-years")
                if both_ns:
                    print("  ⇒ NULL AT BOTH EXTREMES — the ambiguous cases cannot change the conclusion;")
                    print("    no human or LLM coding is required for this result to stand.")
                elif both_sig and same_sign:
                    print("  ⇒ SIGNIFICANT AND SAME-SIGNED AT BOTH EXTREMES — conclusion is robust to the coding.")
                else:
                    print("  ⇒ CONCLUSION FLIPS INSIDE THE BOUNDS — the ambiguous pile is decisive.")
                    print("    This is exactly where the LLM cross-check and the qualitative study earn their keep.")
                globals()["TC_BOUNDS"] = (lo, hi, both_ns, both_sig and same_sign)
            try: TC_SENS.to_csv(f"{OUT_DIR}/techcom_sensitivity.csv", index=False)
            except Exception: pass
````

````python
# ============ 8c · CALLAWAY-SANT'ANNA, HAND-ROLLED (no package dependency) ============
# WHERE CS IS THE RIGHT TOOL: TechCom formation - many adoption cohorts across several years with
# plausibly heterogeneous effects. Under two-way FE, already-treated units act as controls for
# later adopters (Goodman-Bacon), which can flip the sign of an aggregate estimate.
# WHERE CS IS THE WRONG TOOL: the Item 106 design has TWO cohorts and a one-to-two year post
# period. There a transparent 2x2 DiD is more interpretable than heterogeneity-robust machinery.
#
# Implemented directly rather than via the `differences` package, which failed at runtime earlier:
#   ATT(g,t) = [Y_g,t - Y_g,g-1] - [Y_C,t - Y_C,g-1],  C = NOT-YET-TREATED at t
# aggregated by event time, with a firm-level block bootstrap. Deterministic given the seed.
import math
CS_NOTE, CS_TBL = [], None
if G("df") is None or "techcom" not in df.columns:
    print("SKIPPED - no panel or no treatment variable.")
else:
    tcol = "techcom_strict" if "techcom_strict" in df.columns and df["techcom_strict"].std() > 0 else "techcom"
    D = df[["cik", "year", tcol, "roa"]].dropna().sort_values(["cik", "year"]).copy()
    first = D[D[tcol] == 1].groupby("cik")["year"].min().rename("g")
    D = D.merge(first, on="cik", how="left")
    always = set(D.loc[D["g"] <= D["year"].min(), "cik"])
    D = D[~D["cik"].isin(always)]                      # always-treated have no clean pre-period
    cohorts = sorted(int(x) for x in D["g"].dropna().unique())
    n_sw = int(D.dropna(subset=["g"])["cik"].nunique())
    CS_NOTE.append(f"CALLAWAY-SANT'ANNA on {tcol}: {n_sw} clean switchers across "
                   f"{len(cohorts)} cohorts {[int(c) for c in cohorts]}")
    if n_sw < 20 or len(cohorts) < 2:
        CS_NOTE.append("  => GATE HOLDS: fewer than 20 switchers or fewer than 2 cohorts. Reported as")
        CS_NOTE.append("     descriptive only, exactly as pre-registered. CS adds nothing here.")
    else:
        piv = D.pivot_table(index="cik", columns="year", values="roa", aggfunc="mean")
        piv.columns = [int(c) for c in piv.columns]     # cohort years become floats after the merge;
        piv = piv.loc[:, ~piv.columns.duplicated()]     # int-cast avoids label collisions on .loc
        gmap = D.dropna(subset=["g"]).drop_duplicates("cik").set_index("cik")["g"]
        years = sorted(int(x) for x in D["year"].unique())
        # MULTIPLIER BOOTSTRAP (as in Callaway-Sant'Anna). Resampling firms into a SET would
        # discard multiplicity and understate the SE several-fold; exponential firm weights with
        # mean 1 preserve it and keep the estimator deterministic given the seed.
        def att_gt(g, t, w=None):
            g, t = int(g), int(t)
            base = g - 1
            if base == t: return np.nan, 0
            if base not in piv.columns or t not in piv.columns: return np.nan, 0
            treat = [c for c in gmap[gmap == g].index if c in piv.index]
            # Controls must be untreated at BOTH the comparison period t AND the base period g-1.
            # Requiring only g' > t lets a cohort adopting at g-1 sit in the control group while its
            # BASE outcome already contains the treatment effect, which biases pre-period ATTs
            # upward. Threshold on max(t, base) removes that contamination.
            tset = set(treat)
            thresh = max(t, base)
            ctrl = [c for c in piv.index if c not in tset and
                    ((c not in gmap.index) or (int(gmap.get(c, 9999)) > thresh))]
            tr = piv.loc[treat, [base, t]].dropna() if treat else piv.head(0)
            cr = piv.loc[ctrl, [base, t]].dropna() if ctrl else piv.head(0)
            if len(tr) < 3 or len(cr) < 5: return np.nan, 0
            dt = (tr[t] - tr[base]); dc = (cr[t] - cr[base])
            if w is None:
                return float(dt.mean() - dc.mean()), len(tr)
            wt = w.reindex(dt.index).fillna(1.0); wc = w.reindex(dc.index).fillna(1.0)
            if wt.sum() <= 0 or wc.sum() <= 0: return np.nan, 0
            return float(np.average(dt, weights=wt) - np.average(dc, weights=wc)), len(tr)
        def event_agg(w=None):
            acc = {}
            for g in cohorts:
                for t in years:
                    e = int(t - g)
                    if e < -3 or e > 3: continue
                    a, n_ = att_gt(g, t, w)
                    if a == a and n_ > 0:
                        acc.setdefault(e, []).append((a, n_))
            return {e: float(np.average([x[0] for x in v], weights=[x[1] for x in v]))
                    for e, v in acc.items() if v}
        pt = event_agg()
        rs = np.random.default_rng(G("SEED") or 42)
        firms_all = list(piv.index)
        boots = []
        for _ in range(300):
            w = pd.Series(rs.exponential(1.0, len(firms_all)), index=firms_all)
            boots.append(event_agg(w))
        rows = []
        for e in sorted(pt):
            draws = [b[e] for b in boots if e in b]
            se = float(np.std(draws, ddof=1)) if len(draws) > 20 else np.nan
            z = pt[e] / se if (se == se and se > 0) else np.nan
            pv = float(2 * (1 - 0.5 * (1 + math.erf(abs(z) / math.sqrt(2))))) if z == z else np.nan
            rows.append(dict(event_time=e, att=round(pt[e], 4),
                             se=(round(se, 4) if se == se else None),
                             p=(round(pv, 3) if pv == pv else None)))
        CS_TBL = pd.DataFrame(rows)
        CS_NOTE.append("  ATT by event time (tau=-1 is the reference period, 300 firm-block bootstrap):")
        for ln in CS_TBL.to_string(index=False).splitlines(): CS_NOTE.append("    " + ln)
        pre = CS_TBL[CS_TBL.event_time < -1]
        if len(pre) and pre["p"].notna().any():
            CS_NOTE.append(f"  pre-trend check: smallest pre-period p = {pre['p'].dropna().min():.3f} "
                           "(want all pre-period ATTs indistinguishable from zero)")
        post = CS_TBL[CS_TBL.event_time >= 0]
        if len(post):
            CS_NOTE.append(f"  simple post-treatment average ATT = {post['att'].mean():+.4f}")
        CS_NOTE.append("  Controls are NOT-YET-TREATED units only, so already-treated firms never serve")
        CS_NOTE.append("  as controls - the bias two-way FE cannot avoid under staggered adoption.")
        try: CS_TBL.to_csv(f"{OUT_DIR}/callaway_santanna_att.csv", index=False)
        except Exception: pass

        # ---- CROSS-CHECK against the `differences` package (automatic, cached) ----
        # Runs by default: an off-by-default check validates nothing. Three protections make that
        # safe. (1) It is a diagnostic about OUR CODE, never a research result, and can never
        # substitute for the estimate above. (2) The comparison is CACHED against a hash of our own
        # ATTs, so it installs once and later runs read the stored number - reproducible even if the
        # package version drifts. (3) Every failure path is caught; the run continues regardless.
        if G("USE_EXTERNAL_CS"):
            import hashlib
            att_key = hashlib.md5(CS_TBL.round(6).to_csv(index=False).encode()).hexdigest()[:16]
            xc_path = f"{CACHE_DIR}/cs_crosscheck.json" if G("CACHE_DIR") else None
            cached = None
            if xc_path and os.path.exists(xc_path):
                try:
                    _c = json.load(open(xc_path))
                    if _c.get("att_key") == att_key: cached = _c
                except Exception:
                    cached = None
            if cached:
                CS_NOTE.append(f"  CROSS-CHECK vs `differences` (cached): max |difference| = "
                               f"{cached.get('max_diff'):.4f} across {cached.get('n_periods')} periods")
                CS_NOTE.append("    cached against a hash of our own ATTs - recomputed only if they change.")
            else:
                try:
                    if G("pipq"): pipq("differences")
                    from differences import ATTgt
                    E = D.dropna(subset=["roa"]).copy()
                    E["g"] = E["g"].fillna(0).astype(int)
                    att = ATTgt(data=E.set_index(["cik", "year"]), cohort_name="g")
                    att.fit(formula="roa")
                    ext = np.asarray(att.aggregate("event")).ravel()
                    ours = CS_TBL.set_index("event_time")["att"]
                    diffs = [abs(float(ext[j_]) - float(v_)) for j_, v_ in enumerate(ours.values)
                             if j_ < len(ext)]
                    if diffs:
                        md_ = float(max(diffs))
                        CS_NOTE.append(f"  CROSS-CHECK vs `differences`: max |difference| in event-time ATT = "
                                       f"{md_:.4f} across {len(diffs)} periods")
                        CS_NOTE.append("    agreement validates the hand-rolled implementation; a large gap means")
                        CS_NOTE.append("    the two make different control-group choices - investigate rather")
                        CS_NOTE.append("    than silently preferring either.")
                        if xc_path:
                            try:
                                json.dump({"att_key": att_key, "max_diff": md_, "n_periods": len(diffs)},
                                          open(xc_path, "w"))
                            except Exception:
                                pass
                    else:
                        CS_NOTE.append("  CROSS-CHECK: package ran but its output could not be aligned to ours.")
                except Exception as e:
                    CS_NOTE.append(f"  CROSS-CHECK unavailable ({type(e).__name__}) - our estimator is the")
                    CS_NOTE.append("  primary result and is unaffected. This is a note, not a failure.")
        else:
            CS_NOTE.append("  cross-check disabled (USE_EXTERNAL_CS=False) - offline mode.")
    globals()["CS_NOTE"] = CS_NOTE
    print("\n".join(CS_NOTE))
````

````python
# ============ 9 · ROBUSTNESS ============
if G("df") is None or G("d") is None or G("fe_dk") is None:
    print("SKIPPED — estimation incomplete.")
else:
    loto = []
    for drop_t in LEXICON:
        keep_t = [t for t in LEXICON if t != drop_t]
        e2 = Z[keep_t].mean(axis=1)
        tmp = panel.merge(Z[["cik", "year"]].assign(eagi2=e2), on=["cik", "year"])
        tmp = tmp.merge(df[["cik", "year"]].drop_duplicates(), on=["cik", "year"], how="inner")
        cols2 = ["eagi2", "ci"] + [c for c in ["size", "lev", "rnd", "rnd_missing", "capex", "capsoft"] if c in tmp.columns]
        tmp = add_lags(tmp.sort_values(["cik", "year"]), cols2, 1)
        ctrl2 = [c + "_l" for c in cols2 if c != "eagi2" and (c + "_l") in CTRL]
        tmp = tmp.dropna(subset=["eagi2_l", "ci_l"] + ctrl2).set_index(["cik", "year"])   # same conditioning as baseline
        if len(tmp) < 10:
            continue
        b = fe_dk("roa", ["eagi2_l"] + ctrl2, tmp).params["eagi2_l"]
        loto.append((drop_t, b))
    loto = pd.DataFrame(loto, columns=["dropped_term", "beta_roa"]).sort_values("beta_roa")
    if len(loto):
        print("Leave-one-term-out — β(ROA) range: [{:+.4f}, {:+.4f}]".format(loto.beta_roa.min(), loto.beta_roa.max()))

    dw = d.copy()
    for y in ["roa", "q", "sga"]:
        lo, hi = dw[y].quantile([0.01, 0.99])
        dw[y] = dw[y].clip(lo, hi)
    beta_winsor = fe_dk("roa", ["eagi_l"] + CTRL, dw).params["eagi_l"]
    print("Winsorized β(ROA) = {:+.4f}".format(beta_winsor))

    base2 = [c for c in ["eagi", "size", "lev", "rnd", "capex", "capsoft"] if c in df.columns]
    d2 = add_lags(df.sort_values(["cik", "year"]), base2, 2)
    ctrl3 = [c for c in CTRL if c in d2.columns]
    d2 = d2.dropna(subset=["eagi_l"] + ctrl3).set_index(["cik", "year"])
    beta_lag2 = fe_dk("roa", ["eagi_l"] + ctrl3, d2).params["eagi_l"] if len(d2) >= 10 else float("nan")
    print("t−2 lag β(ROA) = {:+.4f}  (n={}) — '_l' columns here are t−2, not t−1".format(beta_lag2, len(d2)))

    beta_placebo = None
    if "placebo_l" in d.columns:
        dp = d.dropna(subset=["placebo_l"])
        beta_placebo = fe_dk("roa", ["placebo_l"] + CTRL, dp).params["placebo_l"]
        print("Placebo β(ROA) = {:+.4f}  (expect ≈ 0)".format(beta_placebo))
````

````python
# ============ 9b · FALSIFICATION & SPECIFICATION CURVE ============
# Two things a tier-1 referee will demand of results this fragile:
#   (a) a PLACEBO run of every firm-level test — if meaningless vocabulary "predicts" the same
#       outcomes, the finding is a document/firm property, not the construct;
#   (b) a SPECIFICATION CURVE — when significance flips across reasonable specs, the honest
#       object is the distribution of estimates, not a chosen point estimate.
FALSI, SPECCURVE = None, None
if G("df") is None:
    print("SKIPPED — no panel.")
else:
    import statsmodels.api as sm
    D = df.copy()
    idxc = "eagi_best" if "eagi_best" in D.columns and D["eagi_best"].notna().sum() > 100 else "eagi"

    # ---------- (a) PLACEBO falsification of the firm-level results ----------
    if "placebo" in D.columns:
        ad = D.dropna(subset=["roa"]).sort_values(["cik", "year"])
        gg = ad.groupby("cik")
        F = pd.DataFrame({"vol": gg["roa"].std(),
                          "worst": gg["roa"].apply(lambda x: float(x.diff().min()) if len(x) > 2 else np.nan),
                          "real": gg[idxc].mean(), "plac": gg["placebo"].mean()}).reset_index()
        for c in ["size", "lev", "rnd"]:
            if c in ad.columns: F[c] = gg[c].mean().values
        if "industry_raw" in ad.columns: F["ind"] = gg["industry_raw"].first().values
        F = F.dropna(subset=["vol", "worst", "real", "plac"])
        if len(F) > 40:
            base = [c for c in ["size", "lev", "rnd"] if c in F.columns]
            rows = []
            for dep in ["vol", "worst"]:
                for xv, lab in [("real", f"{idxc} (construct)"), ("plac", "PLACEBO lexicon")]:
                    X = F[[xv] + base].astype(float)
                    if "ind" in F.columns and F["ind"].nunique() > 1:
                        X = pd.concat([X, pd.get_dummies(F["ind"].astype(str), prefix="i",
                                                         drop_first=True, dtype=float)], axis=1)
                    r = sm.OLS(F[dep].values, sm.add_constant(X)).fit(cov_type="HC1")
                    rows.append(dict(outcome=dep, regressor=lab, beta=round(float(r.params[xv]), 4),
                                     p=round(float(r.pvalues[xv]), 3), n=len(F)))
            # AI index gets the SAME falsification: it is prevalent and may proxy sector/verbosity
            if "ai_idx" in D.columns:
                Fa = F.merge(D.groupby("cik")["ai_idx"].mean().rename("ai").reset_index(), on="cik", how="left").dropna(subset=["ai"])
                if len(Fa) > 40:
                    Xa = Fa[["ai"] + base].astype(float)
                    if "ind" in Fa.columns and Fa["ind"].nunique() > 1:
                        Xa = pd.concat([Xa, pd.get_dummies(Fa["ind"].astype(str), prefix="i",
                                                           drop_first=True, dtype=float)], axis=1)
                    ra = sm.OLS(Fa[dep].values, sm.add_constant(Xa)).fit(cov_type="HC1")
                    rows.append(dict(outcome=dep, regressor="AI index (different construct)",
                                     beta=round(float(ra.params["ai"]), 4),
                                     p=round(float(ra.pvalues["ai"]), 3), n=len(Fa)))
            FALSI = pd.DataFrame(rows)
            print("PLACEBO FALSIFICATION of the firm-level results:")
            print(FALSI.to_string(index=False))
            bad = []
            for dep in ["vol", "worst"]:
                pr = FALSI[(FALSI.outcome == dep) & (FALSI.regressor == "PLACEBO lexicon")]
                rr = FALSI[(FALSI.outcome == dep) & (FALSI.regressor.str.contains("construct"))]
                if len(pr) and len(rr) and float(pr.p.iloc[0]) < 0.10:
                    bad.append(dep)
            if bad:
                print(f"\n  ⚠ PLACEBO ALSO SIGNIFICANT for {bad} → the firm-level result is a")
                print("    document/firm property (size, complexity, verbosity), NOT EA governance.")
                print("    Do not report the adaptability finding as a construct effect.")
            else:
                print("\n  ✓ placebo null while the construct is not → the firm-level result survives falsification.")
            globals()["FALSI"] = FALSI
            try: FALSI.to_csv(f"{OUT_DIR}/placebo_falsification.csv", index=False)
            except Exception: pass
        # is the firm-mean index just size / verbosity?
        if len(F) > 40 and "size" in F.columns:
            r_sz = float(F["real"].corr(F["size"])); r_pl = float(F["real"].corr(F["plac"]))
            print(f"\n  corr(firm-mean {idxc}, firm-mean size) = {r_sz:+.3f}"
                  f" · corr(construct, placebo) = {r_pl:+.3f}")
            # These correlations discriminate where the placebo REGRESSION cannot: if size is a
            # control, it absorbs the shared variance and the placebo can look null even when both
            # measures are the same document property.
            flags = []
            if abs(r_sz) > 0.5: flags.append(f"index is largely FIRM SIZE (|r|={abs(r_sz):.2f})")
            if abs(r_pl) > 0.5: flags.append(f"construct and PLACEBO move together (|r|={abs(r_pl):.2f}) "
                                             "⇒ both are measuring disclosure volume, not content")
            if flags:
                print("  ⚠ CONSTRUCT-VALIDITY FAILURE: " + "; ".join(flags))
                print("    Firm-level results using this index are not interpretable as EA governance.")
            else:
                print("  ✓ index is not collinear with size, and is distinct from the placebo.")
            globals()["FALSI_CORR"] = dict(corr_size=round(r_sz, 3), corr_placebo=round(r_pl, 3),
                                           flagged=bool(flags))

    # ---------- (b) SPECIFICATION CURVE ----------
    def curve(xcol, label):
        out = []
        W = D.dropna(subset=[xcol]).sort_values(["cik", "year"]).copy()
        lg = W[["cik", "year", xcol]].copy(); lg["year"] += 1
        W = W.merge(lg.rename(columns={xcol: "x_L"}), on=["cik", "year"], how="left")
        for c in ["size", "lev", "rnd"]:
            if c in W.columns:
                l2 = W[["cik", "year", c]].copy(); l2["year"] += 1
                W = W.merge(l2.rename(columns={c: c + "_L"}), on=["cik", "year"], how="left")
        ctl_all = [c + "_L" for c in ["size", "lev", "rnd"] if c + "_L" in W.columns]
        for dep in [d_ for d_ in ["roa", "q", "sga"] if d_ in W.columns]:
            for est in ["between", "within", "pooled"]:
                for use_ctl in [True, False]:
                    for use_ind in [True, False]:
                        S = W.dropna(subset=["x_L", dep] + (ctl_all if use_ctl else []))
                        if len(S) < 80: continue
                        S = S.copy()
                        if est == "between":
                            S["xx"] = S.groupby("cik")["x_L"].transform("mean")
                        elif est == "within":
                            S["xx"] = S["x_L"] - S.groupby("cik")["x_L"].transform("mean")
                        else:
                            S["xx"] = S["x_L"]
                        X = S[["xx"] + (ctl_all if use_ctl else [])].astype(float)
                        X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y",
                                                         drop_first=True, dtype=float)], axis=1)
                        if use_ind and "industry_raw" in S.columns and S["industry_raw"].nunique() > 1:
                            X = pd.concat([X, pd.get_dummies(S["industry_raw"].astype(str), prefix="i",
                                                            drop_first=True, dtype=float)], axis=1)
                        try:
                            # standardise the OUTCOME: coefficients across roa / q / sga are
                            # otherwise on different scales and the pooled range is meaningless.
                            yv = S[dep].astype(float)
                            yz = (yv - yv.mean()) / (yv.std() or 1)
                            r = sm.OLS(yz.values, sm.add_constant(X)).fit(
                                cov_type="cluster", cov_kwds={"groups": S["cik"].values})
                            out.append(dict(index=label, outcome=dep, estimator=est,
                                            controls=use_ctl, industry=use_ind,
                                            beta=float(r.params["xx"]), p=float(r.pvalues["xx"]), n=len(S)))
                        except Exception:
                            continue
        return out
    allspecs = []
    for xc, lb in [("ci", "CI (Paper 1)"), (idxc, "EAGI (Paper 2)")]:
        if xc in D.columns: allspecs += curve(xc, lb)
    if allspecs:
        SPECCURVE = pd.DataFrame(allspecs)
        print("\nSPECIFICATION CURVE — distribution of the coefficient across reasonable specs:")
        print("  (β are on a STANDARDISED-OUTCOME scale: sd of the dependent variable per unit x)")
        for lb, gsub in SPECCURVE.groupby("index"):
            pos = (gsub.beta > 0)
            print(f"  {lb}: {len(gsub)} specs · median β={gsub.beta.median():+.4f} · "
                  f"range [{gsub.beta.min():+.4f}, {gsub.beta.max():+.4f}] · "
                  f"{(gsub.p < 0.05).mean():.0%} significant · {pos.mean():.0%} positive · "
                  + ("SIGN UNSTABLE" if 0.15 < pos.mean() < 0.85 else "sign stable"))
            for dep, gd in gsub.groupby("outcome"):
                print(f"      {dep:4s}: median β={gd.beta.median():+.4f} "
                      f"[{gd.beta.min():+.4f}, {gd.beta.max():+.4f}] · "
                      f"{(gd.p < 0.05).mean():.0%} sig · {(gd.beta > 0).mean():.0%} positive")
        # PRE-SPECIFIED PRIMARY: adherence/governance is a firm CHARACTERISTIC, so the
        # theoretically primary estimator is BETWEEN-firm with industry and year FE and controls.
        # Declaring it in advance is what separates a specification curve from spec-shopping.
        prim = SPECCURVE[(SPECCURVE.estimator == "between") & (SPECCURVE.controls) &
                         (SPECCURVE.industry) & (SPECCURVE.outcome == "roa")]
        if len(prim):
            print("\n  PRE-SPECIFIED PRIMARY SPEC (between-firm, ind+year FE, controls, ROA):")
            for _, r_ in prim.iterrows():
                print(f"    {r_['index']}: β={r_['beta']:+.4f} p={r_['p']:.3f} n={int(r_['n'])}")
            globals()["PRIMARY_SPEC"] = prim
        nsig = int((SPECCURVE.p < 0.05).sum()); ntot = len(SPECCURVE)
        print(f"\n  multiple-testing context: {nsig}/{ntot} specs significant at 5%; "
              f"~{0.05*ntot:.0f} expected by chance alone if no effect exists.")
        globals()["SPECCURVE"] = SPECCURVE
        try: SPECCURVE.to_csv(f"{OUT_DIR}/specification_curve.csv", index=False)
        except Exception: pass
````

````python
# ============ 9c · FALSIFICATION OF THE AI RESULTS ============
# The AI findings (architecture baseline → AI adoption; AI × baseline → ROA) face exactly the
# critiques that sank the EA effect claims. Six tests, each capable of killing the result:
#   1 placebo moderator   - does a meaningless lexicon "moderate" AI just as well?
#   2 placebo outcome     - does CI predict ANY text index, not specifically AI?
#   3 reverse timing      - does AI predict CI as strongly as CI predicts AI?
#   4 verbosity / size    - is the AI index just longer filings or bigger firms?
#   5 sector split        - does it survive outside technology industries?
#   6 randomisation inf.  - is the interaction extreme against a permuted null?
AIFALS = None
if G("df") is None or "ai_idx" not in df.columns:
    print("SKIPPED — AI index not built.")
else:
    import statsmodels.api as sm
    W = df.sort_values(["cik", "year"]).copy()
    keep = [c for c in ["ci", "placebo", "ai_idx", "size", "lev", "rnd"] if c in W.columns]
    lg = W[["cik", "year"] + keep].copy(); lg["year"] += 1
    W = W.merge(lg.rename(columns={c: c + "_L" for c in keep}), on=["cik", "year"], how="left")
    ctl = [c + "_L" for c in ["size", "lev", "rnd"] if c + "_L" in W.columns]
    def yr_ind(S, X):
        X = pd.concat([X, pd.get_dummies(S["year"].astype(str), prefix="y", drop_first=True, dtype=float)], axis=1)
        if "industry_raw" in S.columns and S["industry_raw"].nunique() > 1:
            X = pd.concat([X, pd.get_dummies(S["industry_raw"].astype(str), prefix="i",
                                             drop_first=True, dtype=float)], axis=1)
        return sm.add_constant(X)
    def fit(S, y, X):
        return sm.OLS(S[y].values, X).fit(cov_type="cluster", cov_kwds={"groups": S["cik"].values})
    rows = []

    # 1 · placebo moderator: AI × placebo should NOT behave like AI × CI
    for mod, lab in [("ci_L", "CI (construct)"), ("placebo_L", "PLACEBO (falsification)")]:
        if mod not in W.columns: continue
        S = W.dropna(subset=["ai_idx", mod, "roa"] + ctl).copy()
        if len(S) < 80: continue
        S["a_c"] = S["ai_idx"] - S["ai_idx"].mean(); S["m_c"] = S[mod] - S[mod].mean()
        S["ixn"] = S["a_c"] * S["m_c"]
        r = fit(S, "roa", yr_ind(S, S[["a_c", "m_c", "ixn"] + ctl].astype(float)))
        rows.append(dict(test="1 moderator", spec=f"AI x {lab}", beta=round(float(r.params["ixn"]), 4),
                         p=round(float(r.pvalues["ixn"]), 3), n=len(S)))

    # 2 · placebo outcome: does CI predict AI specifically, or any text index?
    for dep, lab in [("ai_idx", "AI index"), ("placebo", "PLACEBO index")]:
        if dep not in W.columns or "ci_L" not in W.columns: continue
        S = W.dropna(subset=["ci_L", dep] + ctl)
        if len(S) < 80: continue
        r = fit(S, dep, yr_ind(S, S[["ci_L"] + ctl].astype(float)))
        rows.append(dict(test="2 outcome", spec=f"CI -> {lab}", beta=round(float(r.params["ci_L"]), 4),
                         p=round(float(r.pvalues["ci_L"]), 3), n=len(S)))

    # 3 · reverse timing: symmetry would undercut any directional reading
    if {"ai_idx_L", "ci"}.issubset(W.columns):
        S = W.dropna(subset=["ai_idx_L", "ci"] + ctl)
        if len(S) > 80:
            r = fit(S, "ci", yr_ind(S, S[["ai_idx_L"] + ctl].astype(float)))
            rows.append(dict(test="3 reverse", spec="AI(t-1) -> CI", beta=round(float(r.params["ai_idx_L"]), 4),
                             p=round(float(r.pvalues["ai_idx_L"]), 3), n=len(S)))

    # 4 · verbosity and size
    F = df.groupby("cik").agg({c: "mean" for c in ["ai_idx", "size", "aux_n_tokens", "ci"]
                               if c in df.columns}).dropna()
    if len(F) > 30:
        msg = " · ".join(f"corr(AI,{c})={F['ai_idx'].corr(F[c]):+.3f}"
                         for c in ["size", "aux_n_tokens", "ci"] if c in F.columns)
        rows.append(dict(test="4 confound", spec=msg, beta=np.nan, p=np.nan, n=len(F)))

    # 5 · sector split: survive outside technology?
    if "industry_raw" in W.columns:
        tech = W["industry_raw"].astype(str).str.contains("tech|softw|info|semic", case=False, na=False)
        for lab, mask in [("tech only", tech), ("non-tech only", ~tech)]:
            S = W[mask].dropna(subset=["ai_idx", "ci_L", "roa"] + ctl).copy()
            if len(S) < 80: continue
            S["a_c"] = S["ai_idx"] - S["ai_idx"].mean(); S["m_c"] = S["ci_L"] - S["ci_L"].mean()
            S["ixn"] = S["a_c"] * S["m_c"]
            r = fit(S, "roa", yr_ind(S, S[["a_c", "m_c", "ixn"] + ctl].astype(float)))
            rows.append(dict(test="5 sector", spec=f"AI x CI, {lab}", beta=round(float(r.params["ixn"]), 4),
                             p=round(float(r.pvalues["ixn"]), 3), n=len(S)))

    # 6 · randomisation inference on the interaction
    S = W.dropna(subset=["ai_idx", "ci_L", "roa"] + ctl).copy()
    if len(S) > 80:
        S["a_c"] = S["ai_idx"] - S["ai_idx"].mean(); S["m_c"] = S["ci_L"] - S["ci_L"].mean()
        base = yr_ind(S, S[["a_c", "m_c"] + ctl].astype(float))
        yv = S["roa"].values
        def bx(v):
            XX = pd.concat([pd.Series(v, index=base.index, name="ixn"), base], axis=1).values
            return float((np.linalg.pinv(XX.T @ XX) @ XX.T @ yv)[0])
        obs = bx((S["a_c"] * S["m_c"]).values)
        rs = np.random.default_rng(G("SEED") or 42)
        dr = np.array([bx((S["a_c"].values * rs.permutation(S["m_c"].values))) for _ in range(1000)])
        rows.append(dict(test="6 permutation", spec="AI x CI, 1000 draws", beta=round(obs, 4),
                         p=round(float((np.abs(dr) >= abs(obs)).mean()), 3), n=len(S)))

    if rows:
        AIFALS = pd.DataFrame(rows)
        print("FALSIFICATION OF THE AI RESULTS:")
        print(AIFALS.to_string(index=False))
        pm = AIFALS[(AIFALS.test == "1 moderator") & (AIFALS.spec.str.contains("PLACEBO"))]
        po = AIFALS[(AIFALS.test == "2 outcome") & (AIFALS.spec.str.contains("PLACEBO"))]
        bad = []
        if len(pm) and float(pm.p.iloc[0]) < 0.10: bad.append("a meaningless lexicon moderates AI too")
        if len(po) and float(po.p.iloc[0]) < 0.10: bad.append("CI predicts the placebo index as well")
        if bad:
            print("\n  ⚠ AI RESULT FAILS FALSIFICATION: " + "; ".join(bad))
            print("    Treat as disclosure-volume artefact, not an architecture-AI relationship.")
        else:
            print("\n  ✓ placebo moderator and placebo outcome are both null → the AI result survives")
            print("    falsification. It remains ASSOCIATIONAL: the identification gradient still applies.")
        globals()["AIFALS"] = AIFALS
        try: AIFALS.to_csv(f"{OUT_DIR}/ai_falsification.csv", index=False)
        except Exception: pass
````

````python
# ============ 10 · EXPORT + TWO-PAPER SUMMARY ============
if G("df") is None or G("results") is None:
    print("SKIPPED — nothing to summarize yet. See RUN DIAGNOSTIC below.")
else:
    rows = []
    for y, (r1, r2) in results.items():
        rows.append(dict(outcome=y, model="H1", beta=r1.params["eagi_l"], se=r1.std_errors["eagi_l"],
                         p=r1.pvalues["eagi_l"], n=int(r1.nobs)))
        rows.append(dict(outcome=y, model="H2_ixn", beta=r2.params["ci_x_eagi_l"], se=r2.std_errors["ci_x_eagi_l"],
                         p=r2.pvalues["ci_x_eagi_l"], n=int(r2.nobs)))
    tab = pd.DataFrame(rows)
    tab.to_csv(f"{OUT_DIR}/eagi_main_results.csv", index=False)
    loto.to_csv(f"{OUT_DIR}/eagi_loto.csv", index=False)
    df.to_csv(f"{OUT_DIR}/panel_with_eagi.csv", index=False)
    try:
        _num = df.select_dtypes("number").round(8)
        PANEL_HASH = int(pd.util.hash_pandas_object(_num, index=False).sum() % (10 ** 12))
        df.to_parquet(f"{OUT_DIR}/panel_frozen.parquet", index=False)
        globals()["PANEL_HASH"] = PANEL_HASH
        print(f"frozen panel → panel_frozen.parquet · hash {PANEL_HASH}")
    except Exception as e:
        print("panel freeze skipped:", type(e).__name__)
    try:
        with pd.ExcelWriter(f"{OUT_DIR}/eagi_results.xlsx", engine="openpyxl") as xw:
            tab.to_excel(xw, sheet_name="main_results", index=False)
            loto.to_excel(xw, sheet_name="leave_one_term_out", index=False)
            df.to_excel(xw, sheet_name="panel_with_eagi", index=False)
            if G("VALIDITY_TBL") is not None: VALIDITY_TBL.to_excel(xw, sheet_name="convergent_validity", index=False)
            if G("INSTR_TABLE") is not None: INSTR_TABLE.to_excel(xw, sheet_name="instrument_variants", index=False)
        print("workbook → eagi_results.xlsx")
    except Exception as e:
        print("xlsx export skipped:", type(e).__name__)

    _key = "roa" if "roa" in results else (list(results)[0] if results else None)
    _estn = int(results[_key][0].nobs) if _key else 0
    _base = float(results[_key][0].params["eagi_l"]) if _key else 0.0
    _se0 = float(results[_key][0].std_errors["eagi_l"]) if _key else float("nan")
    _rng = float(loto.beta_roa.max() - loto.beta_roa.min()) if len(loto) else float("nan")
    _alt = G("ALT_NOTE") or []
    P1 = [x.replace("[P1] ", "").replace("[P1]", "") for x in _alt if x.startswith("[P1]")]
    P2 = [x.replace("[P2] ", "").replace("[P2]", "") for x in _alt if x.startswith("[P2]")]

    L = []; A = L.append
    A("=" * 72)
    A("RUN SUMMARY — ONE PIPELINE, TWO PAPERS")
    A(f"generated {pd.Timestamp.now():%Y-%m-%d %H:%M}")
    A(f"shared sample: {panel.shape[0]} firm-years · {panel.cik.nunique()} firms · "
      f"{int(panel.year.min())}–{int(panel.year.max())} · panel {G('PANEL_PATH') or 'DERIVED from CCAOF tables'}")
    A(f"fiscal-year alignment: {G('FY_ALIGN_NOTE') or 'n/a'}")
    A(f"panel hash (reproducibility): {G('PANEL_HASH') or 'n/a'} · frozen copy panel_frozen.parquet")
    A("REPRODUCIBILITY: seed=" + str(G("SEED")) + " · deterministic rule classifier is PRIMARY · "
      "LLM second classifier " + str(G("LLM_MODE") or "not run"))
    A("  LLM labels are cached to llm_techcom_labels.csv and ship with the replication package,")
    A("  so reruns are identical whether or not the API is reachable.")
    if G("KAPPA_RULES_LLM") is not None:
        A(f"  classifier agreement rules vs LLM (no human input): κ={G('KAPPA_RULES_LLM'):.3f}")
    A("")
    A("#" * 72)
    A("DECLARED ANALYSIS HIERARCHY (fixed before estimation; multiplicity is the main threat")
    A("to a programme this size — every result is defensible alone, the SET needs a structure)")
    A("#" * 72)
    A("  PRIMARY   (1 test per paper · 5% threshold, no adjustment needed)")
    A("    P1: CI → ROA, between-firm, industry+year FE, controls")
    A("    P2: EAGI instrument reliability (α, split-half) — a measurement claim, not an effect")
    A("  SECONDARY (pre-specified · Bonferroni within family)")
    A("    P1: within-firm estimate · inverted-U shape · dimension-level · P1-H3 dispersion")
    A("    P2: H1 levels (3 outcomes → p<0.0167) · H2 moderation · H3 variance battery")
    A("  EXPLORATORY (hypothesis-generating; report effect sizes, not significance verdicts)")
    A("    P1-H3b directional · P1-H3c quintiles · P1-H3d fit-to-potential · composite structure")
    A("    compliance channel · AI control and probe · ERP signals · long horizon · pre-shock")
    A("  Anything in the exploratory tier that survives should be RE-TESTED on new data, not")
    A("  promoted within this one.")
    A("")
    A("#" * 72)
    A("PAPER 1 — CIAOF: Core Integrity adherence and firm outcomes")
    A("#" * 72)
    A(f"Construct: {(G('DERIV_HOW') or {}).get('ci', 'n/a')}")
    A("Claim: firms with higher Core Integrity adherence achieve better outcomes;")
    A("originally specified as an inverted-U (an interior optimum in adherence).")
    A("-" * 72)
    for ln in P1: A("  " + ln)
    if not P1: A("  (alternative-design cell did not run — no between/within decomposition available)")
    if G("STRUCT_NOTE"):
        A("Structure of the composite — weakest link · complementarity · importance · necessity:")
        for ln in STRUCT_NOTE: A("  " + ln)
    if G("H3_NOTE"):
        A("P1-H3 — allocation, not level (does SELECTIVE adherence beat UNIFORM adherence?):")
        for ln in H3_NOTE: A("  " + ln)
    A("-" * 72)
    A("Paper 1 verdict (pre-publication — specify BOTH estimators up front, not as a revision):")
    A("  Theorise adherence as a firm CHARACTERISTIC rather than a decision margin. Report the")
    A("  within-firm estimate (clean identification, low power on a sticky construct) AND the")
    A("  between-firm estimate (powered, confounded) as answering two different questions:")
    A("    within  → does BECOMING more disciplined change outcomes?")
    A("    between → do disciplined FIRMS perform differently?")
    A("  The inverted-U is a shape claim and is therefore inherently cross-sectional; report it only")
    A("  if concave AND the vertex lies inside the observed range (checked above).")
    A("  Robustness: ccai_pca alternative composite (CI α is modest — show both).")
    A("  Identification roadmap for the follow-up: firm-level ERP install-base exposure from an")
    A("  EXTERNAL source (vendor-internal customer data would recreate the conflict CIAOF removed).")
    A("")
    A("#" * 72)
    A("PAPER 2 — EAGI: EA governance disclosure, instrument and outcomes")
    A("#" * 72)
    if G("LEX_ALPHA") is not None:
        A(f"Instrument: α={G('LEX_ALPHA'):.3f} · mean inter-item r={G('LEX_MEAN_R'):.3f} · PC1 share={G('LEX_PC1'):.1%}")
        A(f"  independence benchmark for a {len(LEXICON)}-term mean = {1/np.sqrt(len(LEXICON)):.3f}; observed sd = {df.eagi.std():.3f}")
        if G("LEX_DEAD"): A(f"  terms in <5% of filings: {G('LEX_DEAD')}")
    if G("PREV_TBL") is not None:
        A("Term prevalence — the root cause (filing level vs firm-ever level):")
        for ln in PREV_TBL.round(3).to_string(index=False).splitlines()[:8]: A("  " + ln)
    if G("INSTR_TABLE") is not None:
        A("Instrument variants incl. FIRM-LEVEL aggregation (the CIAOF analogue for measurement):")
        for ln in INSTR_TABLE.to_string(index=False).splitlines(): A("  " + ln)
    if G("I106_NOTE"):
        A("Item 106 staggered-deadline design (feasibility-gated):")
        for ln in I106_NOTE: A("  " + ln)
    if G("CUST_NOTE"):
        A("Customisation VOLUME (XBRL capitalised software) — the 2x2's missing variable:")
        for ln in CUST_NOTE: A("  " + ln)
    if G("META_NOTE"):
        A("Filing-metadata instruments (filing lag · NT 10-K · amendments · SIC · filer status):")
        for ln in META_NOTE: A("  " + ln)
    if G("MAND_NOTE"):
        A("Mandatory-disclosure instruments (audited ITGC weakness · Item 106 cyber governance):")
        for ln in MAND_NOTE: A("  " + ln)
    if G("PROX_NOTE"):
        A("Proximity lexicon (the middle layer between exact phrases and raw roots):")
        for ln in PROX_NOTE: A("  " + ln)
    if G("CEIL_NOTE"):
        A("Lexicon ceiling — deterministic bound on any keyword method:")
        for ln in CEIL_NOTE: A("  " + ln)
    if G("SEM_NOTE"):
        A("Measurement of last resort — semantic scoring and alternative corpus:")
        for ln in SEM_NOTE: A("  " + ln)
    A("-" * 72)
    A("H1 — within-firm FE(firm+year), firm-clustered SE · dep ~ EAGI(t−1) + controls")
    for y in [k for k in ["roa", "q", "sga"] if k in results]:
        r1 = results[y][0]
        A(f"  {y.upper():4s} β={r1.params['eagi_l']:+.4f}  se={r1.std_errors['eagi_l']:.4f}  p={r1.pvalues['eagi_l']:.3f}"
          f"  per-1sd Δ={r1.params['eagi_l'] * (G('SD_EAGI_L') or 0):+.4f}"
          + (f"  | DK p={RESULTS_DK[y][0].pvalues['eagi_l']:.3f}" if y in (G("RESULTS_DK") or {}) else ""))
    A("H2 — centred interaction CI(t−1) × EAGI(t−1)")
    for y in [k for k in ["roa", "q", "sga"] if k in results]:
        r2 = results[y][1]
        A(f"  {y.upper():4s} β_ixn={r2.params['ci_x_eagi_l']:+.4f}  se={r2.std_errors['ci_x_eagi_l']:.4f}"
          f"  p={r2.pvalues['ci_x_eagi_l']:.3f}  per-1sd×1sd Δ="
          f"{r2.params['ci_x_eagi_l'] * (G('SD_EAGI_L') or 0) * (G('SD_CI_L') or 0):+.4f}")
    for _y in [k for k in ["roa", "q", "sga"] if k in results and k in (G("RESULTS_DK") or {})]:
        _sc, _sd2 = results[_y][0].std_errors["eagi_l"], RESULTS_DK[_y][0].std_errors["eagi_l"]
        if _sd2 > 0 and max(_sc / _sd2, _sd2 / _sc) > 2:
            A(f"  ⚠ {_y.upper()}: clustered vs DK SEs differ {max(_sc/_sd2, _sd2/_sc):.1f}× — T≈7 makes DK unreliable; use clustered.")
    A("Better-matched designs (sticky construct → between-firm; theory → adaptability):")
    for ln in P2: A("  " + ln)
    if G("AI_NOTE"):
        A("AI positive control + Paper 3 probe:")
        for ln in AI_NOTE: A("  " + ln)
    if G("AIFALS") is not None:
        A("Falsification of the AI results:")
        for ln in AIFALS.to_string(index=False).splitlines(): A("  " + ln)
    if G("VAR_NOTE"):
        A("Variance battery · pre-shock design · 2x2 heterogeneity:")
        for ln in VAR_NOTE: A("  " + ln)
    if G("H2S_NOTE"):
        A("H2 re-tested with a NON-TEXT moderator + formative-construct dimension analysis:")
        for ln in H2S_NOTE: A("  " + ln)
    if G("COMP_NOTE"):
        A("Compliance channel — is the visible vocabulary the meaningful one?")
        for ln in COMP_NOTE: A("  " + ln)
    if G("FALSI") is not None:
        A("Placebo falsification of the firm-level results:")
        for ln in FALSI.to_string(index=False).splitlines(): A("  " + ln)
    if G("FALSI_CORR") is not None:
        A(f"Construct-validity correlations: corr(index, size)={FALSI_CORR['corr_size']:+.3f} · "
          f"corr(index, placebo)={FALSI_CORR['corr_placebo']:+.3f}"
          + ("  ⚠ FLAGGED — firm-level results not interpretable as the construct"
             if FALSI_CORR["flagged"] else "  ✓ distinct from size and placebo"))
    if G("PRIMARY_SPEC") is not None:
        A("PRE-SPECIFIED PRIMARY SPEC (between-firm, industry+year FE, controls, ROA):")
        for _, r_ in PRIMARY_SPEC.iterrows():
            A(f"  {r_['index']}: β={r_['beta']:+.4f} p={r_['p']:.3f} n={int(r_['n'])}")
    if G("SPECCURVE") is not None:
        A("Specification curve (all reasonable specs; fragile results must be shown as a distribution):")
        for lb, gs_ in SPECCURVE.groupby("index"):
            A(f"  {lb}: {len(gs_)} specs · median β={gs_.beta.median():+.4f} · "
              f"range [{gs_.beta.min():+.4f}, {gs_.beta.max():+.4f}] · "
              f"{(gs_.p < 0.05).mean():.0%} significant · {(gs_.beta > 0).mean():.0%} positive")
    A("Robustness (ROA):")
    if len(loto):
        A(f"  leave-one-term-out β ∈ [{loto.beta_roa.min():+.4f}, {loto.beta_roa.max():+.4f}]")
    A(f"  winsorized β={beta_winsor:+.4f} · t−2 β={beta_lag2:+.4f} · placebo β="
      + ("n/a" if beta_placebo is None else f"{beta_placebo:+.4f}"))
    if G("VALIDITY_TBL") is not None:
        A("Convergent validity (within_firm = the variation FE exploits):")
        for ln in VALIDITY_TBL.to_string(index=False).splitlines(): A("  " + ln)
        VALIDITY_TBL.to_csv(f"{OUT_DIR}/eagi_convergent_validity.csv", index=False)
    if _key:
        _mde1 = 2.802 * _se0 * (G("SD_EAGI_L") or 0)
        A(f"Power: smallest detectable per-1sd {_key.upper()} effect = {_mde1:+.4f} (80% power)")
        _rel = G("LEX_ALPHA")
        if _rel is not None and _rel == _rel:
            A(f"  attenuation-corrected at reliability {float(_rel):.3f}: {_mde1/max(float(_rel),0.01):+.4f}")
            if float(_rel) < 0.3:
                A("  ⇒ NULL IS UNINFORMATIVE ABOUT THE ECONOMICS — predicted by the instrument alone.")
    if G("TC_BOUNDS") is not None:
        _lo, _hi, _ns, _sig = TC_BOUNDS
        A(f"TechCom extreme-coding bounds on β(→ROA): [{_lo:+.4f}, {_hi:+.4f}] · ambiguous share "
          f"{(G('TC_REVIEW_SHARE') or 0):.1%}")
        A("  " + ("null at both extremes → ambiguous cases cannot change the conclusion" if _ns
                  else ("robust at both extremes" if _sig
                        else "conclusion flips inside the bounds → coding of ambiguous cases is decisive")))
    if G("TC_SENS") is not None:
        A("TechCom coding scenarios (raw regex · lower bound · upper bound):")
        for ln in TC_SENS.to_string(index=False).splitlines(): A("  " + ln)
    if G("CS_NOTE"):
        A("Callaway-Sant'Anna (hand-rolled, not-yet-treated controls):")
        for ln in CS_NOTE: A("  " + ln)
    A("Event study — board TechCom formation:")
    for ln in (G("es_note") or "not estimated").splitlines(): A("  " + ln)
    if G("LEXVAL_MIX"): A(f"Lexicon context auto-labels: {G('LEXVAL_MIX')} (human coding sheet in OUT_DIR)")
    A("-" * 72)
    A("Paper 2 verdict:")
    A("  Report as a MEASUREMENT paper: a pre-registered disclosure index, a documented null, and")
    A("  quantified evidence on whether 10-K language can carry the construct at all. Structural")
    A("  measures (board technology committee, named CIO/CTO/Chief Architect) are the bridge to")
    A("  a follow-up effects paper; repository telemetry serves as a tech-subsample validation.")
    A("")
    if G("AGREE_TBL") is not None:
        A("Coding agreement (machine vs human gold standard):")
        for ln in AGREE_TBL.to_string(index=False).splitlines(): A("  " + ln)
    A("=" * 72)
    A("SHARED ARTEFACTS")
    A(f"  results: eagi_main_results.csv · eagi_loto.csv · panel_with_eagi.csv · eagi_results.xlsx")
    A(f"  validation: techcom_evidence.csv · techcom_verified.csv · exec_tech_roles.csv · lexicon_validation_sheet.xlsx")
    A(f"  instrument: eagi_instrument_variants.csv   →  all in {OUT_DIR}")
    A("  Tool linkage: TRVA Navigator operationalises Paper 1 (CIAOF dimensions → archetype gating)")
    A("  and consumes Paper 2's governance framing (governed capacity to choose deviations).")
    A("  Empirical results require NO human coding: rules are primary, LLM labels are cached data.")
    A("  Human coding is a QUALITATIVE FOLLOW-UP study (gold standard files are emitted for it).")
    A("  (legacy note) Human tasks reduced to a GOLD STANDARD only: code techcom_goldstandard.csv (~45 rows) and a")
    A("  slice of lexicon_validation_sheet.xlsx, save with a '_coded' suffix, re-run → κ is computed")
    A("  automatically and machine labels are then defensible for the full corpus.")
    A("Inference notes: SEs firm-clustered; H1 tests 3 outcomes → Bonferroni 5% is p<0.0167;")
    A("  H2 main effects centred; capex unavailable → capitalised-software intensity substituted.")
    A("=" * 72)
    globals()["RUN_SUMMARY_TEXT"] = "\n".join(L)
    open(f"{OUT_DIR}/eagi_run_summary.txt", "w").write(RUN_SUMMARY_TEXT)
    print("\n" + RUN_SUMMARY_TEXT)
````

````python
# ============ 10b · PAPER NUMBERS EXPORT — the authoritative reconciliation artefact ============
# Every bracketed figure in either working paper must be reconciled against a single archived
# source. This cell IS that source. It harvests values from the globals the run already produced
# (never recomputing, so it cannot diverge from what was estimated), writes them keyed and
# machine-readable, renders the figures, and emits a checklist ordered to match Appendix D.2.
PAPER_ROWS, FIGS = [], []
if G("df") is None or G("OUT_DIR") is None:
    print("SKIPPED - run did not reach estimation.")
else:
    import datetime, math
    def _rec(paper, section, key, label, value=None, se=None, p=None, n=None, note=""):
        PAPER_ROWS.append(dict(paper=paper, section=section, key=key, label=label,
                               value=value, se=se, p=p, n=n, note=note))
    def _f(x, nd=6):
        try:
            v = float(x)
            return None if (v != v or math.isinf(v)) else round(v, nd)
        except Exception:
            return None
    # ---- parse the note blocks the analysis cells produced, so nothing is retyped by hand ----
    _num = r"([-+]?\d+\.?\d*(?:[eE][-+]?\d+)?)"
    def _harvest(notes, paper, section):
        for ln in (notes or []):
            t = str(ln).strip()
            if not t: continue
            b = re.search(r"(?:beta|\u03b2|b1|b2)\s*[_a-zA-Z]*\s*=\s*" + _num, t)
            pv = re.search(r"\bp\s*=\s*" + _num, t)
            nn = re.search(r"\bn\s*=\s*(\d+)", t)
            if b or pv:
                lab = re.split(r"(?:beta|\u03b2|b1|b2)\s*[_a-zA-Z]*\s*=", t)[0].strip(" :-\u2014")[:110] or t[:110]
                _rec(paper, section, "", lab,
                     _f(b.group(1)) if b else None, None,
                     _f(pv.group(1), 4) if pv else None,
                     int(nn.group(1)) if nn else None)
    # ---------- PAPER 1 ----------
    _rec("P1", "sample", "n_firmyears", "estimation panel firm-years", len(df))
    _rec("P1", "sample", "n_firms", "distinct firms", int(df["cik"].nunique()))
    _rec("P1", "sample", "year_min", "first year", int(df["year"].min()))
    _rec("P1", "sample", "year_max", "last year", int(df["year"].max()))
    _rec("P1", "sample", "panel_hash", "frozen panel hash", G("PANEL_HASH"))
    if "ci" in df.columns:
        _b = _f(df.groupby("cik")["ci"].mean().std()); _w = _f(df.groupby("cik")["ci"].transform(lambda x: x - x.mean()).std())
        _rec("P1", "4.1", "ci_sd_between", "CI between-firm SD", _b)
        _rec("P1", "4.1", "ci_sd_within", "CI within-firm SD", _w)
        if _b and _w:
            _rec("P1", "4.1", "ci_between_share", "between share of CI variance",
                 _f(_b**2 / (_b**2 + _w**2), 4))
    if G("PRIMARY_SPEC") is not None:
        for _, r_ in PRIMARY_SPEC.iterrows():
            _rec("P1" if "CI" in str(r_["index"]) else "P2", "3.4/4.2", "primary_spec",
                 f"PRE-SPECIFIED PRIMARY - {r_['index']}", _f(r_["beta"]), None, _f(r_["p"], 4), int(r_["n"]))
    _harvest(G("H3_NOTE"), "P1", "4.4-4.6")
    _harvest(G("STRUCT_NOTE"), "P1", "4.7")
    _harvest([x for x in (G("ALT_NOTE") or []) if x.startswith("[P1]")], "P1", "4.2/4.8")
    if G("DIMS") is not None:
        for _, r_ in DIMS.iterrows():
            _rec("P1", "4.7", "dim_" + str(r_["dimension"]), "dimension-level between-firm",
                 _f(r_["beta_between"]), None, _f(r_["p"], 4), int(r_["n"]))
    # ---------- PAPER 2 ----------
    for k, lab in [("LEX_ALPHA", "EAGI Cronbach alpha"), ("LEX_MEAN_R", "mean inter-item r"),
                   ("LEX_PC1", "PC1 variance share")]:
        if G(k) is not None: _rec("P2", "4.1", k.lower(), lab, _f(G(k), 4))
    if G("LEX_DEAD") is not None:
        _rec("P2", "4.1", "dead_terms", "terms present in <5% of filings", len(G("LEX_DEAD")),
             note="; ".join(map(str, G("LEX_DEAD")))[:400])
    if "eagi" in df.columns:
        _rec("P2", "4.1", "eagi_sd", "observed SD of the 20-term index", _f(df["eagi"].std(), 4))
        _rec("P2", "4.1", "eagi_indep_benchmark", "1/sqrt(k) independence benchmark",
             _f(1/np.sqrt(len(LEXICON)), 4))
    for src, sec in [("CEIL_NOTE", "4.3"), ("PROX_NOTE", "4.3"), ("SEM_NOTE", "4.3"),
                     ("AI_NOTE", "4.2/5"), ("MAND_NOTE", "4.6"), ("META_NOTE", "4.6"),
                     ("COMP_NOTE", "4.7"), ("I106_NOTE", "4.6"), ("VAR_NOTE", "4.4"),
                     ("H2S_NOTE", "4.4"), ("CS_NOTE", "4.6"), ("CUST_NOTE", "6.4")]:
        _harvest(G(src), "P2", sec)
    _harvest([x for x in (G("ALT_NOTE") or []) if x.startswith("[P2]")], "P2", "4.4")
    if G("results") is not None:
        for y, (r1, r2) in results.items():
            _rec("P2", "4.4", f"H1_{y}", f"H1 within-firm FE - {y.upper()}",
                 _f(r1.params["eagi_l"]), _f(r1.std_errors["eagi_l"]), _f(r1.pvalues["eagi_l"], 4), int(r1.nobs))
            _rec("P2", "4.4", f"H2_{y}", f"H2 centred interaction - {y.upper()}",
                 _f(r2.params["ci_x_eagi_l"]), _f(r2.std_errors["ci_x_eagi_l"]),
                 _f(r2.pvalues["ci_x_eagi_l"], 4), int(r2.nobs))
    for nm, tbl in [("instrument_variants", G("INSTR_TABLE")), ("variance_battery", G("VARTBL")),
                    ("techcom_sensitivity", G("TC_SENS")), ("placebo_falsification", G("FALSI")),
                    ("ai_falsification", G("AIFALS")), ("cs_att", G("CS_TBL")),
                    ("convergent_validity", G("VALIDITY_TBL")), ("lexicon_ceiling", G("CEIL_TBL")),
                    ("term_prevalence", G("PREV_TBL")), ("coding_agreement", G("AGREE_TBL"))]:
        if tbl is not None and len(tbl):
            try: tbl.to_csv(f"{OUT_DIR}/tbl_{nm}.csv", index=False)
            except Exception: pass
    if G("SPECCURVE") is not None:
        for lb, gs_ in SPECCURVE.groupby("index"):
            _rec("P1" if "CI" in str(lb) else "P2", "4.2", "speccurve",
                 f"specification curve - {lb}: {len(gs_)} specs, "
                 f"{(gs_.p<0.05).mean():.0%} significant, {(gs_.beta>0).mean():.0%} positive",
                 _f(gs_.beta.median()), None, None, len(gs_),
                 note=f"range [{gs_.beta.min():+.4f}, {gs_.beta.max():+.4f}]")
    # ---------- write ----------
    PN = pd.DataFrame(PAPER_ROWS)
    stamp = f"{pd.Timestamp.now():%Y-%m-%d %H:%M}"
    try:
        PN.to_csv(f"{OUT_DIR}/paper_numbers.csv", index=False)
        json.dump({r["key"] or f"{r['paper']}_{r['section']}_{i}": r for i, r in enumerate(PAPER_ROWS)},
                  open(f"{OUT_DIR}/paper_numbers.json", "w"), indent=1, default=str)
    except Exception as e:
        print("csv/json export failed:", type(e).__name__)
    L2 = []; A2 = L2.append
    A2(f"# Paper numbers - authoritative export\n")
    A2(f"generated {stamp} - notebook {G('NB_VERSION')} - panel hash {G('PANEL_HASH')}\n")
    A2("Every bracketed figure in either working paper reconciles to a row below. Values are")
    A2("harvested from the run's own outputs; nothing here is recomputed or retyped.\n")
    for pap, title in [("P1", "Paper 1 - CIAOF"), ("P2", "Paper 2 - EAGI")]:
        A2(f"\n## {title}\n")
        sub = PN[PN.paper == pap]
        for sec in sorted(sub.section.unique()):
            A2(f"\n### Section {sec}\n")
            A2("| statistic | value | se | p | n |")
            A2("|---|---|---|---|---|")
            for _, r_ in sub[sub.section == sec].iterrows():
                A2(f"| {str(r_['label'])[:100]} | {r_['value']} | {r_['se'] if r_['se'] is not None else ''} "
                   f"| {r_['p'] if r_['p'] is not None else ''} | {r_['n'] if r_['n'] is not None else ''} |")
    A2("\n\n## Verbatim run blocks (source of truth for anything not parsed above)\n")
    for nm in ["CEIL_NOTE", "PROX_NOTE", "SEM_NOTE", "AI_NOTE", "VAR_NOTE", "H2S_NOTE", "H3_NOTE",
               "STRUCT_NOTE", "COMP_NOTE", "I106_NOTE", "MAND_NOTE", "META_NOTE", "CUST_NOTE",
               "CS_NOTE", "ALT_NOTE"]:
        v = G(nm)
        if v:
            A2(f"\n### {nm}\n```")
            for ln in v: A2(str(ln))
            A2("```")
    if G("RUN_SUMMARY_TEXT"):
        A2("\n\n## Full run summary\n```")
        A2(RUN_SUMMARY_TEXT); A2("```")
    try:
        open(f"{OUT_DIR}/PAPER_NUMBERS.md", "w").write("\n".join(L2))
    except Exception as e:
        print("markdown export failed:", type(e).__name__)
    # ---------- figures ----------
    try:
        import matplotlib
        matplotlib.use("Agg")
        import matplotlib.pyplot as plt
        FD = f"{OUT_DIR}/figures"; os.makedirs(FD, exist_ok=True)
        def _save(fig, name, caption):
            fig.tight_layout(); fig.savefig(f"{FD}/{name}.png", dpi=200); plt.close(fig)
            FIGS.append((name, caption))
        if G("SPECCURVE") is not None:
            for lb, gs_ in SPECCURVE.groupby("index"):
                g2 = gs_.sort_values("beta").reset_index(drop=True)
                fig, ax = plt.subplots(figsize=(7, 3.4))
                cols = ["#c0392b" if pp < .05 else "#9a9a9a" for pp in g2.p]
                ax.bar(range(len(g2)), g2.beta, color=cols)
                ax.axhline(0, color="k", lw=.8)
                ax.set_xlabel("specification (ordered by coefficient)")
                ax.set_ylabel("beta (standardised outcome)")
                ax.set_title(f"Specification curve - {lb}")
                _save(fig, f"fig_speccurve_{re.sub(r'[^A-Za-z0-9]+','_',str(lb))}",
                      "distribution of the coefficient across all defensible specifications; red = p<0.05")
        if G("PREV_TBL") is not None and len(PREV_TBL):
            t = PREV_TBL.sort_values("share_firms_ever").tail(20)
            fig, ax = plt.subplots(figsize=(7, 5))
            ax.barh(t["term"].astype(str), t["share_firms_ever"], color="#1a6b72")
            ax.axvline(.05, color="#c0392b", ls="--", lw=1, label="5% threshold")
            ax.set_xlabel("share of firms using the term at least once"); ax.legend()
            ax.set_title("Lexicon term prevalence")
            _save(fig, "fig_term_prevalence", "term prevalence at firm-ever level; the 5% line is the usability threshold")
        if G("INSTR_TABLE") is not None and len(INSTR_TABLE):
            t = INSTR_TABLE.dropna(subset=["spearman_brown"]).sort_values("spearman_brown")
            fig, ax = plt.subplots(figsize=(7, max(3, .32*len(t))))
            ax.barh(t["variant"].astype(str), t["spearman_brown"], color="#6b6457")
            ax.axvline(.5, color="#c0392b", ls="--", lw=1, label="SB = 0.50")
            ax.set_xlabel("Spearman-Brown"); ax.legend()
            ax.set_title("Instrument reliability across constructions")
            _save(fig, "fig_instrument_variants", "no construction reaches conventional reliability")
        if G("VARTBL") is not None and len(VARTBL):
            fig, ax = plt.subplots(figsize=(7, 3.6))
            lab = VARTBL["risk_measure"].astype(str) + " ~ " + VARTBL["regressor"].astype(str)
            ax.barh(lab, VARTBL["p"], color=["#1a7a4a" if pp < .05 else "#9a9a9a" for pp in VARTBL["p"]])
            ax.axvline(.05, color="#c0392b", ls="--", lw=1)
            ax.set_xlabel("p-value"); ax.set_title("Variance battery")
            _save(fig, "fig_variance_battery", "risk-measure results; significance alone is insufficient - see placebo columns")
        if G("CS_TBL") is not None and len(CS_TBL):
            t = CS_TBL.dropna(subset=["att"])
            fig, ax = plt.subplots(figsize=(6.5, 3.6))
            ax.errorbar(t["event_time"], t["att"],
                        yerr=[1.96*(s if s == s else 0) for s in t["se"].fillna(0)],
                        fmt="o-", color="#1a6b72", capsize=3)
            ax.axhline(0, color="k", lw=.8); ax.axvline(-1, color="#c8a84b", ls="--", lw=1)
            ax.set_xlabel("event time (reference tau = -1)"); ax.set_ylabel("ATT")
            ax.set_title("Callaway-Sant'Anna event-time ATT")
            _save(fig, "fig_cs_event_study", "not-yet-treated controls, multiplier bootstrap; pre-periods should be indistinguishable from zero")
        if G("DIMS") is not None and len(DIMS):
            t = DIMS.sort_values("beta_between")
            fig, ax = plt.subplots(figsize=(7, 3.2))
            ax.barh(t["dimension"].astype(str).str.replace("kpi_", "", regex=False),
                    t["beta_between"], color=["#c0392b" if b < 0 else "#1a7a4a" for b in t["beta_between"]])
            ax.axvline(0, color="k", lw=.8); ax.set_xlabel("between-firm beta")
            ax.set_title("CIAOF dimension-level associations")
            _save(fig, "fig_dimension_level", "the composite can average opposing dimension effects")
        if "ci" in df.columns and "roa" in df.columns:
            b_ = df.groupby("cik").agg(ci=("ci", "mean"), roa=("roa", "mean")).dropna()
            if len(b_) > 20:
                b_["q"] = pd.qcut(b_["ci"], 5, labels=["Q1 lowest", "Q2", "Q3", "Q4", "Q5 highest"],
                                  duplicates="drop")
                m_ = b_.groupby("q", observed=True)["roa"].mean()
                fig, ax = plt.subplots(figsize=(6, 3.2))
                ax.bar(m_.index.astype(str), m_.values, color="#c8a84b")
                ax.set_ylabel("mean ROA (raw)"); ax.set_xlabel("adherence quintile")
                ax.set_title("ROA by adherence quintile (descriptive)")
                _save(fig, "fig_quintiles_raw", "raw means without controls; the paper reports adjusted means")
        print(f"figures written: {len(FIGS)} -> {FD}")
        for nm, cap in FIGS: print(f"    {nm}.png  - {cap}")
        open(f"{FD}/CAPTIONS.md", "w").write(
            "# Figure captions\n\n" + "\n".join(f"- **{n}.png** - {c}" for n, c in FIGS))
    except Exception as e:
        print("figure export skipped:", type(e).__name__, str(e)[:90])
    globals()["PAPER_NUMBERS"] = PN
    print(f"\nPAPER NUMBERS EXPORT -> {OUT_DIR}")
    print(f"  paper_numbers.csv   {len(PN)} keyed statistics")
    print( "  paper_numbers.json  same, machine-readable")
    print( "  PAPER_NUMBERS.md    reconciliation document (use this for Appendix D.2)")
    print(f"  tbl_*.csv           {sum(1 for _ in os.listdir(OUT_DIR) if _.startswith('tbl_'))} result tables")
    print(f"  figures/            {len(FIGS)} figures + CAPTIONS.md")
    print(f"  panel hash          {G('PANEL_HASH')}")
````
